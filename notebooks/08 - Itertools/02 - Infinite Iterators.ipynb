# Itertools => Infinite Iterators

Infinite iterators never stop on their own. Always limit them with `islice`, `takewhile`, or a finite `zip`.

| Function | Purpose | Example |
|---|---|---|
| `count(start, step)` | Count forever | `count(10, 5)` gives 10, 15, 20, ... |
| `cycle(iterable)` | Repeat the items forever | `cycle("AB")` gives A, B, A, B, ... |
| `repeat(x, times)` | Repeat one object | `repeat("x", 3)` gives x, x, x |

---

## `count(start=0, step=1)`

Counts forever: `start`, `start + step`, `start + 2 * step`, and so on.

```python
from itertools import count, islice

list(islice(count(10), 5))       # [10, 11, 12, 13, 14]
list(islice(count(0, 5), 4))     # [0, 5, 10, 15]
list(islice(count(5, -2), 4))    # [5, 3, 1, -1]
```

- `step` can be negative or a float.
- With floats, rounding errors can add up. The documentation suggests `(start + step * i for i in count())` for better accuracy.
- `zip(count(1), items)` numbers items. `enumerate(items, start=1)` does the same and is usually simpler.

## `cycle(iterable)`

Repeats the items of an iterable forever.

```python
from itertools import cycle

list(islice(cycle("ABC"), 7))    # ['A', 'B', 'C', 'A', 'B', 'C', 'A']
```

- It saves a copy of every item, so it may need significant memory for large inputs.
- Cycling an empty iterable produces nothing.
- Typical uses: alternating colors, round-robin assignment, repeating patterns.

## `repeat(object, times=None)`

Repeats the **same** object. Infinite when `times` is not given.

```python
from itertools import repeat

list(repeat("x", 3))                     # ['x', 'x', 'x']
list(map(pow, range(5), repeat(2)))      # [0, 1, 4, 9, 16]
```

Its main use is supplying a **constant stream** to `map()` or `zip()`.

## Ways to Stop an Infinite Iterator

| Tool | Stops when |
|---|---|
| `islice(it, n)` | After `n` items |
| `takewhile(pred, it)` | The predicate first fails |
| `zip(finite, it)` | The finite iterable ends |
| `next(it)` | You take exactly one value |
| `break` in a `for` loop | Your condition is met |

## Common Mistakes

- `list(count())` never finishes.
- `for n in count(): ...` without `break` never finishes.
- `sorted(cycle(...))` and `len(count())` do not work on infinite iterators.
- Forgetting that `count()` remembers its position between `next()` calls.

````python
from itertools import count, cycle, repeat, islice, takewhile

# count(start, step)
print(list(islice(count(10), 5)))                # [10, 11, 12, 13, 14]
print(list(islice(count(0, 5), 4)))              # [0, 5, 10, 15]
print(list(islice(count(5, -2), 4)))             # [5, 3, 1, -1]
print(list(islice(count(0.5, 0.5), 4)))          # [0.5, 1.0, 1.5, 2.0]

# Number items with zip(count(...), items)
print(list(zip(count(1), "abc")))                # [(1, 'a'), (2, 'b'), (3, 'c')]

# Stop an infinite stream with takewhile
print(list(takewhile(lambda n: n < 20, count(0, 7))))   # [0, 7, 14]

# count() remembers its position
counter = count()
print(next(counter), next(counter), next(counter))      # 0 1 2

# cycle
print(list(islice(cycle("ABC"), 7)))             # ['A', 'B', 'C', 'A', 'B', 'C', 'A']

colors = cycle(["red", "green"])
rows = ["row1", "row2", "row3", "row4", "row5"]
print([(row, next(colors)) for row in rows])     # alternating colors

print(list(islice(cycle([]), 5)))                # empty input gives empty output

# repeat
print(list(repeat("x", 3)))                      # ['x', 'x', 'x']
print(list(map(pow, range(5), repeat(2))))       # [0, 1, 4, 9, 16]
print(list(zip("abc", repeat(0))))               # pair every item with a constant
````

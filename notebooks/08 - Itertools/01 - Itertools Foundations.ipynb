# Itertools => Foundations

`itertools` groups its tools in three families. All of them return iterators.

| Family | Tools |
|---|---|
| Infinite iterators | `count`, `cycle`, `repeat` |
| Terminating iterators | `accumulate`, `batched`, `chain`, `chain.from_iterable`, `compress`, `dropwhile`, `filterfalse`, `groupby`, `islice`, `pairwise`, `starmap`, `takewhile`, `tee`, `zip_longest` |
| Combinatoric iterators | `product`, `permutations`, `combinations`, `combinations_with_replacement` |

---

## What Is `itertools`?

`itertools` is a standard-library module of fast, memory-efficient **iterator building blocks**.

The functions are small on purpose. You combine them like Lego pieces, which the documentation calls an **iterator algebra**.

```python
import itertools
```

## Why Use It?

- **Lazy**: values are produced one at a time, only when requested.
- **Memory-efficient**: no big intermediate lists.
- **Composable**: the output of one tool is the input of the next.
- **Short and fast**: the tools are implemented in C.

## Everything Returns an Iterator

Calling an itertools function does not compute the result. It returns an **iterator**.

```python
chained = itertools.chain([1, 2], [3, 4])
print(chained)          # <itertools.chain object ...>
print(list(chained))    # [1, 2, 3, 4]
print(list(chained))    # []  (an iterator is single use)
```

## Import Styles

```python
import itertools                      # itertools.chain(...)
from itertools import chain, islice   # chain(...), islice(...)
import itertools as it                # it.chain(...)
```

Avoid `from itertools import *`. It hides where names come from.

## The Pipeline Mindset

Chain small lazy steps instead of building lists:

```python
from itertools import count, islice

odd_squares = (n * n for n in count() if n % 2 == 1)
list(islice(odd_squares, 5))       # [1, 9, 25, 49, 81]
```

`count()` is infinite, but `islice` stops after five results, so the pipeline never runs away.

## Five Golden Rules

1. **Infinite iterators must be limited** with `islice`, `takewhile`, or a finite `zip`.
2. **Iterators are single use.** Once consumed, they are empty.
3. **`groupby` groups consecutive items.** Sort the data first.
4. **Nothing runs until you consume it.** Use `list()`, `for`, or `next()`.
5. **`tee` stores data in memory.** Use it carefully.

````python
import itertools
from itertools import chain, count, islice

# Everything returns an iterator
chained = chain([1, 2], [3, 4])
print(type(chained).__name__)
print(list(chained))                 # [1, 2, 3, 4]
print(list(chained))                 # []: an iterator is single use

# Nothing runs until it is consumed
it = chain("ab", "cd")
print(next(it), next(it))            # a b
print(list(it))                      # ['c', 'd']

# Public tools in this Python version
print([name for name in dir(itertools) if not name.startswith("_")])

# Pipeline: infinite source + lazy filter + limit
odd_squares = (n * n for n in count() if n % 2 == 1)
print(list(islice(odd_squares, 5)))  # [1, 9, 25, 49, 81]

# The same idea with functions
def is_odd(n):
    return n % 2 == 1

def square(n):
    return n * n

print(sum(map(square, islice(filter(is_odd, count()), 5))))   # 165

# Import styles
import itertools as it_tools
print(list(it_tools.islice("ABCDEFG", 3)))
````

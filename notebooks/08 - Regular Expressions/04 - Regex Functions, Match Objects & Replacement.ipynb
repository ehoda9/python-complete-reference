# 08.4 - Regex APIs => Search, Match, Find, Split & Replace

## Search Functions

### `re.search()`

Searches anywhere and returns the first `Match`, or `None`.

### `re.match()`

Checks only at the beginning of the string.

### `re.fullmatch()`

Requires the entire string to match.

## Find Functions

### `re.findall()`

Returns all matches.

Its result depends on capturing groups:

- No capturing groups => list of matched strings
- One capturing group => list of captured strings
- Multiple capturing groups => list of tuples

### `re.finditer()`

Returns an iterator of `Match` objects.

Use it when you need:

- Match positions
- Groups
- More detailed match information
- Memory-friendly iteration over many results

## Split

```python
re.split(pattern, string, maxsplit=0)
```

With capturing groups, captured delimiters are included in the result.

## Replace

```python
re.sub(pattern, repl, string, count=0)
```

`repl` can be:

- A replacement string
- A function that receives a `Match` object

## `re.subn()`

Returns:

```text
(new_string, replacement_count)
```

## `count`

For `find`/`sub` related operations where supported, `count`/`maxsplit` can limit work.

## Compiled Patterns

```python
pattern = re.compile(r"\d+")
```

Then:

```python
pattern.search(text)
pattern.findall(text)
pattern.finditer(text)
pattern.sub(...)
```

Compiled patterns are useful when the same regex is reused.

## Pattern Object

A compiled pattern is a `re.Pattern`.

Useful attributes:

- `.pattern`
- `.flags`
- `.groups`
- `.groupindex`

## Match Object

A successful match is a `re.Match`.

## Source

https://docs.python.org/3/library/re.html

````python
import re

text = "ID: 123, ID: 456, ID: 789"

# findall: return all matched strings
print(re.findall(r"\d+", text))

# finditer: return an iterator of Match objects
for match in re.finditer(r"\d+", text):
    print(match.group(), match.span())

# split: split by a regex pattern
print(re.split(r",\s*", text))

# sub: replace matches
print(re.sub(r"\d+", "XXX", text))

# sub with a function: replacement depends on the Match
def double(match):
    return str(int(match.group()) * 2)

print(re.sub(r"\d+", double, text))

# compile: create a reusable regex Pattern
pattern = re.compile(r"\d+")

print(pattern.findall(text))
print(pattern.pattern)
print(pattern.groups)
````

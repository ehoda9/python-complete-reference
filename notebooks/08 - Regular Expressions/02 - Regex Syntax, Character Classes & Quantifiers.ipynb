# 08.2 - Regex Syntax => Character Classes & Quantifiers

## Character Classes

| Pattern | Meaning |
|---|---|
| `.` | Any character except newline by default |
| `\d` | Unicode decimal digit |
| `\D` | Not a Unicode decimal digit |
| `\w` | Unicode word character |
| `\W` | Not a Unicode word character |
| `\s` | Unicode whitespace |
| `\S` | Not whitespace |
| `[abc]` | `a`, `b`, or `c` |
| `[^abc]` | Any character except `a`, `b`, `c` |
| `[a-z]` | Lowercase range |
| `[0-9]` | Digit range |

## Quantifiers

| Pattern | Meaning |
|---|---|
| `*` | 0 or more |
| `+` | 1 or more |
| `?` | 0 or 1 |
| `{m}` | Exactly `m` |
| `{m,n}` | Between `m` and `n` |
| `{m,}` | At least `m` |

Quantifiers are normally **greedy**.

Add `?` after a quantifier for the lazy form:

```text
*?   +?   ??   {m,n}?
```

## Anchors and Boundaries

| Pattern | Meaning |
|---|---|
| `^` | Start of string / line depending on flags |
| `$` | End of string / line depending on flags |
| `\A` | Start of string |
| `\Z` | End of string |
| `\b` | Word boundary |
| `\B` | Not a word boundary |

## Important Unicode Note

In Python's default string regex mode, `\d`, `\w`, and `\s` follow Unicode-aware behavior.

Use `re.ASCII` when ASCII-only semantics are required.

## Escaping Special Characters

Characters such as these have special regex meanings:

```text
. ^ $ * + ? { } [ ] \ | ( )
```

Use a backslash when you need their literal meaning:

```python
r"\."
r"\?"
r"\["
```

## `re.escape()`

Useful when user-provided text should be matched literally instead of interpreted as regex syntax.

## Source

https://docs.python.org/3/library/re.html

````python
import re

text = "Order 123, Order 456, Order 789."

# re.findall(pattern, string): return all matching strings
numbers = re.findall(r"\d+", text)

print(numbers)

# Character class
print(re.findall(r"[A-Z]", "Python Regex"))

# Quantifier
print(re.findall(r"\w{4}", "Python code test"))

# Boundary
print(re.findall(r"\bcat\b", "cat scatter category cat"))

# Literal user text
user_text = "a+b"
safe_pattern = re.escape(user_text)

print(re.fullmatch(safe_pattern, "a+b"))
````

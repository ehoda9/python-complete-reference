# 11.1 - Comprehensions => Foundations

## What Is a Comprehension?

A **comprehension** is a compact expression that builds a new collection from an existing iterable, with an optional filter.

You describe *what the result should contain* instead of writing the loop that builds it.

```python
squares = [n * n for n in range(5)]   # [0, 1, 4, 9, 16]
```

## Anatomy

```python
[expression for item in iterable if condition]
```

| Part | Role |
|---|---|
| `expression` | Produces each element of the result |
| `for item in iterable` | The source: gives items one at a time |
| `if condition` | Optional filter: keeps only the items where the condition is true |

The `for` and `if` clauses read like nested blocks written from top to bottom. The `expression` runs at the innermost level.

## Loop vs Comprehension

```python
# Loop
result = []
for n in range(10):
    if n % 2 == 0:
        result.append(n * n)

# Comprehension (same result)
result = [n * n for n in range(10) if n % 2 == 0]
```

## The Four Kinds

| Kind | Syntax | Result | Notes |
|---|---|---|---|
| List comprehension | `[x for x in it]` | `list` | Ordered, keeps duplicates |
| Set comprehension | `{x for x in it}` | `set` | Unordered, unique items |
| Dict comprehension | `{k: v for k, v in it}` | `dict` | Unique keys, the last value wins |
| Generator expression | `(x for x in it)` | generator | Lazy, single use |

There is **no tuple comprehension**. `(x for x in it)` is a generator expression.

Use `tuple(x for x in it)` when you need a tuple.

## Any Iterable Works

The source can be any iterable:

- Strings, lists, tuples, sets, ranges
- Dictionaries (`.items()`, `.keys()`, `.values()`)
- Files, generators
- `zip()` and `enumerate()` results

## Transform, Filter, or Both

| Goal | Example |
|---|---|
| Transform every item | `[w.upper() for w in words]` |
| Keep some items unchanged | `[n for n in nums if n > 0]` |
| Both | `[w.upper() for w in words if len(w) > 3]` |

## Unpacking in the Loop Target

The loop target can unpack values, exactly like a `for` statement:

```python
pairs = [(1, "a"), (2, "b")]
[f"{num}{letter}" for num, letter in pairs]   # ['1a', '2b']
```

## Comprehension vs `map()` and `filter()`

```python
list(map(str.upper, words))             # map
[w.upper() for w in words]              # comprehension

list(filter(lambda n: n > 0, nums))     # filter
[n for n in nums if n > 0]              # comprehension
```

Comprehensions usually read better because they need no `lambda`.

`map()` with an existing function such as `str.upper` is also a good choice.

## Key Rules

- A comprehension **builds a new object**. It never changes the source.
- The loop variable is local to the comprehension and does not leak into the surrounding code.
- Use a comprehension to **build a value**, not for side effects such as `print()`.
- If the expression is hard to read on one line, use a regular loop.

````python
words = ["python", "is", "fun", "and", "powerful"]
nums = [-2, -1, 0, 1, 2, 3]

# Loop vs comprehension (same result)
loop_result = []
for n in range(10):
    if n % 2 == 0:
        loop_result.append(n * n)

comp_result = [n * n for n in range(10) if n % 2 == 0]
print(loop_result == comp_result, comp_result)

# Transform, filter, or both
print([w.upper() for w in words])                 # transform
print([n for n in nums if n > 0])                 # filter
print([w.upper() for w in words if len(w) > 3])   # both

# The four kinds
print([c for c in "hello"])                       # list keeps duplicates
print({c for c in "hello"} == set("helo"))        # set keeps unique items only
print({w: len(w) for w in words})                 # dict
gen = (n * n for n in range(5))                   # generator: lazy
print(type(gen).__name__, list(gen))

# There is no tuple comprehension: wrap a generator expression
print(tuple(n * n for n in range(5)))

# Any iterable works
print([ch * 2 for ch in "abc"])                               # string
print([key for key in {"x": 1, "y": 2}])                      # dict keys
print([f"{i}:{w}" for i, w in enumerate(words, start=1)])     # enumerate
print([a + b for a, b in zip([1, 2, 3], [10, 20, 30])])       # zip

# Unpacking in the loop target
pairs = [(1, "a"), (2, "b")]
print([f"{num}{letter}" for num, letter in pairs])

# map / filter equivalents
print(list(map(str.upper, words)) == [w.upper() for w in words])
print(list(filter(lambda n: n > 0, nums)) == [n for n in nums if n > 0])

# The source is never modified
source = [1, 2, 3]
doubled = [n * 2 for n in source]
print(source, doubled)
````

# 10.1 - Type Hinting => Foundations

## What Is Type Hinting?

**Type hints** are annotations that describe the expected types of values in Python code.

They improve:

- Readability
- Editor support
- Static analysis
- Refactoring
- API documentation
- Early detection of many type-related mistakes

## Important

Python type hints generally do **not** enforce types at runtime by themselves.

```python
def add(a: int, b: int) -> int:
    return a + b
```

A type checker can analyze this annotation, but Python normally does not automatically reject every incorrect call at runtime.

## Variable Annotations

```python
name: str = "Ahmed"
age: int = 25
```

A bare annotation is also possible:

```python
age: int
```

## Function Annotations

```python
def greet(name: str) -> str:
    return f"Hello, {name}"
```

The syntax is:

```text
parameter: Type
return: Type
```

## `None`

Use:

```python
def log(message: str) -> None:
    ...
```

to indicate that the function does not return a useful value.

## Type Hints vs `isinstance()`

| Type hints | `isinstance()` |
|---|---|
| Static/documentation-oriented | Runtime check |
| Used by tools | Executed by Python |
| Does not enforce by itself | Actually checks a value |

## Built-in Generic Syntax

Modern Python can use built-in collections directly:

```python
list[str]
dict[str, int]
tuple[int, str]
set[int]
```

This syntax is preferred in modern Python when compatible with your supported versions.

## Type Checker

Type hints are interpreted by external/static analysis tools such as:

- mypy
- pyright
- IDE language servers

````python
# Variable annotations
name: str = "Ahmed"
age: int = 25

# Function annotations
def greet(name: str) -> str:
    return f"Hello, {name}"

def log(message: str) -> None:
    print(message)

print(greet(name))
log("Type hints improve readability.")

# Type hints do not automatically enforce runtime types.
print(greet(123))  # Python itself does not reject this call just because of the annotation.
````

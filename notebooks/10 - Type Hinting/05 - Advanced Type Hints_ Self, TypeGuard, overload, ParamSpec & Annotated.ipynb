# 10.5 - Advanced Type Hinting

## `Self`

`Self` represents the current enclosing class.

Useful when a method returns `self` and subclasses should preserve the more specific return type.

```python
class Builder:
    def set_name(self, name: str) -> Self:
        self.name = name
        return self
```

It is especially useful for fluent APIs and alternative constructors.

## `TypeGuard`

A function returning `TypeGuard[T]` tells a type checker:

> If this function returns `True`, treat the value as type `T`.

Useful for custom type-narrowing helpers.

## `TypeIs`

Modern typing also provides `TypeIs`, which can express predicates that narrow to an intersection-like compatible type in both branches.

Use it when the predicate truly establishes that relationship.

## `overload`

`@overload` describes multiple call signatures for a function.

Important:

- Overloaded signatures are for static type checkers.
- Only the final implementation is executed at runtime.

## `ParamSpec`

`ParamSpec` captures a callable's parameter specification.

Useful when writing decorators that preserve the original function's parameter types.

```python
P = ParamSpec("P")

def decorator(func: Callable[P, T]) -> Callable[P, T]:
    ...
```

## `Concatenate`

`Concatenate` can describe decorators that add parameters to a callable's signature.

## `Annotated`

`Annotated[T, metadata]` attaches metadata to a type annotation.

The metadata can be interpreted by frameworks/tools.

It does not automatically enforce validation.

## `Any`

`Any` disables much of the static type checking for that value.

Use it carefully.

`Any` is not the same as `object`.

## `object` vs `Any`

- `object` means the value can be any object, but type-safe code cannot assume specific operations without narrowing.
- `Any` allows essentially unchecked operations by type checkers.

## `Never`

`Never` represents a type with no values.

Useful for functions that never return normally:

```python
def fail() -> Never:
    raise RuntimeError
```

## `NoReturn`

`NoReturn` has the same type-system meaning as `Never`; `Never` is the modern spelling.

````python
from functools import wraps
from typing import Annotated, Callable, Never, ParamSpec, Self, TypeGuard, TypeIs, TypeVar, overload

T = TypeVar("T")
P = ParamSpec("P")


class Builder:
    def __init__(self):
        self.name = ""

    def set_name(self, name: str) -> Self:
        self.name = name
        return self


def is_int(value: object) -> TypeGuard[int]:
    return isinstance(value, int)


def is_string(value: object) -> TypeIs[str]:
    return isinstance(value, str)


@overload
def stringify(value: int) -> str: ...

@overload
def stringify(value: str) -> str: ...

def stringify(value: int | str) -> str:
    return str(value)


def trace(func: Callable[P, T]) -> Callable[P, T]:
    @wraps(func)
    def wrapper(*args: P.args, **kwargs: P.kwargs) -> T:
        return func(*args, **kwargs)
    return wrapper


def fail(message: str) -> Never:
    raise RuntimeError(message)


Score = Annotated[int, "0..100"]


builder = Builder().set_name("Python")

value: object = 10
if is_int(value):
    print(value + 1)

print(stringify(100))
print(stringify("Python"))
print(builder.name)
````

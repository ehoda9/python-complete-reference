# Abstract Base Classes (ABCs)

An abstract base class defines the interface that subclasses must provide.

| Tool | Purpose |
|---|---|
| `ABC` | Base class for abstract classes (`from abc import ABC`) |
| `@abstractmethod` | Marks a method that subclasses must implement |
| `ABCMeta` | The metaclass behind `ABC` |
| Abstract properties | `@property` combined with `@abstractmethod` |
| Abstract class and static methods | `@classmethod` or `@staticmethod` combined with `@abstractmethod` |
| Implementation in an abstract method | Subclasses can still call it with `super()` |
| ABC vs duck typing | An explicit, enforced contract vs an informal one |

---

## Why Abstract Classes?

An abstract base class defines an expected interface for subclasses.

Use ABCs when you want to state:

> "Any concrete subclass must provide these operations."

Python's `abc` module provides this infrastructure.

## `ABC`

The easiest modern form is:

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    ...
```

## `@abstractmethod`

```python
@abstractmethod
def area(self):
    ...
```

A concrete subclass must implement abstract methods before it can be instantiated.

## `ABCMeta`

`ABC` uses `ABCMeta` as its metaclass.

You may define an ABC directly with:

```python
class Shape(metaclass=ABCMeta):
    ...
```

but inheriting from `ABC` is usually simpler.

## Abstract Properties and Class/Static Methods

Abstract behavior can also be combined with:

- `@property`
- `@classmethod`
- `@staticmethod`

For these combinations, `@abstractmethod` should be the innermost decorator.

## Abstract Methods Can Have an Implementation

An abstract method can contain code, and subclasses can call it with `super()`.

Being abstract means the subclass must override it before instantiation; it does not necessarily mean the method body must be empty.

## ABC vs Duck Typing

- **ABC** => explicit interface contract enforced at instantiation time.
- **Duck typing** => behavior-based, usually without requiring inheritance.

Choose based on the design and API you need.

## Source

- `abc` module: https://docs.python.org/3/library/abc.html

````python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def area(self):
        # Interface requirement
        pass


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


shape = Rectangle(4, 5)

print(shape.area())
print(isinstance(shape, Shape))

# Shape() would raise TypeError because area() is abstract.
````

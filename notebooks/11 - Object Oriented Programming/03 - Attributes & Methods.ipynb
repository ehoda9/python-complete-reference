# Attributes & Methods

Attributes hold data. Methods hold behavior.

| Topic | Point |
|---|---|
| Instance attributes | Belong to a specific object |
| Class attributes | Defined in the class body and shared by all instances |
| Attribute lookup | Instance first, then the class, then the base classes (MRO) |
| Shadowing | An instance attribute can hide a class attribute with the same name |
| Instance methods | Receive the instance as `self` |
| Method chaining | A method that returns `self` |
| Attribute tools | `getattr`, `setattr`, `hasattr`, `delattr`, `vars`, `dir` |
| Mutable class attributes | Avoid them when each instance needs its own state |

---

## Instance Attributes

Instance attributes belong to a specific object.

Usually created inside `__init__`:

```python
self.name = name
self.age = age
```

Two instances can have different values.

## Class Attributes

Class attributes are defined in the class body:

```python
class User:
    users_count = 0
```

They belong to the class namespace and can be accessed through the class and, when not shadowed, through instances.

## Attribute Lookup

For an expression such as:

```python
obj.name
```

Python performs attribute lookup through the object/class machinery. A simplified mental model:

```text
instance data
   =>
class data
   =>
base classes according to MRO
```

Descriptors can change the exact lookup behavior.

## Shadowing

An instance attribute can shadow a class attribute with the same name.

```python
class User:
    role = "user"

u = User()
u.role = "admin"
```

Now `u.role` comes from the instance, while `User.role` remains `"user"`.

## Instance Methods

Normal methods receive the instance as `self`.

They can:

- Read instance attributes
- Modify instance attributes
- Call other instance methods
- Access class attributes
- Call class methods / static methods

## Method Chaining

A method can return `self` to support chaining:

```python
obj.set_name(...).set_age(...)
```

Use this only when it improves readability.

## Attribute Tools

| Tool | Purpose |
|---|---|
| `getattr(obj, name)` | Get an attribute dynamically |
| `setattr(obj, name, value)` | Set dynamically |
| `hasattr(obj, name)` | Check whether lookup succeeds |
| `delattr(obj, name)` | Delete dynamically |
| `vars(obj)` | Get `obj.__dict__` when available |
| `dir(obj)` | Inspect accessible names |

## Mutable Class Attributes

Avoid class-level mutable state when each instance should have its own data.

Bad:

```python
class User:
    tags = []
```

All instances can share the same list.

Prefer:

```python
class User:
    def __init__(self):
        self.tags = []
```

## Source

- Python Classes: https://docs.python.org/3/tutorial/classes.html

````python
class User:
    users_count = 0

    def __init__(self, name):
        self.name = name
        self.tags = []
        User.users_count += 1

    def add_tag(self, tag):
        self.tags.append(tag)
        return self

    def describe(self):
        return f"{self.name}: {self.tags}"


user = User("Ahmed")
user.add_tag("python").add_tag("oop")

print(user.describe())
print(User.users_count)
print(getattr(user, "name"))
print(hasattr(user, "email"))
print(vars(user))
````

# Class Methods, Static Methods & Alternative Constructors

## Three Common Method Types

| Method | First Parameter | Works With |
|---|---|---|
| Instance method | `self` | A particular instance |
| Class method | `cls` | The class |
| Static method | No automatic `self`/`cls` | No instance/class state |

## `@classmethod`

- Receives the class as `cls`.
- Can be called through the class or an instance.
- Useful for operations involving class-level state.
- Commonly used for **alternative constructors**.

Example idea:

```python
User.from_string("Ahmed,20")
```

## `@staticmethod`

- Receives no automatic instance/class argument.
- Useful for a function that is logically related to a class but does not need object or class state.

## Alternative Constructors

A class method can create instances using another input format:

```python
@classmethod
def from_string(cls, text):
    ...
    return cls(...)
```

Using `cls` rather than the class name keeps inheritance-friendly behavior.

## Decorators Commonly Used With Classes

Important OOP-related decorators:

- `@property`
- `@classmethod`
- `@staticmethod`
- `@abstractmethod`
- `@dataclass`
- `@cached_property`
- `@functools.total_ordering`
- `@functools.singledispatchmethod`

## Decorator Order Matters

When combining decorators, order can affect the resulting descriptor/behavior.

For abstract methods combined with `classmethod`, `staticmethod`, or `property`, `@abstractmethod` should be the innermost decorator.

- `abc`: https://docs.python.org/3/library/abc.html
- `functools`: https://docs.python.org/3/library/functools.html

````python
class User:
    users_count = 0

    def __init__(self, name, age):
        self.name = name
        self.age = age
        User.users_count += 1

    @classmethod
    def from_string(cls, text):
        # cls: the class receiving the call
        name, age = text.split(",")
        return cls(name, int(age))

    @classmethod
    def show_count(cls):
        return cls.users_count

    @staticmethod
    def is_adult(age):
        # No self or cls is needed
        return age >= 18


user = User.from_string("Ahmed,20")

print(user.name, user.age)
print(User.show_count())
print(User.is_adult(user.age))
print(user.is_adult(15))
````

# Object Oriented Programming => Foundations

The vocabulary of object oriented programming.

| Term | Meaning |
|---|---|
| Class | Blueprint / type used to create objects |
| Instance | Object created from a class |
| Attribute | Data associated with an object or class |
| Method | Function defined inside a class |
| State | Current data of an object |
| Behavior | Actions an object can perform |
| Instantiation | Creating an instance from a class |
| Encapsulation | Bundling data and behavior, with controlled access |
| Inheritance | Creating a class from another class |
| Polymorphism | Same interface/operation, different implementations |
| Abstraction | Exposing essential behavior while hiding implementation details |
| Composition | Building objects from other objects |

---

## What Is OOP?

- **OOP (Object Oriented Programming)** is a programming paradigm for organizing data and behavior into objects.
- A **class** is a blueprint that defines the structure and behavior of a type.
- An **object / instance** is a concrete object created from a class.
- **Attributes** represent data/state.
- **Methods** represent behavior.
- Python is a **multi-paradigm** language: procedural, object-oriented, functional, and more.
- Python's object model is dynamic: classes are objects too, and classes can be modified at runtime.

## Core Mental Model

```text
Class
  =>
Blueprint / Type
  =>
Instances (Objects)
  =>
State + Behavior
```

Example:

```text
Car
  Attributes => model, color, price
  Methods    => start(), stop(), accelerate()
```

## Everything Is An Object

Almost every value you use in Python is an object:

```text
10       => int object
"hello"  => str object
[1, 2]   => list object
function => function object
Class    => class/type object
```

Use `type()` and `__class__` to inspect an object's type.

## OOP Is Not Only About Inheritance

A strong Python OOP design also uses:

- Encapsulation
- Composition
- Properties
- Class and static methods
- Polymorphism / duck typing
- Abstract base classes
- Special (dunder) methods
- Dataclasses
- Enums
- Descriptors
- Multiple inheritance and MRO when appropriate

## Python Naming

Class names normally use **PascalCase**:

```text
User
BankAccount
OrderItem
```

Methods and instance attributes normally use **snake_case**:

```text
get_balance()
account_number
```

````python
# A simple class groups data and behavior together.
class Car:
    def __init__(self, model, color):
        self.model = model
        self.color = color

    def describe(self):
        return f"{self.color} {self.model}"


car = Car("Sedan", "Black")

print(car.describe())
print(type(car))
print(car.__class__)
````

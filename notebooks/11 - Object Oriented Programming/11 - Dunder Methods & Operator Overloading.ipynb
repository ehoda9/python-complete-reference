# Special (Dunder) Methods & Operator Overloading

## What Are Dunder Methods?

Dunder = **double underscore**.

Examples:

```text
__init__
__str__
__repr__
__len__
__eq__
```

These methods are special protocol hooks used by Python syntax and built-ins.

They are not merely ordinary methods you call manually.

## Common Representation Methods

| Method | Used By |
|---|---|
| `__str__` | `str(obj)`, `print(obj)` |
| `__repr__` | `repr(obj)`, interactive/debug representation |
| `__format__` | `format(obj, spec)` |
| `__bytes__` | `bytes(obj)` |

A good `__repr__` should generally be useful for debugging.

## Object Lifecycle

| Method | Role |
|---|---|
| `__new__` | Create/return instance |
| `__init__` | Initialize instance |
| `__del__` | Finalization hook; use with care |

`__del__` is not a substitute for deterministic resource management. Prefer context managers for resources.

## Size / Truth / Containers

| Method | Protocol |
|---|---|
| `__len__` | `len(obj)` |
| `__bool__` | `bool(obj)` / truth testing |
| `__iter__` | `iter(obj)` / iteration |
| `__next__` | `next(obj)` |
| `__contains__` | `item in obj` |
| `__getitem__` | `obj[key]` |
| `__setitem__` | `obj[key] = value` |
| `__delitem__` | `del obj[key]` |
| `__call__` | `obj(...)` |

## Comparison Methods

```text
__eq__  ==
__ne__  !=
__lt__  <
__le__  <=
__gt__  >
__ge__  >=
```

Special comparison methods may return `NotImplemented` when they do not support the other operand.

## Arithmetic Operator Overloading

Common methods:

```text
__add__      +
__sub__      -
__mul__      *
__truediv__  /
__floordiv__ //
__mod__      %
__pow__      **
```

Reflected forms include:

```text
__radd__
__rsub__
__rmul__
...
```

In-place forms include:

```text
__iadd__
__isub__
__imul__
...
```

## Hashing

`__hash__` is used by `hash()` and hash-based collections.

Important rule:

> Objects that compare equal must have the same hash.

If a class overrides `__eq__` without explicitly restoring compatible hashing, Python normally makes instances unhashable.

Do not make mutable, equality-based objects hashable in a way that allows the hash to change while stored in a set/dict.

## Special Method Lookup

For implicit operations such as `len(obj)` and arithmetic, Python looks up special methods on the object's type, not by simply checking instance attributes.

## Source

- Python Data Model: https://docs.python.org/3/reference/datamodel.html

````python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __str__(self):
        return f"${self.amount:.2f}"

    def __repr__(self):
        return f"Money({self.amount!r})"

    def __add__(self, other):
        if not isinstance(other, Money):
            return NotImplemented
        return Money(self.amount + other.amount)

    def __eq__(self, other):
        if not isinstance(other, Money):
            return NotImplemented
        return self.amount == other.amount


price1 = Money(10)
price2 = Money(15)
total = price1 + price2

print(price1)
print(repr(price1))
print(total)
print(price1 == Money(10))
````

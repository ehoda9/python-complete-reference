# Inheritance & `super()`

Inheritance lets a class reuse and extend another class.

| Topic | Syntax | Meaning |
|---|---|---|
| Inheritance | `class Child(Parent):` | The child gets the parent's attributes and methods |
| Method overriding | same method name in the child | The child's version replaces the parent's |
| `super()` | `super().method()` | Call the next implementation in the method resolution order |
| `isinstance()` | `isinstance(obj, Class)` | Is the object an instance of the class or a subclass? |
| `issubclass()` | `issubclass(Sub, Base)` | Is one class a subclass of another? |
| Inheritance vs composition | is-a vs has-a | Choose the relationship that fits |

---

## Inheritance

Inheritance allows a class to derive from another class.

```python
class Child(Parent):
    ...
```

- Parent is also called **base class** or **superclass**.
- Child is also called **derived class** or **subclass**.

A subclass can:

- Reuse inherited behavior
- Add new behavior
- Override inherited methods
- Extend initialization

## Method Overriding

A subclass can define a method with the same name as a parent method.

The subclass implementation is then selected for normal method lookup.

## `super()`

`super()` gives access to the next implementation in the method resolution order.

Common use:

```python
super().__init__(...)
```

Important:

> `super()` does not simply mean "call my parent class".

It follows the class's **MRO** and is designed to cooperate with inheritance, including multiple inheritance.

## Do Not Hard-Code the Parent When Cooperative Inheritance Is Intended

Prefer:

```python
super().method()
```

over:

```python
Parent.method(self)
```

when you want the inheritance chain to cooperate.

## `isinstance()` vs `issubclass()`

```python
isinstance(obj, Class)
issubclass(Child, Parent)
```

`isinstance()` checks an object.

`issubclass()` checks class relationships.

## Inheritance vs Composition

Inheritance expresses an **is-a** relationship.

Composition often expresses a **has-a** relationship.

Use inheritance when the subtype relationship and substitutability make sense. Use composition when an object is built from or delegates to other objects.

## Source

- Python Classes / Inheritance: https://docs.python.org/3/tutorial/classes.html
- Python Data Model: https://docs.python.org/3/reference/datamodel.html

````python
class Employee:
    def __init__(self, name):
        self.name = name

    def describe(self):
        return f"Employee: {self.name}"


class Developer(Employee):
    def __init__(self, name, language):
        super().__init__(name)
        self.language = language

    def describe(self):
        return f"Developer: {self.name} - {self.language}"


developer = Developer("Ahmed", "Python")

print(developer.describe())
print(isinstance(developer, Developer))
print(isinstance(developer, Employee))
print(issubclass(Developer, Employee))
````

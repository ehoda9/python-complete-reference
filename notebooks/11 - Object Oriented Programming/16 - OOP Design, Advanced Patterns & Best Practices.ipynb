# OOP Design, Advanced Patterns & Best Practices

Guidelines for designing classes that are easy to use and to change.

| Guideline | Idea |
|---|---|
| Clear responsibilities | A class has one focused purpose |
| Composition over inheritance | Easier to change |
| Program to an interface | Depend on behavior, not on a concrete class |
| SOLID | Five design principles |
| Shallow inheritance | Deep trees make behavior hard to trace |
| Small public API | Expose only the operations users need |
| `@property` | Controlled attribute access |
| Dataclasses | For data-centric objects |
| Protocols and `__call__` | Containers and function-like objects through dunder methods |

---

## 1. Prefer Clear Responsibilities

A class should have a focused purpose.

Good:

```text
User        => user data/behavior
EmailSender => email delivery
Report      => report representation
```

Avoid giant "God classes" that do everything.

## 2. Composition Over Inheritance

When both are possible, composition can be easier to change:

```text
Order
  =>
PaymentService
ShippingService
NotificationService
```

The order uses these objects instead of inheriting from them.

Inheritance is still useful when there is a genuine subtype relationship and shared contract.

## 3. Program to an Interface

Code can depend on behavior:

```python
service.send(...)
```

instead of requiring one concrete implementation everywhere.

This works naturally with duck typing and can also be formalized with ABCs.

## 4. SOLID Overview

| Principle | Idea |
|---|---|
| S - Single Responsibility | One main reason for a class to change |
| O - Open/Closed | Extend behavior without constantly modifying stable code |
| L - Liskov Substitution | Subtypes should remain usable where the base type is expected |
| I - Interface Segregation | Prefer small focused interfaces over large ones |
| D - Dependency Inversion | Depend on abstractions/behavior rather than concrete details |

SOLID is a design guideline, not a Python syntax rule.

## 5. Avoid Deep Inheritance Trees

Deep inheritance can make behavior hard to trace.

Prefer simple hierarchies and composition when inheritance does not add real value.

## 6. Keep Public APIs Small

Expose the operations users actually need.

Internal helpers can use `_` naming conventions.

## 7. Use `@property` for Controlled Attribute Access

Properties are useful when:

- A value is derived
- Validation is needed
- You want to change the implementation without changing the public syntax

## 8. Use Dataclasses for Data-Centric Objects

If most of a class is:

```text
fields + simple representation + equality
```

a dataclass may be clearer than handwritten boilerplate.

## 9. Prefer Explicit Protocols

For containers, use Python's protocols through dunder methods:

```text
__len__
__iter__
__getitem__
__contains__
```

For callable objects:

```text
__call__
```

For context managers:

```text
__enter__
__exit__
```

## 10. Context Managers

Objects can manage resources using:

```python
with resource:
    ...
```

Implement:

```python
__enter__()
__exit__(...)
```

For async resources:

```python
__aenter__()
__aexit__(...)
```

## 11. `__call__`

An object with `__call__` can be used like a function:

```python
obj(...)
```

This is useful for stateful callable objects.

## 12. Introspection

Useful OOP inspection tools:

```text
type()
isinstance()
issubclass()
dir()
vars()
getattr()
hasattr()
Class.__dict__
Class.mro()
Class.__mro__
obj.__class__
```

## 13. Advanced Function Dispatch

`functools.singledispatchmethod` can provide type-based dispatch for methods.

Dispatch is based on the first non-`self` / non-`cls` argument.

Use it when type-specific method behavior is genuinely useful; it is not a replacement for normal polymorphism.

## 14. OOP Checklist

Before creating a class, ask:

```text
Does this object represent a clear concept?
Does it own state?
Does it have meaningful behavior?
Should this be composition instead?
Is inheritance actually needed?
What should be public?
What should be internal?
Would a dataclass be clearer?
Does the object need special methods?
```

````python
from functools import singledispatchmethod


class Formatter:
    @singledispatchmethod
    def format_value(self, value):
        return str(value)

    @format_value.register
    def _(self, value: int):
        return f"INT:{value}"

    @format_value.register
    def _(self, value: list):
        return f"LIST:{len(value)} items"


formatter = Formatter()

print(formatter.format_value("hello"))
print(formatter.format_value(10))
print(formatter.format_value([1, 2, 3]))
````

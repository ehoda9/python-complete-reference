# Polymorphism, Duck Typing & Composition

Different objects can be used through the same interface.

| Idea | Meaning |
|---|---|
| Polymorphism | Code works through a common interface while objects behave differently |
| Duck typing | What an object can do matters more than its exact type |
| Built-in polymorphism | `len()`, `+` and `for` work with many types |
| Composition | One object contains or uses another object |
| Delegation | An object forwards some work to another object |
| Association, aggregation, composition | Relationships of different strength |

---

## Polymorphism

Polymorphism means code can work through a common interface while different objects provide different implementations.

Example:

```python
animal.speak()
```

Different classes can implement `speak()` differently.

## Duck Typing

Python often cares more about what an object **can do** than its exact class.

> "If it behaves like the required object, use it."

Example:

```python
def start_engine(vehicle):
    vehicle.start()
```

The function can work with any object that provides a compatible `start()` operation.

This is one reason Python OOP can be flexible without a large inheritance hierarchy.

## Built-in Polymorphism

The language itself uses polymorphism constantly:

```python
len(list_obj)
len(string_obj)
len(dict_obj)
```

Each type implements the required behavior differently.

Operators are polymorphic too:

```python
1 + 2
"Py" + "thon"
```

## Composition

Composition means one object contains or uses another object.

```text
Car
 |
 +-- Engine
 +-- Transmission
```

This is a **has-a** relationship.

Composition often reduces coupling compared with inheritance.

## Delegation

An object can forward some work to another object:

```python
self.engine.start()
```

The outer object exposes a higher-level API while another object performs the specialized operation.

## Association / Aggregation / Composition

These terms describe object relationships with different strengths:

| Relationship | Idea |
|---|---|
| Association | Objects know/use each other |
| Aggregation | Whole uses objects that can exist independently |
| Composition | Whole strongly owns/contains component objects |

In practical Python code, the boundaries are often conceptual rather than enforced by the language.

## Source

- Python Data Model: https://docs.python.org/3/reference/datamodel.html

````python
class EmailNotification:
    def send(self, message):
        return f"Email: {message}"


class SMSNotification:
    def send(self, message):
        return f"SMS: {message}"


def notify(service, message):
    # Duck typing: only the required .send() behavior matters.
    return service.send(message)


print(notify(EmailNotification(), "Hello"))
print(notify(SMSNotification(), "Hello"))


class Engine:
    def start(self):
        return "Engine started"


class Car:
    def __init__(self, engine):
        self.engine = engine

    def start(self):
        # Composition + delegation
        return self.engine.start()


print(Car(Engine()).start())
````

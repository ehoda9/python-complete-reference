# Data Structures => References, Identity & Copying

A variable is a name that refers to an object. Assigning one name to another does **not** copy the object.

| Tool | Purpose | Example |
|---|---|---|
| `b = a` | Another name for the **same** object | `b = a` |
| `id(x)` | A number that identifies an object | `id(a)` |
| `is` / `is not` | Are they the same object? | `a is b` |
| `==` | Are the values equal? | `a == c` |
| `.copy()` | Shallow copy of a list, dict or set | `a.copy()` |
| `list(a)`, `a[:]` | Other ways to make a shallow copy | `list(a)` |
| `copy.copy(x)` | Shallow copy of any object | `copy.copy(a)` |
| `copy.deepcopy(x)` | Copy the object **and everything inside it** | `copy.deepcopy(a)` |

```python
import copy
```

---

## Names Refer to Objects

```python
a = [1, 2, 3]
b = a            # the same list, with two names
b.append(4)
print(a)         # [1, 2, 3, 4]
```

Changing the object through one name is visible through the other.

---

## Mutable and Immutable

| Kind | Types | Changed in place? |
|---|---|---|
| Immutable | `int`, `float`, `bool`, `str`, `tuple`, `frozenset`, `bytes` | No. Operations create a new object |
| Mutable | `list`, `dict`, `set`, `bytearray` | Yes |

* `x += 1` on an immutable object **rebinds** the name to a new object.
* A tuple is immutable, but a mutable object **inside** it can still change.

---

## Shallow Copy

A shallow copy creates a **new outer container**. The objects inside are **shared**.

```python
original = [[1, 2], [3, 4]]
shallow = original.copy()
shallow[0].append(99)     # the inner list is shared, so original changes too
```

Ways to make one: `.copy()`, `list(a)`, `a[:]` and `copy.copy(a)`.

---

## Deep Copy

`copy.deepcopy()` copies the object and **everything inside it**. The copy is fully independent.

```python
deep = copy.deepcopy(original)
```

---

## Which Copy?

| Situation | Use |
|---|---|
| A flat list, dict or set (no nested mutable objects) | A shallow copy |
| Nested mutable objects that must stay independent | `copy.deepcopy()` |
| You only need another name | `b = a` |

---

### Important Notes

* `==` compares values. `is` compares identity.
* `[[0] * 3] * 3` repeats the **same** inner list three times. Use `[[0] * 3 for _ in range(3)]`.
* Never use `is` to compare numbers or strings.
* Deep copies of large structures are slow and use memory.

````python
import copy

# b = a: two names, one object
a = [1, 2, 3]
b = a
b.append(4)
print(a, b, a is b, id(a) == id(b))

# == compares values, is compares identity
c = [1, 2, 3, 4]
print(a == c, a is c)

# Immutable objects cannot change in place: += rebinds the name
x = "py"
y = x
y += "thon"
print(x, y)

t = (1, [2, 3])
t[1].append(4)               # the tuple is immutable, but the list inside it is not
print(t)

# Shallow copy: a new outer list, shared inner objects
original = [[1, 2], [3, 4]]
shallow = original.copy()
shallow.append([5, 6])
shallow[0].append(99)
print(original)              # the inner list is shared
print(shallow)
print(shallow is original, shallow[0] is original[0])

# Other ways to make a shallow copy
print(list(original) == original, original[:] == original, copy.copy(original) == original)
print(list(original) is original, original[:] is original, copy.copy(original) is original)

# Deep copy: everything is copied
deep = copy.deepcopy(original)
deep[0].append(100)
print(original[0], deep[0], deep[0] is original[0])

# Dictionaries work the same way
settings = {"colors": ["red"], "size": 10}
shallow_dict = settings.copy()
shallow_dict["colors"].append("blue")
print(settings["colors"])
deep_dict = copy.deepcopy(settings)
deep_dict["colors"].append("green")
print(settings["colors"], deep_dict["colors"])

# A classic trap: repeating a list repeats the reference
grid = [[0] * 2] * 2
grid[0][0] = 1
print(grid)
safe_grid = [[0] * 2, [0] * 2]
safe_grid[0][0] = 1
print(safe_grid)
````

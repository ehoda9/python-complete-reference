# Lists

## What is a List?

A list is an **ordered, mutable collection** used to store multiple values.

### List Characteristics

1. **Ordered**
   - Items have a defined order.
   - Each item has an index.

2. **Mutable**
   - Items can be added, removed, or changed after creation.

3. **Allows Duplicates**
   - The same value can appear multiple times.

4. **Can Contain Different Data Types**
   - A single list can contain different types of objects.

5. **Can Be Nested**
   - A list can contain other lists.

---

## List Operators

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Concatenation | `[1, 2] + [3, 4]` | `[1, 2, 3, 4]` |
| `*` | Repetition | `[1, 2] * 2` | `[1, 2, 1, 2]` |
| `in` | Membership | `2 in [1, 2, 3]` | `True` |
| `not in` | Negative Membership | `5 not in [1, 2, 3]` | `True` |

---

# Indexing

Indexing is used to access a single item.

Syntax:

`list[index]`

### Positive Indexing

Starts from `0`.

| Index | `0` | `1` | `2` | `3` |
|---|---|---|---|---|
| Item | `"One"` | `"Two"` | `"Three"` | `"Four"` |

### Negative Indexing

Starts from `-1` from the end.

| Index | `-4` | `-3` | `-2` | `-1` |
|---|---|---|---|---|
| Item | `"One"` | `"Two"` | `"Three"` | `"Four"` |

---

# Slicing

Used to access multiple items.

Syntax:

`list[start:end:step]`

- `start` => included
- `end` => excluded
- `step` => movement between items
- Missing `start` => beginning
- Missing `end` => end
- Missing `step` => `1`
- Negative `step` => can move backwards

---

# Modifying Lists

Lists are mutable, so their contents can be changed.

### Change One Item

`list[index] = value`

### Change Multiple Items

`list[start:end] = iterable`

The replacement does not have to contain the same number of items.

Example:

`my_list[0:3] = ["A"]`

can replace three items with one item.

---

# List Methods

## 1. Adding Items

| Method | Purpose | Example |
|---|---|---|
| `.append(x)` | Add one item to the end | `items.append(10)` |
| `.extend(iterable)` | Add multiple items | `items.extend([10, 20])` |
| `.insert(i, x)` | Insert item at a specific index | `items.insert(1, "A")` |

### Difference

`append()` adds the entire object as **one item**.

`extend()` adds the elements of an iterable individually.

---

## 2. Removing Items

| Method | Purpose | Example |
|---|---|---|
| `.remove(x)` | Remove first matching value | `items.remove(10)` |
| `.pop([i])` | Remove and return item | `items.pop()` |
| `.clear()` | Remove all items | `items.clear()` |

### Important

- `remove()` uses a **value**.
- `pop()` uses an **index**.
- `pop()` returns the removed item.
- `pop()` without an index removes the last item.
- `remove()` raises `ValueError` if the value does not exist.
- `pop()` raises `IndexError` if the index does not exist.

---

## 3. Searching & Counting

| Method | Purpose | Example |
|---|---|---|
| `.index(x)` | Return index of first matching value | `items.index(10)` |
| `.count(x)` | Count occurrences | `items.count(10)` |

### `index()`

Returns the index of the first matching item.

Raises `ValueError` if the item is not found.

---

## 4. Ordering

| Method | Purpose | Example |
|---|---|---|
| `.sort()` | Sort the list in place | `items.sort()` |
| `.reverse()` | Reverse the list in place | `items.reverse()` |

### `sort()`

Can use:

- `reverse=True`
- `key=...`

Example:

`items.sort(reverse=True)`

### Important

`.sort()` changes the original list and returns `None`.

`.reverse()` changes the original list and returns `None`.

---

## 5. Copying

### `.copy()`

Creates a **shallow copy** of the list.

Example:

`new_list = old_list.copy()`

This creates a different list object.

---

# Important List Concepts

### `len()`

Returns the number of items.

`len(my_list)`

### `in`

Checks whether an item exists.

`"Python" in my_list`

### `not in`

Checks whether an item does not exist.

`"Java" not in my_list`

### Iteration

Lists are iterable and can be used with loops.

`for item in my_list:`

---

# Nested Lists

A list can contain other lists.

Example:

`matrix = [[1, 2], [3, 4], [5, 6]]`

Accessing nested items:

`matrix[0][1]`

Result:

`2`

---

# Important Notes

- Lists preserve insertion order.
- Lists are mutable.
- Lists allow duplicate values.
- Lists can contain different data types.
- Lists can contain other lists.
- List indexing starts from `0`.
- Negative indexing starts from `-1`.
- Slicing creates a new list.
- Most list methods modify the original list.
- Methods such as `append()`, `extend()`, `insert()`, `remove()`, `sort()`, and `reverse()` modify the list in place.
- `pop()` both modifies the list and returns the removed item.
- `copy()` creates a shallow copy.

---

## `del`

`del` removes items by position or by slice.

| Form | Effect |
|---|---|
| `del items[i]` | Remove the item at index `i` |
| `del items[1:3]` | Remove a slice |
| `del items[:]` | Remove every item (like `.clear()`) |
| `del items` | Delete the name itself

```python
# ==========================================
# 1. Creating Lists
# ==========================================

myAwesomeList = ["One", "Two", "One", 1, 100.5, True]
empty_list = []



# ==========================================
# 2. Indexing
# ==========================================
print(myAwesomeList[1])     # Two
print(myAwesomeList[-1])    # True
# ==========================================
# 3. Slicing
# ==========================================

print(myAwesomeList[1:4])
print(myAwesomeList[:4])
print(myAwesomeList[1:])
print(myAwesomeList[::1])
# Reverse
print(myAwesomeList[::-1])

# ==========================================
# 4. Modifying Items
# ==========================================
myAwesomeList[1] = "Changed"

# ==========================================
# 5. Slice Assignment
# ==========================================

myAwesomeList[0:3] = ["A"]


# ==========================================
# 6. Adding Items
# ==========================================

numbers = [1, 2, 3]

numbers.append(4)

numbers.extend([5, 6])

numbers.insert(1, 100)


# ==========================================
# 7. append() vs extend()
# ==========================================

items = [1, 2, 3]

items.append([4, 5])
print(items)
# [1, 2, 3, [4, 5]]

items = [1, 2, 3]

items.extend([4, 5])
print(items)
# [1, 2, 3, 4, 5]


# ==========================================
# 8. Removing Items
# ==========================================

items = ["A", "B", "C", "B"]

items.remove("B")

items.pop(0)

items.clear()


# ==========================================
# 9. Searching & Counting
# ==========================================

numbers = [10, 20, 30, 20, 40]

print(numbers.index(20))
print(numbers.count(20))


# ==========================================
# 10. Sorting
# ==========================================

numbers = [5, 1, 8, 3, 2]

numbers.sort()
print(numbers)

numbers.sort(reverse=True) # reverse the sequence of elements after sort it
print(numbers)


# ==========================================
# 11. Reverse
# ==========================================

numbers = [1, 2, 3, 4, 5]

numbers.reverse()
print(numbers)


# ==========================================
# 12. Copy # Shallow copy
# ==========================================

original = [1, 2, 3]

copied = original.copy()

print(original)
print(copied)

print(original is copied)   # False


# ==========================================
# 13. List Operators
# ==========================================

print([1, 2] + [3, 4]) # concat

print([1, 2] * 2) # Math operator

print(2 in [1, 2, 3]) # membership
print(5 not in [1, 2, 3]) # membership

# ==========================================
# 15. Nested Lists
# ==========================================

matrix = [
    [1, 2],
    [3, 4],
    [5, 6]
]

print(matrix[0])
print(matrix[0][1])     # 2


# ==========================================
# 16. Iteration
# ==========================================

for item in ["Python", "Java", "C++"]:
    print(item)
```

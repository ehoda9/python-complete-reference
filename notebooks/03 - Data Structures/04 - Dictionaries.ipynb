# Dictionaries

A `dict` is a **mutable mapping** that stores data as **key-value pairs**.

## Dictionary Characteristics

| Feature | Dictionary |
|---|---|
| Syntax | `{"key": "value"}` |
| Ordered | ✅* |
| Mutable | ✅ |
| Indexed | ❌ |
| Key-based Access | ✅ |
| Duplicate Keys | ❌ |
| Duplicate Values | ✅ |
| Key Type | Hashable |
| Value Type | Any object |

\* Dictionaries preserve insertion order in modern Python.

---

# Key : Value

Each dictionary item consists of:

`key: value`

Example:

`{"name": "Mahmoud"}`

- `"name"` => key
- `"Mahmoud"` => value

---

# Dictionary Keys

Dictionary keys must be **hashable**.

### Valid Keys

- `int`
- `float`
- `str`
- `bool`
- `tuple` containing only hashable elements
- `frozenset`

### Invalid Keys

- `list`
- `dict`
- `set`

### Important

Keys must be **unique**.

If the same key appears more than once, the last value replaces the previous one.

---

# Dictionary Values

Values can be **any Python object**.

A value can be:

- Number
- String
- Boolean
- List
- Tuple
- Set
- Frozenset
- Dictionary
- Function
- Object
- `None`
- etc.

---

# Accessing Values

Dictionary values are accessed using their keys.

`user["name"]`

### `.get()`

`.get(key)` can also retrieve a value.

Difference:

| Method | Missing Key |
|---|---|
| `user["name"]` | Raises `KeyError` |
| `user.get("name")` | Returns `None` by default |

`.get()` can also provide a default value:

`user.get("email", "Not Found")`

---

# Adding & Updating Items

### Add

`user["email"] = "example@email.com"`

### Update

`user["age"] = 25`

If the key exists => value is updated.

If the key does not exist => new key-value pair is added.

---

# Removing Items

| Method | Purpose |
|---|---|
| `.pop(key)` | Remove and return value |
| `.popitem()` | Remove and return the last inserted key-value pair |
| `.clear()` | Remove all items |

`del user["name"]` can also remove a specific key-value pair.

---

# Dictionary Methods

## Accessing Keys & Values

| Method | Purpose |
|---|---|
| `.keys()` | Return dictionary keys |
| `.values()` | Return dictionary values |
| `.items()` | Return key-value pairs |

---

## Updating

| Method | Purpose |
|---|---|
| `.update()` | Add/update multiple key-value pairs |
| `.setdefault()` | Get a value and optionally insert a default if the key does not exist |

---

## Copying

`.copy()` creates a **shallow copy**.

---

# Iterating Through a Dictionary

By default, iterating over a dictionary gives its keys.

`for key in user:`

### Keys

`for key in user.keys():`

### Values

`for value in user.values():`

### Keys & Values

`for key, value in user.items():`

---

# Nested Dictionaries

A dictionary can contain other dictionaries as values.

Example:

`users = {`

`    "user1": {"name": "Ahmed", "age": 20},`

`    "user2": {"name": "Mona", "age": 22}`

`}`

Access nested values:

`users["user1"]["name"]`

---

# Dictionary Length

`len(dictionary)`

Returns the number of **key-value pairs**.

For example:

`{"a": 1, "b": 2}`

has length `2`.

---

# Creating Dictionaries

### Literal

`user = {"name": "Ahmed", "age": 20}`

### Empty Dictionary

`user = {}`

### `dict()`

`user = dict(name="Ahmed", age=20)`

### From Key-Value Pairs

`dict([("name", "Ahmed"), ("age", 20)])`

### `dict.fromkeys()`

`dict.fromkeys(keys, value)` creates a dictionary whose keys come from an iterable and whose values are all the same object.

`dict.fromkeys(["a", "b"], 0)` => `{"a": 0, "b": 0}`

Without a value, every value is `None`.

Avoid a mutable value such as `[]`: every key would share the same list.

---

# Important Notes

- Dictionary keys must be hashable.
- Dictionary values can be any object.
- Keys are unique.
- Values can be duplicated.
- Dictionaries are mutable.
- Dictionaries preserve insertion order.
- Dictionaries are accessed by keys, not numeric indexes.
- A dictionary can contain other containers as values.
- A dictionary can be nested inside another dictionary.

```python
# ==========================================
# 1. Creating a Dictionary
# ==========================================

user = {
    "name": "Mahmoud",
    "age": 20,
    "country": "Egypt",
    "skills": ["HTML", "CSS", "JS"],
    "rating": 10.5
}


# ==========================================
# 2. Accessing Values
# ==========================================

user["name"]

user.get("country")

user.get("email", "Not Found")


# ==========================================
# 3. Keys, Values & Items
# ==========================================

user.keys()

user.values()

user.items()


# ==========================================
# 4. Adding & Updating
# ==========================================

user["email"] = "example@email.com"

user["age"] = 21

user.update({
    "country": "Egypt",
    "rating": 9.5
})


# ==========================================
# 5. Removing Items
# ==========================================

user.pop("rating")

user.popitem()

del user["country"]

user.clear()


# ==========================================
# 6. Nested Dictionary
# ==========================================

languages = {
    "one": {
        "name": "HTML",
        "progress": "80%"
    },
    "two": {
        "name": "CSS",
        "progress": "90%"
    },
    "three": {
        "name": "JavaScript",
        "progress": "90%"
    }
}


languages["one"]

languages["three"]["name"]


# ==========================================
# 7. Dictionary Length
# ==========================================

len(languages)

len(languages["two"])


# ==========================================
# 8. Dictionary from Variables
# ==========================================

framework_one = {
    "name": "Vue",
    "progress": "80%"
}

framework_two = {
    "name": "React",
    "progress": "80%"
}

framework_three = {
    "name": "Angular",
    "progress": "80%"
}

all_frameworks = {
    "one": framework_one,
    "two": framework_two,
    "three": framework_three
}


# ==========================================
# 9. Dictionary Iteration
# ==========================================

for key in all_frameworks:
    pass

for key in all_frameworks.keys():
    pass

for value in all_frameworks.values():
    pass

for key, value in all_frameworks.items():
    pass


# ==========================================
# 10. Dictionary Creation
# ==========================================

user = dict(
    name="Mahmoud",
    age=20
)

user = dict([
    ("name", "Mahmoud"),
    ("age", 20)
])
```

# Regex => Groups, Alternation & Assertions

Groups capture parts of a match. Assertions check the surroundings.

| Feature | Syntax | Meaning |
|---|---|---|
| Capturing group | `(...)` | Capture the matched text |
| Non-capturing group | `(?:...)` | Group without capturing |
| Named group | `(?P<name>...)` | Capture with a name |
| Alternation | `a\|b` | Match `a` or `b` |
| Backreference | `\1`, `(?P=name)` | Match text captured earlier |
| Lookahead | `(?=...)`, `(?!...)` | Followed / not followed by |
| Lookbehind | `(?<=...)`, `(?<!...)` | Preceded / not preceded by |
| Match object | `re.Match` | Holds the result of a successful match |

---

## Capturing Groups

Parentheses create capturing groups:

```text
(...)
```

They let you extract part of a match.

## Non-Capturing Groups

Use:

```text
(?:...)
```

when you need grouping but do not need to capture the text.

## Named Groups

Use:

```text
(?P<name>...)
```

This gives the group a readable name.

## Alternation

The `|` operator means OR:

```text
cat|dog
```

## Backreferences

A backreference matches text captured earlier:

```text
(\w+)\s+\1
```

This can detect repeated words.

Named backreferences can use:

```text
(?P=name)
```

## Lookahead

Positive:

```text
(?=...)
```

The following pattern must match, but it is not consumed.

Negative:

```text
(?!...)
```

The following pattern must not match.

## Lookbehind

Positive:

```text
(?<=...)
```

The preceding pattern must match.

Negative:

```text
(?<!...)
```

The preceding pattern must not match.

Lookbehind patterns have important fixed-length restrictions in the standard `re` engine for many forms.

## Match Object

Regex matching functions may return a `re.Match` object.

Useful methods:

| Method | Purpose |
|---|---|
| `group()` | Get matched text |
| `group(n)` | Get a numbered group |
| `groups()` | Get all captured groups |
| `groupdict()` | Get named groups as a dictionary |
| `start()` | Start index |
| `end()` | End index |
| `span()` | `(start, end)` |
| `string` | Original string |
| `re` | Compiled pattern |

## Group Numbering

Groups are numbered from left to right:

```text
( first group )
( second group )
```

Nested groups still receive their own numbers.

## Source

````python
import re

text = "Ahmed: 25"

# re.search(pattern, string): return the first Match object
match = re.search(r"(?P<name>\w+):\s*(?P<age>\d+)", text)

if match:
    print(match.group())
    print(match.group("name"))
    print(match.group("age"))
    print(match.groupdict())
    print(match.span())

# Non-capturing group
print(re.findall(r"(?:Mr|Ms)\.?\s+\w+", "Mr Ahmed, Ms Mona"))

# Alternation
print(re.findall(r"cat|dog", "cat dog bird"))

# Lookahead
print(re.findall(r"\d+(?= USD)", "10 USD 20 EUR 30 USD"))

# Backreference
print(re.findall(r"\b(\w+)\s+\1\b", "hello hello world"))
````

## Exceptions Handling => Try | Except | Else | Finally

Python provides four main keywords for handling exceptions:

| Keyword   | Purpose                                     |
| --------- | ------------------------------------------- |
| `try`     | Contains code that may cause an exception   |
| `except`  | Handles a matching exception                |
| `else`    | Runs only if no exception occurred in `try` |
| `finally` | Runs whether an exception occurred or not   |

### Basic Structure

```text
try:
    # Code that may cause an exception

except SomeError:
    # Handle the exception

else:
    # Runs when no exception occurs

finally:
    # Runs in both cases
```

### Important

* `try` must be followed by at least one `except` or `finally`.
* `else` is optional.
* `finally` is optional.
* `else` runs only when the `try` block finishes successfully.
* `finally` normally runs whether the code succeeds or fails.
* If an exception is not handled by any `except`, it continues upward and may terminate the program.


````python
# try: tests code that may raise an exception
try:
    number = int(input("Enter a number: "))
    result = 10 / number

# except: handles a specific exception
except ZeroDivisionError:
    print("Cannot divide by zero.")

# else: runs only when no exception occurs
else:
    print("Result:", result)

# finally: runs whether an exception happened or not
finally:
    print("Finished.")
````

# Type Conversion

Type conversion means changing a value from one data type to another.

Python provides built-in functions for converting between common data types.

## Common Type Conversion Functions

| Function | Converts To | Example | Result |
|---|---|---|---|
| `int()` | `int` | `int("10")` | `10` |
| `float()` | `float` | `float("10.5")` | `10.5` |
| `str()` | `str` | `str(10)` | `"10"` |
| `bool()` | `bool` | `bool(10)` | `True` |
| `list()` | `list` | `list("ABC")` | `['A', 'B', 'C']` |
| `tuple()` | `tuple` | `tuple("ABC")` | `('A', 'B', 'C')` |
| `set()` | `set` | `set("ABC")` | `{'A', 'B', 'C'}` |
| `dict()` | `dict` | `dict([("a", 1)])` | `{'a': 1}` |

## Converting to `int`

Used to convert compatible values into integers.

- `int("10")` => `10`
- `int(10.9)` => `10`
- `int("10.5")` => `ValueError`

> Converting a `float` to `int` removes the decimal part. It does not round the number.

## Converting to `float`

Used to convert compatible values into floating-point numbers.

- `float("10")` => `10.0`
- `float("10.5")` => `10.5`
- `float(10)` => `10.0`

## Converting to `str`

Almost any Python object can be converted to a string representation.

- `str(100)` => `"100"`
- `str(10.5)` => `"10.5"`
- `str(True)` => `"True"`

## Converting to `bool`

`bool()` follows Python's **truthy / falsy** rules.

- `bool(1)` => `True`
- `bool(0)` => `False`
- `bool("Hello")` => `True`
- `bool("")` => `False`
- `bool([1, 2])` => `True`
- `bool([])` => `False`

## Converting Between Collections

Some iterable objects can be converted into other collection types.

- `list()` => creates a list
- `tuple()` => creates a tuple
- `set()` => creates a set

Example:

`list("ABC")` => `['A', 'B', 'C']`

## Important Notes

- Type conversion creates a value of the target type.
- Not every value can be converted successfully.
- Invalid conversions can raise errors such as `ValueError` or `TypeError`.
- `input()` always returns a `str`, so numeric input usually needs conversion.

Example:

`age = int(input("Enter your age: "))`

Here, `input()` returns a string first, then `int()` converts it to an integer.

```python
# ---------------------
# -- Type Conversion --
# ---------------------

# String => Integer
age = int("20")

# String => Float
price = float("99.5")

# Integer / Float => String
number = str(100)
decimal = str(10.5)

# Value => Boolean
bool(1)
bool(0)
bool("Python")
bool("")

# String => List / Tuple / Set
list("ABC")
tuple("ABC")
set("ABC")

# List => Tuple
tuple([1, 2, 3])

# Tuple => List
list((1, 2, 3))

# List => Set
set([1, 2, 2, 3])

# User input
age = int(input("Enter your age: "))
```

# String Formatting

String formatting builds text that contains values. Python has three styles. Use **f-strings** by default.

| Style | Example | Notes |
|---|---|---|
| f-string | `f"{name} is {age}"` | Modern default, fastest to read |
| `str.format()` | `"{} is {}".format(name, age)` | Useful when the template is stored separately |
| `%` operator | `"%s is %d" % (name, age)` | Old style, still seen in legacy code |
| `format()` built-in | `format(3.14159, ".2f")` | Formats one value |

---

## f-Strings

Put `f` before the quotes. Anything inside `{}` is a Python expression.

```python
name = "Ann"
f"Hello {name}"           # 'Hello Ann'
f"{2 + 3}"                # '5'
f"{name.upper()}"         # 'ANN'
f"{{literal braces}}"     # '{literal braces}'
```

### Important

* Use `{{` and `}}` to print real braces.
* Before Python 3.12, an expression could not reuse the same quote type as the f-string. From 3.12 it can.

---

## Format Specification

Everything after `:` inside `{}` is the **format spec**.

`{value:[[fill]align][sign][#][0][width][grouping][.precision][type]}`

| Part | Values | Meaning |
|---|---|---|
| `align` | `<` `>` `^` `=` | Left, right, center, pad after the sign |
| `fill` | any character | Padding character (before the align) |
| `sign` | `+` `-` (space) | Show sign always, only for negatives, or a space |
| `#` | | Alternate form (`0x`, `0b`, `0o` prefix) |
| `0` | | Zero padding |
| `width` | number | Minimum width |
| `grouping` | `,` `_` | Thousands separator |
| `.precision` | number | Digits after the point (floats), max length (strings) |
| `type` | see below | How to present the value |

### Type Codes

| Code | Meaning | Example | Result |
|---|---|---|---|
| `s` | String | `f"{'hi':s}"` | `hi` |
| `d` | Integer | `f"{42:d}"` | `42` |
| `f` | Fixed-point | `f"{3.14159:.2f}"` | `3.14` |
| `e` | Scientific | `f"{12345.678:.2e}"` | `1.23e+04` |
| `%` | Percentage | `f"{0.256:.1%}"` | `25.6%` |
| `b` `o` `x` `X` | Binary, octal, hex | `f"{255:x}"` | `ff` |
| `,` | Thousands separator | `f"{1234567:,}"` | `1,234,567` |

---

## Alignment and Padding

```python
f"{'hi':>6}"      # '    hi'   right
f"{'hi':<6}"      # 'hi    '   left
f"{'hi':^6}"      # '  hi  '   center
f"{'hi':*^8}"     # '***hi***' fill + center
f"{42:05d}"       # '00042'    zero padding
```

Numbers align to the **right** by default. Strings align to the **left**.

---

## Dynamic Width and Precision

The width and precision can come from variables:

```python
width, precision = 10, 3
f"{3.14159265:{width}.{precision}f}"   # '     3.142'
```

---

## Conversions: `!r`, `!s`, `!a`

| Flag | Calls | Use |
|---|---|---|
| `!s` | `str()` | Human-readable (default) |
| `!r` | `repr()` | Debug view, shows quotes |
| `!a` | `ascii()` | Escapes non-ASCII characters |

```python
f"{'hi'!r}"    # "'hi'"
```

---

## Debugging With `=`

`{x=}` prints the expression **and** its value (Python 3.8+).

```python
x = 10
f"{x=}"           # 'x=10'
f"{x * 2 = }"     # 'x * 2 = 20'
f"{3.14159=:.2f}" # '3.14159=3.14'
```

---

## Formatting Dates

`datetime` objects accept date format codes:

```python
f"{now:%Y-%m-%d %H:%M}"
```

---

## `str.format()`

| Field | Example | Result |
|---|---|---|
| Automatic | `"{} {}".format("a", "b")` | `a b` |
| Positional | `"{1} {0}".format("a", "b")` | `b a` |
| Named | `"{x} {y}".format(x=1, y=2)` | `1 2` |
| Attribute / key | `"{p.x} {d[k]}".format(p=obj, d=data)` | reads `obj.x` and `data["k"]` |

The same format spec works: `"{:>8.2f}".format(3.14159)`.

---

## `%` Formatting (Legacy)

| Code | Meaning |
|---|---|
| `%s` | String |
| `%d` | Integer |
| `%f` / `%.2f` | Float |
| `%5s` | Width 5 |

```python
"%s is %d years old" % ("Ann", 30)
```

---

## Which One?

| Situation | Use |
|---|---|
| Normal code | f-string |
| Template stored in a variable or file | `str.format()` |
| Reading old code | `%` (recognize it, do not write new code with it) |
| Logging messages | `logger.info("x=%s", x)` (see Logging) |

````python
from datetime import datetime

name, age = "Ann", 30

# The three styles
print(f"{name} is {age}")
print("{} is {}".format(name, age))
print("%s is %d" % (name, age))
print(format(3.14159, ".2f"))

# Expressions inside f-strings
print(f"{2 + 3}", f"{name.upper()}", f"{{literal braces}}")

# Alignment, fill and padding
print(f"[{'hi':>6}] [{'hi':<6}] [{'hi':^6}] [{'hi':*^8}]")
print(f"{42:05d}", f"{7:>4}", f"{'left':<6}|")

# Numbers
print(f"{3.14159:.2f}", f"{1234567:,}", f"{1234567:_}", f"{0.256:.1%}")
print(f"{12345.678:.2e}", f"{255:x}", f"{255:#x}", f"{255:b}", f"{255:#o}")
print(f"{5:+d}", f"{-5:+d}", f"{5: d}")

# Dynamic width and precision
width, precision = 10, 3
print(f"[{3.14159265:{width}.{precision}f}]")

# Conversions
print(f"{'hi'!s}", f"{'hi'!r}", f"{'é'!a}")

# Debugging with =
x = 10
print(f"{x=}")
print(f"{x * 2 = }")
print(f"{3.14159=:.2f}")

# Dates
moment = datetime(2026, 9, 20, 14, 30)
print(f"{moment:%Y-%m-%d %H:%M}")

# str.format(): positional, named, attribute and key access
class Point:
    x, y = 1, 2

data = {"k": "value"}
print("{1} {0}".format("a", "b"))
print("{x} {y}".format(x=1, y=2))
print("{p.x} {d[k]}".format(p=Point(), d=data))
print("{:>8.2f}".format(3.14159))

# % formatting (legacy)
print("%5s|%-5s|%05.1f" % ("ab", "cd", 3.14159))

# A small aligned table
rows = [("apple", 3, 1.5), ("banana", 12, 0.25), ("cherry", 100, 12.0)]
for item, qty, price in rows:
    print(f"{item:<10}{qty:>5}{price:>8.2f}")
````

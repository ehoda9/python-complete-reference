# Keywords Reference
Keywords are **reserved words** in Python - you can't use them as variable, function, or class names because they have a fixed, special meaning in the language's syntax.

To see the full list interactively, use:
```python
help("keywords")
```

````python
import keyword

# help("keywords") prints the list interactively - uncomment to try it:
# help("keywords")

print("Total keywords:", len(keyword.kwlist))
print(keyword.kwlist)
print(keyword.iskeyword("for"))   # True
print(keyword.iskeyword("data"))  # False
````

### `False`
Boolean value representing false.

````python
is_valid = False
print(is_valid, type(is_valid))
````

### `None`
Represents the absence of a value (Python's "null").

````python
result = None
print(result is None)
````

### `True`
Boolean value representing true.

````python
is_ready = True
print(is_ready, type(is_ready))
````

### `and`
Logical AND — True only if both operands are True.

````python
x = 5
print(x > 0 and x < 10)
````

### `as`
Creates an alias — used with `import`, `with`, and `except`.

````python
import math as m
print(m.pi)
````

### `assert`
Tests a condition and raises `AssertionError` if it's False. Used for debugging/sanity checks.

````python
x = 10
assert x > 0, "x must be positive"
print("Assertion passed")
````

### `async`
Declares an asynchronous function (a coroutine).

````python
import asyncio

async def greet():
    return "Hello, async world!"

print(asyncio.run(greet()))
````

### `await`
Pauses an `async` function until the awaited coroutine finishes.

````python
import asyncio

async def get_value():
    await asyncio.sleep(0)
    return 42

print(asyncio.run(get_value()))
````

### `break`
Exits the nearest enclosing loop immediately.

````python
for i in range(10):
    if i == 3:
        break
    print(i)
````

### `class`
Defines a new class - a blueprint for creating objects.

````python
class Dog:
    def __init__(self, name):
        self.name = name

d = Dog("Rex")
print(d.name)
````

### `continue`
Skips the rest of the current loop iteration and moves to the next one.

````python
for i in range(5):
    if i % 2 == 0:
        continue
    print(i)
````

### `def`
Defines a function.

````python
def add(a, b):
    return a + b

print(add(2, 3))
````

### `del`
Deletes a variable, a list item, or a dictionary key.

````python
x = [1, 2, 3]
del x[1]
print(x)
````

### `elif`
"Else if" — checks another condition if the previous `if`/`elif` was False.

````python
score = 75
if score >= 90:
    print("A")
elif score >= 70:
    print("B")
else:
    print("C")
````

### `else`
Runs when the `if` (and all `elif`) conditions are False; also used with loops and `try`.

````python
x = 5
if x > 10:
    print("big")
else:
    print("small")
````

### `except`
Catches and handles an exception raised inside a `try` block.

````python
try:
    1 / 0
except ZeroDivisionError:
    print("Cannot divide by zero")
````

### `finally`
A block that always runs after `try`/`except`, whether an error happened or not.

````python
try:
    x = 1
finally:
    print("This always runs")
````

### `for`
Starts a loop that iterates over a sequence (list, range, string, ...).

````python
for fruit in ["apple", "banana"]:
    print(fruit)
````

### `from`
Imports specific names from a module instead of the whole module.

````python
from math import sqrt
print(sqrt(16))
````

### `global`
Declares that a variable inside a function refers to the global (module-level) scope.

````python
counter = 0

def increment():
    global counter
    counter += 1

increment()
print(counter)
````

### `if`
Starts a conditional statement — runs the block only if the condition is True.

````python
x = 7
if x > 5:
    print("x is greater than 5")
````

### `import`
Loads a module so its functions/classes/variables can be used.

````python
import math
print(math.floor(4.7))
````

### `in`
Checks membership (`x in y`), or is used to iterate in a `for` loop.

````python
print(3 in [1, 2, 3])
````

### `is`
Checks identity — whether two names point to the exact same object (not just equal value).

````python
a = None
print(a is None)
````

### `lambda`
Creates a small, anonymous, single-expression function.

````python
square = lambda x: x ** 2
print(square(5))
````

### `nonlocal`
Declares that a variable refers to the nearest enclosing (non-global) function scope.

````python
def outer():
    x = 1
    def inner():
        nonlocal x
        x += 1
    inner()
    return x

print(outer())
````

### `not`
Logical NOT — inverts a boolean value.

````python
print(not False)
````

### `or`
Logical OR — True if at least one operand is True.

````python
print(False or True)
````

### `pass`
A null operation / placeholder statement — does absolutely nothing.

````python
def todo():
    pass

print("Function defined, does nothing yet")
````

### `raise`
Manually triggers (raises) an exception.

````python
try:
    raise ValueError("Custom error")
except ValueError as e:
    print(e)
````

### `return`
Exits a function and optionally sends a value back to the caller.

````python
def square(x):
    return x * x

print(square(4))
````

### `try`
Starts a block used to test code for exceptions.

````python
try:
    print(10 / 2)
except ZeroDivisionError:
    print("Error")
````

### `while`
Starts a loop that keeps running as long as its condition is True.

````python
i = 0
while i < 3:
    print(i)
    i += 1
````

### `with`
Used for context managers — e.g. automatically closing a file when done.

````python
with open("example.txt", "w") as f:
    f.write("Hello")

print("File written and closed automatically")
````

### `yield`
Used inside a generator function to produce a value and pause execution until resumed.

````python
def counter():
    yield 1
    yield 2
    yield 3

for num in counter():
    print(num)
````

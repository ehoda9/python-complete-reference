# Type Hinting => TypeVar, Generics & Special Type Forms

Generic types, exact values and class-level markers.

| Tool | Purpose |
|---|---|
| `TypeVar` | A type that stays consistent across related positions |
| Constrained `TypeVar` | `TypeVar("T", int, float)`: one of the listed types |
| Bounded `TypeVar` | `TypeVar("T", bound=BaseClass)`: a subtype of the bound |
| `Generic` | Generic classes such as `Box[T]` |
| `Literal` | A limited set of exact values |
| `Final` | A name that should not be reassigned |
| `ClassVar` | An attribute that belongs to the class, not to instances |
| `Required` / `NotRequired` | Mark keys of a `TypedDict` |

---

## `TypeVar`

`TypeVar` represents a type that stays consistent across related positions.

```python
T = TypeVar("T")

def first(items: list[T]) -> T:
    return items[0]
```

The type checker can infer that the returned value has the same type as the elements.

## Constrained TypeVar

```python
T = TypeVar("T", int, float)
```

`T` must be one of the listed types.

## Bounded TypeVar

```python
T = TypeVar("T", bound=BaseClass)
```

`T` can be a subtype of the bound.

## Generic Classes

```python
T = TypeVar("T")

class Box(Generic[T]):
    ...
```

Modern Python also supports type parameter syntax:

```python
class Box[T]:
    ...
```

and:

```python
def first[T](items: list[T]) -> T:
    ...
```

## `Literal`

Use `Literal` for a limited set of exact values:

```python
Literal["read", "write"]
```

Useful when values are fixed and meaningful.

## `Final`

Marks a name as not intended to be reassigned.

```python
MAX_RETRIES: Final = 3
```

Type checkers can flag reassignment.

## `ClassVar`

Marks an attribute as class-level rather than instance-level for typing purposes.

```python
class User:
    users_count: ClassVar[int] = 0
```

`ClassVar` is especially relevant when type checkers analyze classes and dataclasses.

## `Required` / `NotRequired`

These are mainly used with `TypedDict` to describe which dictionary keys must or need not exist.

## Source

https://docs.python.org/3/library/typing.html

````python
from typing import ClassVar, Final, Generic, Literal, TypeVar

T = TypeVar("T")

# TypeVar: input and return keep the same type
def first(items: list[T]) -> T:
    return items[0]

print(first([10, 20, 30]))
print(first(["a", "b"]))

# Generic class
class Box(Generic[T]):
    def __init__(self, value: T):
        self.value = value

box = Box("Python")
print(box.value)

# Literal: only specific values are valid for static checking
def set_mode(mode: Literal["read", "write"]) -> str:
    return mode.upper()

print(set_mode("read"))

# Final: type checkers can flag reassignment
MAX_RETRIES: Final[int] = 3

class User:
    # ClassVar: class-level attribute
    users_count: ClassVar[int] = 0
````

# Boolean

## What is Boolean?

`bool` is Python's built-in type for representing one of two truth values:

- `True`
- `False`

Boolean values are used to represent whether a condition or expression is true or false.

---

# Boolean Values

| Value | Meaning |
|---|---|
| `True` | The condition is true |
| `False` | The condition is false |

`True` and `False` are Boolean constants.

---

# Boolean Expressions

Comparisons produce Boolean values.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `>` | Greater than | `10 > 5` | `True` |
| `<` | Less than | `10 < 5` | `False` |
| `>=` | Greater than or equal | `10 >= 10` | `True` |
| `<=` | Less than or equal | `10 <= 5` | `False` |
| `==` | Equal to | `10 == 10` | `True` |
| `!=` | Not equal to | `10 != 5` | `True` |

---

# `bool()`

`bool()` converts an object to a Boolean value.

Syntax:

`bool(value)`

---

# Truthy Values

An object is **truthy** when `bool(object)` returns `True`.

Examples include:

- Non-zero numbers
- Non-empty strings
- Non-empty lists
- Non-empty tuples
- Non-empty sets
- Non-empty dictionaries
- `True`

Examples:

`bool(10)` => `True`

`bool("Hello")` => `True`

`bool([1, 2])` => `True`

---

# Falsy Values

An object is **falsy** when `bool(object)` returns `False`.

Important built-in falsy values:

| Value | Type |
|---|---|
| `False` | `bool` |
| `None` | `NoneType` |
| `0` | `int` |
| `0.0` | `float` |
| `0j` | `complex` |
| `""` | `str` |
| `[]` | `list` |
| `()` | `tuple` |
| `{}` | `dict` |
| `set()` | `set` |
| `range(0)` | `range` |

### Important

Empty containers are falsy.

Non-empty containers are generally truthy.

---

# Boolean in Conditions

Python automatically evaluates an object's truth value when it is used in a condition.

Example:

`if "Hello":`

The string is non-empty, so it is treated as `True`.

Example:

`if []:`

The list is empty, so it is treated as `False`.

This is called **truth-value testing**.

---

# Important Notes

- `True` and `False` are Boolean values.
- `bool` is a built-in Python type.
- `True` and `False` are case-sensitive.
- `true` and `false` are not valid Python Boolean constants.
- `bool()` returns either `True` or `False`.
- Boolean values are commonly produced by comparison and logical expressions.
- `1` is truthy and `0` is falsy, but they are not the same type as `True` and `False`.

```python
# ==========================================
# 1. Boolean Values
# ==========================================

is_logged_in = True
is_admin = False


# ==========================================
# 2. Boolean Expressions
# ==========================================

10 > 5
10 < 5

10 >= 10
10 <= 5

10 == 10
10 != 5


# ==========================================
# 3. bool()
# ==========================================

bool(True)
bool(False)

bool(100)
bool(-10)
bool(0)

bool("Hello")
bool("")

bool([1, 2, 3])
bool([])

bool((1, 2, 3))
bool(())

bool({1, 2, 3})
bool(set())

bool({"name": "Mahmoud"})
bool({})

bool(None)


# ==========================================
# 4. Truthy Values
# ==========================================

bool(100)
bool(-100)
bool(0.5)

bool("Python")
bool([1, 2, 3])
bool((1, 2))
bool({1, 2})
bool({"name": "Mahmoud"})


# ==========================================
# 5. Falsy Values
# ==========================================

bool(False)
bool(None)

bool(0)
bool(0.0)
bool(0j)

bool("")

bool([])
bool(())
bool({})
bool(set())

bool(range(0))


# ==========================================
# 6. Truth-Value Testing
# ==========================================

name = "Mahmoud"

if name:
    pass

items = []

if items:
    pass


# ==========================================
# 7. bool vs int
# ==========================================

print(type(True))
print(type(1))

print(True == 1)
print(False == 0)
```

# Containers Comparison

Python provides several built-in container / collection types for storing multiple values.

## Main Built-in Containers

| Type | Ordered | Mutable | Allows Duplicates | Indexed | Hashable | Main Syntax |
|---|---:|---:|---:|---:|---:|---|
| `list` | ✅ | ✅ | ✅ | ✅ | ❌ | `[1, 2, 3]` |
| `tuple` | ✅ | ❌ | ✅ | ✅ | ✅* | `(1, 2, 3)` |
| `set` | ❌ | ✅ | ❌ | ❌ | ❌ | `{1, 2, 3}` |
| `frozenset` | ❌ | ❌ | ❌ | ❌ | ✅ | `frozenset({1, 2, 3})` |
| `dict` | ✅** | ✅ | Keys: ❌ | Key-based | ❌ | `{"a": 1}` |
| `range` | ✅ | ❌ | N/A | ✅ | ✅ | `range(5)` |

### Notes

- **Ordered** => preserves element order.
- **Mutable** => can be changed after creation.
- **Duplicates** => whether the container can contain duplicate elements.
- **Indexed** => supports access using numeric indexes such as `items[0]`.
- **Hashable** => can be used as a dictionary key or as an element of a set.

\* A tuple is hashable only if all of its elements are hashable.

\** Dictionaries preserve insertion order in modern Python.

---

# Quick Selection Guide

| If you need... | Use |
|---|---|
| Ordered, changeable collection | `list` |
| Ordered, immutable collection | `tuple` |
| Unique, changeable values | `set` |
| Unique, immutable values | `frozenset` |
| Key => Value relationships | `dict` |
| Immutable sequence of numbers | `range` |

---

# Nested Containers

Containers can often contain other containers.

## General Rule

A container can contain another object if the inner object is allowed as an element/value/key of the outer container.

### Examples

| Outer Container | Can Contain |
|---|---|
| `list` | Almost any Python object |
| `tuple` | Almost any Python object |
| `set` | Hashable objects only |
| `frozenset` | Hashable objects only |
| `dict` keys | Hashable objects only |
| `dict` values | Almost any Python object |

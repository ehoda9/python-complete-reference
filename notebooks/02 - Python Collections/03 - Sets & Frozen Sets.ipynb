# Sets & Frozen Sets

## Set

A `set` is an **unordered, mutable collection of unique elements**.

### Set Characteristics

| Feature | Set |
|---|---|
| Syntax | `{1, 2, 3}` |
| Ordered | ❌ |
| Indexed | ❌ |
| Slicing | ❌ |
| Mutable | ✅ |
| Allows Duplicates | ❌ |
| Elements | Hashable only |
| Hashable | ❌ |

### Same as Lists / Tuples

Sets can:

- Contain different data types
- Be nested inside containers that accept them
- Be iterated over
- Use `len()`
- Use membership operators: `in`, `not in`

### Main Differences

- Sets do not support indexing.
- Sets do not support slicing.
- Sets contain unique elements.
- Set elements must be hashable.

---

# Creating a Set

Use curly braces:

`my_set = {1, 2, 3}`

### Empty Set مهم

`set()` => empty set

`{}` => empty dictionary, **not** an empty set.

---

# Set Elements & Hashability

Set elements must be **hashable**.

### Can Be Set Elements

- `int`
- `float`
- `bool`
- `str`
- `tuple` containing only hashable elements
- `frozenset`

### Cannot Be Set Elements

- `list`
- `dict`
- `set`

### Important

A set can contain a tuple:

`{(1, 2), (3, 4)}`

But a tuple containing an unhashable object cannot be a set element:

`{([1, 2], 3)}` => invalid

---

# Set Operations

Sets are especially useful for mathematical set operations.

| Operator | Operation | Example |
|---|---|---|
| `\|` | Union | `A \| B` |
| `&` | Intersection | `A & B` |
| `-` | Difference | `A - B` |
| `^` | Symmetric Difference | `A ^ B` |

### Meaning

- **Union** => all elements from both sets
- **Intersection** => elements common to both sets
- **Difference** => elements in the first set but not the second
- **Symmetric Difference** => elements that exist in either set, but not both

---

# Set Methods

## Adding

| Method | Purpose |
|---|---|
| `.add(x)` | Add one element |
| `.update(iterable)` | Add multiple elements |

## Removing

| Method | Purpose |
|---|---|
| `.remove(x)` | Remove element; raises `KeyError` if not found |
| `.discard(x)` | Remove element; does nothing if not found |
| `.pop()` | Remove and return an arbitrary element |
| `.clear()` | Remove all elements |

## Set Operations

| Method | Purpose |
|---|---|
| `.union()` | Union |
| `.intersection()` | Intersection |
| `.difference()` | Difference |
| `.symmetric_difference()` | Symmetric difference |

## Relationship Testing

| Method | Purpose |
|---|---|
| `.issubset()` | Check whether another set contains all elements |
| `.issuperset()` | Check whether the set contains all elements of another set |
| `.isdisjoint()` | Check whether two sets have no common elements |

# Frozen Set

A `frozenset` is an **immutable version of a set**.

### Characteristics

| Feature | `set` | `frozenset` |
|---|---|---|
| Ordered | ❌ | ❌ |
| Indexed | ❌ | ❌ |
| Mutable | ✅ | ❌ |
| Allows Duplicates | ❌ | ❌ |
| Hashable | ❌ | ✅ |
| Elements must be hashable | ✅ | ✅ |

### Important

`frozenset` supports set operations, but it does not have methods that modify the collection.

For example:

- `set.add()` => available
- `frozenset.add()` => not available
- `set.remove()` => available
- `frozenset.remove()` => not available

---

# Why Use frozenset?

Use `frozenset` when you need:

- A set that cannot be modified.
- A set as an element inside another set.
- A set as a dictionary key.

Example:

`my_set = {frozenset({1, 2}), frozenset({3, 4})}`

Example dictionary key:

`data = {frozenset({1, 2}): "value"}`

---

# Important Rules

### Set

`set` => mutable + unique + unhashable

### Frozen Set

`frozenset` => immutable + unique + hashable

### Remember

> **Set = mutable unique collection**

> **Frozenset = immutable unique collection**

```python
# ==========================================
# 1. Creating Sets
# ==========================================

my_set = {"Ahmed", 100, 100.5, True}

empty_set = set()

empty_dict = {}


# ==========================================
# 2. Duplicate Elements
# ==========================================

my_set = {1, 2, 3, 2, 1, 3}


# ==========================================
# 3. No Indexing / Slicing
# ==========================================

my_set = {1, 2, 3, 4, 5}

# my_set[0]
# my_set[0:3]


# ==========================================
# 4. Hashable Elements
# ==========================================

valid_set = {
    1,
    2.5,
    True,
    "Python",
    (1, 2),
    frozenset({3, 4})
}

# Invalid:
# { [1, 2] }
# { {"name": "Ahmed"} }
# { {1, 2, 3} }


# ==========================================
# 5. Adding Elements
# ==========================================

numbers = {1, 2, 3}

numbers.add(4)

numbers.update([5, 6, 7])


# ==========================================
# 6. Removing Elements
# ==========================================

numbers.remove(7)

numbers.discard(100)

numbers.pop()

numbers.clear()


# ==========================================
# 7. Set Operations
# ==========================================

A = {1, 2, 3, 4}
B = {3, 4, 5, 6}

union = A | B

intersection = A & B

difference = A - B

symmetric_difference = A ^ B


# ==========================================
# 8. Set Methods
# ==========================================

union = A.union(B)

intersection = A.intersection(B)

difference = A.difference(B)

symmetric_difference = A.symmetric_difference(B)


# ==========================================
# 9. Set Relationships
# ==========================================

A = {1, 2, 3}
B = {1, 2}
C = {10, 20}

A.issuperset(B)

B.issubset(A)

A.isdisjoint(C)


# ==========================================
# 10. Frozenset
# ==========================================

my_frozen_set = frozenset({1, 2, 3, 4})


# Cannot modify a frozenset

# my_frozen_set.add(5)
# my_frozen_set.remove(1)


# ==========================================
# 11. Frozenset as Set Element
# ==========================================

nested_set = {
    frozenset({1, 2}),
    frozenset({3, 4})
}


# ==========================================
# 12. Frozenset as Dictionary Key
# ==========================================

data = {
    frozenset({1, 2}): "Group A"
}
```

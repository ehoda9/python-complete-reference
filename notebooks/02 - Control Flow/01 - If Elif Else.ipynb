# Control Flow - If, Elif, Else

Control flow determines which parts of the program should run based on conditions.

| Form | Syntax | Use |
|---|---|---|
| `if` | `if condition:` | Run a block only when the condition is `True` |
| `if - else` | `if condition: ... else: ...` | Choose between two blocks |
| `if - elif - else` | | Choose between several conditions |
| Nested `if` | | An `if` inside another `if` |
| One-line `if` | `if condition: statement` | A one-statement block on one line |
| Ternary | `value_if_true if condition else value_if_false` | Choose between two values |

---

Conditional statements are used to **make decisions**.

---

## `if`

Runs a block of code only when the condition is `True`.

### Syntax

`if condition:`

```
# code
```

---

## `if - else`

* `if` runs when the condition is `True`.
* `else` runs when the condition is `False`.

### Syntax

`if condition:`

```
# code
```

`else:`

```
# code
```

---

## `if - elif - else`

Used when there are multiple conditions.

* `if` checks the first condition.
* `elif` checks another condition if the previous condition was `False`.
* `else` runs if all conditions are `False`.
* Python executes the **first matching condition** and skips the rest.

### Syntax

`if condition1:`

```
# code
```

`elif condition2:`

```
# code
```

`elif condition3:`

```
# code
```

`else:`

```
# code
```

### Important

You can have:

* One `if`
* Multiple `elif`
* Zero or one `else`

---

## Nested `if`

A nested `if` is an `if` statement inside another `if` statement.

Used when a decision depends on another decision.

### Syntax

`if condition1:`

```
`if condition2:`

    `# code`

`else:`

    `# code`
```

`else:`

```
`# code`
```

The inner `if` is checked only if the outer `if` condition is `True`.

---

## One-Line `if`

When the block contains only one statement, Python allows the statement to be written on the same line.

### Syntax

`if condition: statement`

Example:

`if age >= 18: print("Adult")`

This is valid, but multiple statements on one line can make code harder to read.

---

## One-Line `if - else`

An `if - else` can also be written on one line:

`if condition: statement1`

`else: statement2`

This is different from the **ternary conditional operator**.

---

# Ternary Conditional Operator

The ternary conditional operator is used to choose between two values based on a condition.

### Syntax

`value_if_true if condition else value_if_false`

### Example

`status = "Adult" if age >= 18 else "Minor"`

If `age >= 18` is `True` => `"Adult"` is assigned.

Otherwise => `"Minor"` is assigned.

---

## Ternary vs One-Line `if - else`

### One-Line `if - else`

Used to execute statements:

`if age >= 18: print("Adult")`

`else: print("Minor")`

### Ternary

Used to produce/select a value:

`status = "Adult" if age >= 18 else "Minor"`

---

## Nested Ternary

A ternary expression can contain another ternary expression.

Example:

`result = "A" if score >= 90 else "B" if score >= 80 else "C"`

This works, but too many nested ternaries can make code difficult to read.

For complex decisions, normal `if - elif - else` is usually clearer.

---

## Important Notes

* Conditions are evaluated as `True` or `False`.
* Python uses indentation to define the code block.
* `if` can be used without `else`.
* `elif` is checked only when previous conditions were `False`.
* Only the first matching branch in an `if - elif - else` chain runs.
* Nested `if` means an `if` inside another conditional block.
* A ternary returns/selects a value; it is an expression.
* Use normal `if - elif - else` when the logic becomes complicated.


````python
# ----------------------------------
# -- Control Flow / If Statements --
# ----------------------------------

age = 20

# ----------------
# -- if --
# ----------------

if age >= 18:
    print("Adult")


# ----------------
# -- if / else --
# ----------------

if age >= 18:
    print("Adult")
else:
    print("Minor")


# ----------------
# -- if / elif / else --
# ----------------

score = 85

if score >= 90:
    grade = "A"
elif score >= 80:
    grade = "B"
elif score >= 70:
    grade = "C"
else:
    grade = "F"


# ----------------
# -- Nested if --
# ----------------

age = 20
has_id = True

if age >= 18:
    if has_id:
        print("Access allowed")
    else:
        print("ID required")
else:
    print("Underage")


# ----------------
# -- One-Line if --
# ----------------

if age >= 18: print("Adult")


# ----------------
# -- One-Line if / else --
# ----------------

if age >= 18: print("Adult")
else: print("Minor")


# ----------------------------------
# -- Ternary Conditional Operator --
# ----------------------------------

status = "Adult" if age >= 18 else "Minor"

# Another example
number = 10
result = "Even" if number % 2 == 0 else "Odd"


# ----------------
# -- Nested Ternary --
# ----------------

score = 85

grade = "A" if score >= 90 else "B" if score >= 80 else "C"
````

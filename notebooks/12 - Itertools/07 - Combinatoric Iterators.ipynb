# 12.7 - Itertools => Combinatoric Iterators

Combinatoric iterators generate **arrangements** of the input items: pairs, selections, orderings and grids.

## The Four Tools

| Tool | Order matters? | Items can repeat? | Number of results |
|---|---|---|---|
| `product(*iterables, repeat=1)` | Yes | Yes | product of the input sizes (`n ** r` with `repeat=r`) |
| `permutations(iterable, r=None)` | Yes | No | `math.perm(n, r)` |
| `combinations(iterable, r)` | No | No | `math.comb(n, r)` |
| `combinations_with_replacement(iterable, r)` | No | Yes | `math.comb(n + r - 1, r)` |

Examples with `"ABC"` and `r = 2`:

| Tool | Result |
|---|---|
| `product("ABC", repeat=2)` | `AA AB AC BA BB BC CA CB CC` |
| `permutations("ABC", 2)` | `AB AC BA BC CA CB` |
| `combinations("ABC", 2)` | `AB AC BC` |
| `combinations_with_replacement("ABC", 2)` | `AA AB AC BB BC CC` |

## `product`: Nested Loops in One Call

`product(A, B)` is the same as `((a, b) for a in A for b in B)`.

```python
from itertools import product

list(product([1, 2], "ab"))        # [(1, 'a'), (1, 'b'), (2, 'a'), (2, 'b')]
list(product([0, 1], repeat=3))    # all 3-bit binary combinations
```

- `repeat=r` means the same iterable `r` times.
- It reads all inputs into memory first, so the inputs must be **finite**.

## `permutations`: Every Ordering

```python
from itertools import permutations

list(permutations("ABC"))          # all 6 orderings (r defaults to the full length)
list(permutations("ABC", 2))       # 6 ordered pairs
```

## `combinations`: Selections Without Order

```python
from itertools import combinations

list(combinations("ABC", 2))       # [('A', 'B'), ('A', 'C'), ('B', 'C')]
```

Typical uses: every pair from a list, every subset of a fixed size.

## `combinations_with_replacement`

Like `combinations`, but an item may be chosen more than once.

## Order and Uniqueness Rules

- Results follow the **order of the input positions**. Sorted input gives sorted output.
- Items are unique by **position, not by value**. Repeated input values produce repeated results.

```python
list(combinations("AAB", 2))
# [('A', 'A'), ('A', 'B'), ('A', 'B')]     the two 'A' items are different positions
```

## Watch the Size

The number of results grows very fast.

- Check the count with `math.comb()`, `math.perm()` or `n ** r` **before** listing.
- Iterate lazily instead of calling `list()` on large results.

## Key Rules

- Order matters and repeats allowed: `product`.
- Order matters and no repeats: `permutations`.
- Order does not matter and no repeats: `combinations`.
- Order does not matter and repeats allowed: `combinations_with_replacement`.

````python
import math
from itertools import product, permutations, combinations, combinations_with_replacement

items = "ABC"

# The four tools, r = 2
print(["".join(t) for t in product(items, repeat=2)])
print(["".join(t) for t in permutations(items, 2)])
print(["".join(t) for t in combinations(items, 2)])
print(["".join(t) for t in combinations_with_replacement(items, 2)])

# product: a nested loop in one call
print(list(product([1, 2], "ab")))
print(list(product([1, 2], "ab")) == [(a, b) for a in [1, 2] for b in "ab"])
print(list(product([0, 1], repeat=3)))               # 3-bit binary combinations

# permutations: r defaults to the full length
print(len(list(permutations("ABC"))))                # 6

# combinations: every pair from a list
print(list(combinations(["Ann", "Bob", "Cy"], 2)))

# Counts match the math formulas
n, r = 5, 3
print(len(list(product(range(n), repeat=r))) == n ** r)
print(len(list(permutations(range(n), r))) == math.perm(n, r))
print(len(list(combinations(range(n), r))) == math.comb(n, r))
print(len(list(combinations_with_replacement(range(n), r))) == math.comb(n + r - 1, r))

# Uniqueness is by POSITION, not by value
print(list(combinations("AAB", 2)))

# Sorted input gives sorted output
print(list(combinations([1, 2, 3, 4], 2)))

# Parameter grid: every combination of settings
learning_rates = [0.1, 0.01]
depths = [3, 5]
for lr, depth in product(learning_rates, depths):
    print(f"lr={lr}, depth={depth}")

# Check the size BEFORE listing
print(math.perm(10, 5))                              # 30240 results: fine
print(math.perm(20, 10))                             # 670442572800 results: never call list()
````

# File Handling

File handling is used to create, read, write, and update files using Python.

Python commonly uses the `open()` function to work with files.

## `open()`

### Syntax

`open(file, mode)`

Example:

`file = open("data.txt", "r")`

The `mode` determines what Python is allowed to do with the file.

---

## File Modes

| Mode  | Name   | Purpose                | File Must Exist? | Creates File? |
| ----- | ------ | ---------------------- | ---------------- | ------------- |
| `"r"` | Read   | Read file contents     | Yes              | No            |
| `"w"` | Write  | Write new content      | No               | Yes           |
| `"a"` | Append | Add content to the end | No               | Yes           |
| `"x"` | Create | Create a new file      | No               | Yes           |

### `"r"` => Read

* Opens the file for reading.
* This is the default mode.
* Gives `FileNotFoundError` if the file does not exist.
* Does not allow writing.

### `"w"` => Write

* Opens the file for writing.
* Creates the file if it does not exist.
* If the file already exists, its previous content is **overwritten**.

### `"a"` => Append

* Opens the file for adding content.
* Creates the file if it does not exist.
* New content is added to the **end** of the file.
* Existing content is preserved.

### `"x"` => Create

* Creates a new file.
* Gives `FileExistsError` if the file already exists.
* Useful when you want to make sure you do not overwrite an existing file.

---

## Closing a File

When using `open()` directly, close the file when you finish working with it.

`file = open("data.txt", "r")`

`file.close()`

A better approach is using `with`.

---

## `with open()`

The `with` statement automatically closes the file after the block finishes.

### Syntax

`with open("data.txt", "r") as file:`

```
`# work with the file`
```

This is the recommended way to work with files.

---

## Reading a File

### `read()`

Reads the entire file.

`file.read()`

### `readline()`

Reads one line.

`file.readline()`

### `readlines()`

Reads all lines and returns them as a list.

`file.readlines()`

### Iterating Through a File

A file can be used as an iterable.

`for line in file:`

```
`# process line`
```

---

## Writing to a File

### `write()`

Writes a string to the file.

`file.write("Hello Python")`

### `writelines()`

Writes multiple strings from an iterable.

`file.writelines(["Hello\n", "Python\n"])`

`writelines()` does **not** automatically add new lines.

---

## Important Notes

* `open()` returns a file object.
* `"r"` is the default mode.
* `"w"` can overwrite existing content.
* `"a"` preserves existing content and adds to the end.
* `"x"` fails if the file already exists.
* Use `with open(...)` when possible.
* Text files are usually handled as strings.
* `read()` returns a string.
* `readlines()` returns a list of lines.
* `write()` expects a string.
* Always consider the file encoding when working with text files.


````python
# -------------------
# -- File Handling --
# -------------------

# -------------------
# -- Write --
# -------------------

with open("data.txt", "w") as file:
    file.write("Hello Python\n")
    file.write("File Handling\n")


# -------------------
# -- Read --
# -------------------

with open("data.txt", "r") as file:
    content = file.read()


# -------------------
# -- Read Line --
# -------------------

with open("data.txt", "r") as file:
    first_line = file.readline()


# -------------------
# -- Read Lines --
# -------------------

with open("data.txt", "r") as file:
    lines = file.readlines()


# -------------------
# -- Iterate --
# -------------------

with open("data.txt", "r") as file:
    for line in file:
        pass


# -------------------
# -- Append --
# -------------------

with open("data.txt", "a") as file:
    file.write("New line\n")


# -------------------
# -- Create --
# -------------------

with open("new_file.txt", "x") as file:
    file.write("New file")


# -------------------
# -- writelines() --
# -------------------

lines = [
    "Python\n",
    "Java\n",
    "C++\n"
]

with open("languages.txt", "w") as file:
    file.writelines(lines)
````

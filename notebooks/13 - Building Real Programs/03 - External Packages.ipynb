# Modules => External Packages

Python has a large ecosystem of **third-party packages** that are not included in the Python standard library.

| Task | Command |
|---|---|
| Install | `pip install package_name` |
| Install a version | `pip install requests==2.32.0` |
| Upgrade | `pip install --upgrade requests` |
| Uninstall | `pip uninstall requests` |
| List installed packages | `pip list` |
| Show one package | `pip show requests` |
| Inside Jupyter | `%pip install requests` |
| Create a virtual environment | `python -m venv .venv` |
| Save / restore the packages | `pip freeze > requirements.txt` / `pip install -r requirements.txt` |

---

Examples:

* `requests` => HTTP requests
* `numpy` => numerical computing
* `pandas` => data analysis
* `matplotlib` => visualization
* `opencv-python` => computer vision
* `flask` => web applications
* `django` => web applications

You usually install third-party packages using `pip`.

---

## Install a Package

In a terminal:

`pip install package_name`

Example:

`pip install requests`

After installation:

`import requests`

---

## Install a Specific Version

`pip install requests==2.32.0`

This installs the specified version if available.

---

## Upgrade a Package

`pip install --upgrade requests`

---

## Uninstall a Package

`pip uninstall requests`

---

## Check Installed Packages

`pip list`

or:

`pip show requests`

---

## Import After Installation

Installing and importing are two different steps:

1. **Install** the package using `pip`.
2. **Import** it in your Python code.

Example:

`pip install requests`

Then:

`import requests`

---

## Jupyter Notebook

Inside Jupyter, you can run installation commands using:

`%pip install requests`

This is often preferable in notebooks because it targets the Python environment associated with the current kernel.

Then:

`import requests`

---

## Package vs Module

A **module** is typically a single `.py` file.

A **package** is a collection of modules organized together.

A package can contain multiple modules and subpackages.

### Important

Not every installed package is imported using exactly the same name as its installation name.

Example:

`pip install opencv-python`

but:

`import cv2`

So the **pip package name** and the **Python import name** can be different.

---

## Standard Library vs External Packages

| Type                | Installation                 | Example                  |
| ------------------- | ---------------------------- | ------------------------ |
| Standard Library    | No separate installation     | `math`, `random`, `json` |
| Third-party Package | Usually installed with `pip` | `requests`, `numpy`      |
| Custom Module       | Created by you               | `math_tools.py`          |

### Important Notes

* `pip` is used to install Python packages.
* Installing a package does not automatically import it.
* Use `import` inside Python code.
* `%pip install ...` is useful in Jupyter.
* External packages may have dependencies of their own.
* Virtual environments are commonly used to keep project dependencies isolated.

## Virtual Environments (`venv`)

A **virtual environment** is a folder with its own Python and its own installed packages. Each project can have its own versions without affecting the others.

| Task | Command |
|---|---|
| Create one | `python -m venv .venv` |
| Activate (Windows, cmd) | `.venv\Scripts\activate` |
| Activate (Windows, PowerShell) | `.venv\Scripts\Activate.ps1` |
| Activate (Git Bash on Windows) | `source .venv/Scripts/activate` |
| Activate (macOS and Linux) | `source .venv/bin/activate` |
| Leave it | `deactivate` |
| Save the installed packages | `pip freeze > requirements.txt` |
| Install them again | `pip install -r requirements.txt` |
| Use the `pip` of the current Python | `python -m pip install package_name` |

* Create the environment once per project and activate it every time you open a new terminal.
* Do not commit the environment folder. Commit `requirements.txt` instead.
* To remove an environment, delete its folder.

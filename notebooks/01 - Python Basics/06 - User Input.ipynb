# User Input

The `input()` function is used to get data from the user while the program is running.

| Tool | Purpose |
|---|---|
| `input()` | Read a line of text. The program waits for `Enter` |
| `input("message")` | Show a prompt first |
| Return type | Always a `str` |
| `int(input())` / `float(input())` | Convert the text to a number |
| `input().split()` | Read several values on one line |

---

## Basic Syntax

`input()`

The program waits for the user to enter something and press `Enter`.

## Input With a Message

`input("Enter your name: ")`

The text inside `input()` is called the **prompt**.

## Important

`input()` always returns a `str`.

Even if the user enters a number, Python receives it as a string.

Example:

`age = input("Enter your age: ")`

If the user enters `20`:

`age` contains `"20"`, not `20`.

## Type Conversion With Input

Use type conversion when you need a specific data type.

| Input | Example | Result Type |
|---|---|---|
| Text | `input()` | `str` |
| Integer | `int(input())` | `int` |
| Decimal | `float(input())` | `float` |
| Boolean | `bool(input())` | `bool` |

### Common Examples

`name = input("Enter your name: ")`

`age = int(input("Enter your age: "))`

`price = float(input("Enter the price: "))`

## Multiple Values

`input().split()` can be used to receive multiple values.

Example:

`names = input().split()`

If the user enters:

`Ahmed Mona Ali`

The result is:

`["Ahmed", "Mona", "Ali"]`

### Multiple Numbers

Because `split()` still returns strings, convert each value to `int` when needed.

`numbers = [int(x) for x in input().split()]`

## Important Notes

- `input()` pauses the program until the user presses `Enter`.
- The return value of `input()` is always a string.
- Use `int()` or `float()` when numeric input is required.
- `input().split()` separates input into multiple strings.
- Invalid conversion can raise `ValueError`.

```python
# ----------------
# -- User Input --
# ----------------

# Basic input
name = input("Enter your name: ")

# Input is always a string
age = input("Enter your age: ")

# Convert input to int
age = int(input("Enter your age: "))

# Convert input to float
price = float(input("Enter the price: "))

# Multiple strings
names = input("Enter names: ").split()

# Multiple integers
numbers = [int(x) for x in input("Enter numbers: ").split()]
```

# Operators

Operators combine values. Each family has its own section below.

| Family | Operators | Use |
|---|---|---|
| Arithmetic | `+` `-` `*` `/` `//` `%` `**` | Calculations |
| Assignment | `=` `+=` `-=` `*=` `/=` and more | Store and update values |
| Comparison | `==` `!=` `<` `>` `<=` `>=` | Compare values (can be chained) |
| Logical | `and` `or` `not` | Combine conditions |
| Identity | `is` `is not` | Same object? |
| Membership | `in` `not in` | Is a value inside a container? |
| Bitwise | `&` `\|` `^` `~` `<<` `>>` | Work on the binary digits of integers |

---

# Arithmetic Operators

Arithmetic operators are used to perform mathematical operations.

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `10 + 5` | `15` |
| `-` | Subtraction | `10 - 5` | `5` |
| `*` | Multiplication | `10 * 5` | `50` |
| `/` | Division | `10 / 5` | `2.0` |
| `%` | Modulus | `10 % 3` | `1` |
| `**` | Exponentiation | `2 ** 3` | `8` |
| `//` | Floor Division | `10 // 3` | `3` |

---

The `+` operator can also be used with other compatible types, such as strings and lists.

---

### Operator Precedence

Multiplication is performed before addition.

`5 + 10 * 100` => `1005`

Use parentheses to change the order:

`(5 + 10) * 100` => `1500`

`*` can also repeat sequences:

`"Hi" * 3` => `"HiHiHi"`

---

### Important notes

The `/` operator always returns a `float`.

`100 / 20` => `5.0`

Even when the result is mathematically an integer:

`10 / 2` => `5.0`

Division by zero raises `ZeroDivisionError`.

### Important Difference

`/` returns the normal division result.

`//` returns the floor-divided result.

Floor division rounds toward negative infinity, not simply toward zero.

---
### Arithmetic with Different Numeric Types

Python can perform arithmetic between compatible numeric types.

Examples:

- `int + int` => `int`
- `int + float` => `float`
- `float + float` => `float`
- `int + complex` => `complex`

### Division by Zero

`10 / 0` => `ZeroDivisionError`

`10 // 0` => `ZeroDivisionError`

`10 % 0` => `ZeroDivisionError`

```python
# ==========================================
# 1. Arithmetic Operators
# ==========================================


# ------------------------------------------
# Addition +
# ------------------------------------------

print(10 + 30)          # + Addition
print(-30 - 20)         # Subtraction -
print(10 * 3)           # Multiplication *
print(100 / 20)         # Division / always returns float

# Multiplication has higher precedence than addition
# Parentheses change the order
print(5 + 10 * 2)        # 25
print((5 + 10) * 2)      # 30

print(9 % 2)            # Modulus %

# Check even / odd
print(10 % 2 == 0)      # True
print(7 % 2 == 0)       # False


# ------------------------------------------
# Exponentiation **
print(2 ** 5)            # 32
print(2 * 2 * 2 * 2 * 2) # 32

# ------------------------------------------
print(100 // 20)         # Floor Division //
# Difference between / and //
print(7 / 2)             # 3.5
print(7 // 2)            # 3

# Floor means toward negative infinity
print(-7 // 2)           # -4


# ------------------------------------------
# Arithmetic with different numeric types
# ------------------------------------------

print(10 + 2.5)          # 12.5
print(type(10 + 2.5))    # float

print(10 + 2j)           # (10+2j)
print(type(10 + 2j))     # complex


# ------------------------------------------
# Operators with sequences
# ------------------------------------------

print("Hi" * 3)          # HiHiHi
print([1, 2] * 3)        # [1, 2, 1, 2, 1, 2]
```

# Assignment Operators

Assignment operators are used to assign a value to a variable or update its current value.

| Operator | Name | Example | Equivalent To |
|---|---|---|---|
| `=` | Assignment | `x = 10` | Assign `10` to `x` |
| `+=` | Add and Assign | `x += 5` | `x = x + 5` |
| `-=` | Subtract and Assign | `x -= 5` | `x = x - 5` |
| `*=` | Multiply and Assign | `x *= 5` | `x = x * 5` |
| `/=` | Divide and Assign | `x /= 5` | `x = x / 5` |
| `**=` | Exponentiate and Assign | `x **= 2` | `x = x ** 2` |
| `%=` | Modulus and Assign | `x %= 3` | `x = x % 3` |
| `//=` | Floor Divide and Assign | `x //= 3` | `x = x // 3` |

### Important Notes

- `=` assigns a value to a variable.
- The other operators perform an operation and then store the result back in the variable.
- `/=` usually changes the result to `float` because `/` always performs true division.
- `//=` uses floor division.
- Assignment operators can be used with compatible data types.

### Example

If:

`x = 10`

Then:

`x += 5`

means:

`x = x + 5`

So `x` becomes `15`.

```python
# --------------------------
# -- Assignment Operators --
# --------------------------

x = 10

x += 5      # x = x + 5
x -= 2      # x = x - 2
x *= 3      # x = x * 3
x /= 2      # x = x / 2
x **= 2     # x = x ** 2
x %= 3      # x = x % 3
x //= 2     # x = x // 2
```

# Comparison Operators

Comparison operators are used to compare two values.

The result is always a Boolean value: `True` or `False`.

| Operator | Name | Example | Result |
|---|---|---|---|
| `==` | Equal | `10 == 10` | `True` |
| `!=` | Not Equal | `10 != 5` | `True` |
| `>` | Greater Than | `10 > 5` | `True` |
| `<` | Less Than | `10 < 5` | `False` |
| `>=` | Greater Than or Equal | `10 >= 10` | `True` |
| `<=` | Less Than or Equal | `10 <= 5` | `False` |

### Important Notes

- `==` checks if two values are equal.
- `=` is **assignment**, not comparison.
- `!=` checks if two values are different.
- `>=` means greater than **or equal to**.
- `<=` means less than **or equal to**.
- Comparison operators return `True` or `False`.
- They are commonly used with `if`, `while`, and logical operators.

```python
# --------------------------
# -- Comparison Operators --
# --------------------------

x = 10
y = 5

x == y
x != y
x > y
x < y
x >= y
x <= y

# Comparison with variables
age = 20

age >= 18
age < 18

# Strings can also be compared
"Python" == "Python"
"Python" != "Java"
```

# Chained Comparisons

Comparison operators can be **chained**. `a < b < c` means `a < b and b < c`, and `b` is evaluated only once.

| Chain | Means |
|---|---|
| `1 < x < 10` | `1 < x and x < 10` |
| `a == b == c` | `a == b and b == c` |
| `0 <= i < len(items)` | `i` is a valid index |
| `10 > x >= 5` | `10 > x and x >= 5` |
| `a < b > c` | Legal, but confusing. Avoid it |

---

### Important Notes

* The chain stops at the first false comparison (short-circuit), so later parts are not evaluated.
* Each middle value is evaluated **once**.
* Chains can mix `<`, `<=`, `>`, `>=`, `==`, `!=`, `is` and `in`. Keep them readable.

```python
x = 5

print(1 < x < 10)              # True
print(1 < x < 3)               # False
print(0 <= x <= 5)             # True
print(10 > x >= 5)             # True
print(3 == 3 == 3, 3 == 3 == 4)

# The same check written out with and
print(1 < x and x < 10)

# Short-circuit: the second comparison is never evaluated when the first one is false
print(5 < 3 < 1 / 0)           # False, and no ZeroDivisionError
```

# Logical Operators

Logical operators combine conditions.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `and` | True if both sides are true | `True and False` | `False` |
| `or` | True if at least one side is true | `True or False` | `True` |
| `not` | Reverses the truth value | `not True` | `False` |

---

### Important Notes

* `and` and `or` **short-circuit**: Python stops as soon as the result is known.
* `and` and `or` return one of their **operands**, not always `True` or `False`.
  * `a and b` returns `a` if `a` is falsy, otherwise `b`.
  * `a or b` returns `a` if `a` is truthy, otherwise `b`.
* Precedence: `not` first, then `and`, then `or`. Use parentheses when in doubt.

```python
print(True and False, True or False, not True)
print(5 > 3 and 2 < 4, 5 > 3 and 2 > 4)

# or returns the first truthy operand, and returns the first falsy operand (or the last one)
print(0 or "default", "" or "empty", "text" or "unused")
print(0 and "unused", "a" and "b")
print(not 0, not "text", not [])

# Short-circuit: the right side is not evaluated when the left side decides the result
print(True or 1 / 0)
print(False and 1 / 0)

# Precedence: not, then and, then or
print(not True or True, not (True or True))
```

# Identity Operators

Identity operators check whether two names refer to the **same object**.

| Operator | Meaning | Example |
|---|---|---|
| `is` | Same object | `a is b` |
| `is not` | Different objects | `a is not b` |

---

### `is` vs `==`

| Operator | Compares |
|---|---|
| `==` | **Values**: are they equal? |
| `is` | **Identity**: is it the very same object? |

---

### Important Notes

* Use `is` and `is not` for `None`, `True` and `False`: `value is None`.
* Do not use `is` to compare numbers or strings. Equal values can be different objects.
* `id(obj)` returns a number that identifies an object while it exists.

```python
a = [1, 2, 3]
b = a                # the same object
c = [1, 2, 3]        # an equal but separate object

print(a == c, a is c)
print(a is b, id(a) == id(b))
print(a is not c)

value = None
print(value is None, value is not None)
```

# Membership Operators

Membership operators are used to check whether a value exists inside a sequence or collection.

They return a Boolean value: `True` or `False`.

| Operator | Name           | Example              | Result |
| -------- | -------------- | -------------------- | ------ |
| `in`     | Membership     | `2 in [1, 2, 3]`     | `True` |
| `not in` | Not Membership | `5 not in [1, 2, 3]` | `True` |

## `in`

Checks whether a value exists inside a collection.

Example:

`"Python" in ["Python", "Java", "C++"]`

=> `True`

## `not in`

Checks whether a value does **not** exist inside a collection.

Example:

`"Java" not in ["Python", "C++"]`

=> `True`

## With Different Data Types

### Lists

`"Python" in ["Python", "Java"]`

### Tuples

`10 in (5, 10, 15)`

### Sets

`10 in {5, 10, 15}`

### Strings

`"Py" in "Python"`

### Dictionaries

Membership checks **keys**, not values.

`"name" in {"name": "Mahmoud", "age": 20}`

=> `True`

`"Mahmoud" in {"name": "Mahmoud", "age": 20}`

=> `False`

To check dictionary values:

`"Mahmoud" in {"name": "Mahmoud", "age": 20}.values()`

=> `True`

## Important Notes

* `in` and `not in` return `True` or `False`.
* They are commonly used with `if` and `while`.
* With strings, they check whether a substring exists.
* With dictionaries, they check **keys** by default.
* `not in` is the opposite of `in`.


```python
# --------------------------
# -- Membership Operators --
# --------------------------

# Lists
"Python" in ["Python", "Java", "C++"]
"PHP" not in ["Python", "Java", "C++"]

# Tuples
10 in (5, 10, 15)
20 not in (5, 10, 15)

# Sets
10 in {5, 10, 15}
20 not in {5, 10, 15}

# Strings
"Py" in "Python"
"Java" not in "Python"

# Dictionaries
user = {
    "name": "Mahmoud",
    "age": 20
}

"name" in user
"email" not in user

# Check dictionary values
"Mahmoud" in user.values()

# Using membership with if
name = "Mahmoud"

if name in ["Ahmed", "Mahmoud", "Mona"]:
    print("Name found")
```

# Bitwise Operators

Bitwise operators work on the **binary digits** of integers.

| Operator | Name | Example | Result |
|---|---|---|---|
| `&` | AND | `0b1100 & 0b1010` | `0b1000` (8) |
| `\|` | OR | `0b1100 \| 0b1010` | `0b1110` (14) |
| `^` | XOR | `0b1100 ^ 0b1010` | `0b0110` (6) |
| `~` | NOT | `~5` | `-6` |
| `<<` | Shift left | `1 << 3` | `8` |
| `>>` | Shift right | `16 >> 2` | `4` |

---

### Important Notes

* `x << n` multiplies by `2 ** n`. `x >> n` divides by `2 ** n` and rounds down.
* `~x` is the same as `-x - 1`.
* `0b1010` is a binary literal. `0o12` is octal and `0xA` is hexadecimal.
* The same symbols `&`, `|` and `^` are used for set operations.

```python
a, b = 0b1100, 0b1010          # 12 and 10

print(a & b, a | b, a ^ b)      # 8 14 6
print(~5, ~-1)                  # -6 0
print(1 << 3, 16 >> 2, 5 << 1)  # 8 4 10
print(0b1010 == 0o12 == 0xA == 10)
```

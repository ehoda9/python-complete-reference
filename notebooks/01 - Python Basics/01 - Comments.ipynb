# Comments

Comments are notes for people. Python ignores them.

| Kind | Syntax | Notes |
|---|---|---|
| Single-line comment | `# note` | Starts with `#` |
| Inline comment | `x = 5  # note` | After code, on the same line |
| Several lines | one `#` per line | Python has no dedicated multi-line comment syntax |
| Commenting out code | `# print("skip")` | Stops a line from running without deleting it |

---

### Why use comments?

1. **Information about the file**
   - Explain what the file or section is about.

2. **Usage instructions**
   - Explain how to use or run specific code.

3. **License / History**
   - Add license information, author, version, or change history.

4. **Explain code**
   - Add notes to make code easier to understand and maintain.

5. **Temporarily comment out code**
   - Prevent a line of code from running without deleting it.
  
### Important notes

- **Single-line comment:** starts with `#`
- **Inline comment:** comment after code
- Python does **not** have a dedicated multiline comment syntax.

```python
# This prints "Hello from notebook"
print("Hello from notebook")  # Inline comment

# print("Hello from notebook")
# This code will not run because it is commented out.
```

# Variables

A variable is a name that refers to a value.

| Topic | Rule |
|---|---|
| Assignment | `name = value` |
| Start with | A letter (`a-z`, `A-Z`) or `_`, never a digit |
| Can contain | Letters, digits and `_`. No spaces or special characters |
| Case | Case-sensitive: `name`, `Name` and `NAME` are different |
| Keywords | Cannot be used as names (`help("keywords")`) |
| What a variable is | A name, not the object. The type belongs to the object |

---

### What is a Variable?

A variable is a name that refers to an object/value.

### Syntax
`[Variable Name] [Assignment Operator] [Value]`



### Variable Naming Rules
1. Can start with: Letters: `a-z`, `A-Z` or Underscore: `_`
2. Cannot start with a number `1name`.
3. Can contain: Letters & Numbers: `0-9` & Underscores: `_`
4. Cannot contain spaces or special characters.
5. Variable names are **case-sensitive**.
   - `name` != `Name` != `NAME`
6. Cannot use Python **keywords** as variable names.
   - Reserved Words : `help("keywords")`
### Important Notes
- A variable is a **name/reference**, not the object itself.
- Assignment does not require declaring the data type.
- The type belongs to the object, not the variable.
- A variable can refer to objects of different types during its lifetime.

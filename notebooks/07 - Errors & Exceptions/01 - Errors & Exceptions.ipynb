## Errors & Exceptions => Raising Exceptions

* An **exception** is an error that occurs while a program is running.
* Exceptions interrupt the normal flow of the program.
* Python provides an error message that helps you understand what went wrong.
* A **traceback** shows where the error occurred and points to the line that caused it.
* Exceptions have different types, such as:

  * `SyntaxError`
  * `NameError`
  * `TypeError`
  * `ValueError`
  * `IndexError`
  * `KeyError`
  * `ZeroDivisionError`
  * `FileNotFoundError`
* Python provides many built-in exception types.
* The `raise` keyword is used to raise an exception manually.

### Common Exception Types

| Exception           | Usually happens when                         |
| ------------------- | -------------------------------------------- |
| `SyntaxError`       | Python code has invalid syntax               |
| `NameError`         | A name/variable does not exist               |
| `TypeError`         | An operation uses an inappropriate type      |
| `ValueError`        | The type is correct but the value is invalid |
| `IndexError`        | A sequence index does not exist              |
| `KeyError`          | A dictionary key does not exist              |
| `ZeroDivisionError` | Dividing by zero                             |
| `FileNotFoundError` | A requested file does not exist              |

### `raise`

`raise` allows you to create an exception yourself when a condition is invalid.

```text
raise ValueError("Invalid value")
```

This immediately raises the exception and stops normal execution unless it is handled.


````python
# raise: manually raises an exception with a custom message
age = -5

if age < 0:
    raise ValueError("Age cannot be negative")
````

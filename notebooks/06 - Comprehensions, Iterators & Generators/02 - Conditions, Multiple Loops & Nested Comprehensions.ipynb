# Comprehensions => Conditions, Multiple Loops & Nested Comprehensions

Every comprehension pattern is a small variation of `[expression for item in iterable if condition]`.

| Pattern | Syntax |
|---|---|
| Filter | `[x for x in it if cond]` |
| Choose a value | `[a if cond else b for x in it]` |
| Several conditions | `[x for x in it if c1 and c2]` |
| Several loops | `[(x, y) for x in xs for y in ys]` |
| Flatten | `[x for row in rows for x in row]` |
| Nested (list of lists) | `[[f(x) for x in row] for row in rows]` |
| Transpose | `[list(col) for col in zip(*rows)]` |

---

## Filtering With `if`

An `if` placed **after** the `for` clause is a filter. Items that fail the test are dropped.

```python
[n for n in nums if n > 0]
```

## Choosing a Value With `if / else`

An `if / else` placed **before** the `for` clause is a conditional expression. It picks a value for every item.

```python
["even" if n % 2 == 0 else "odd" for n in nums]
```

| Goal | Where the condition goes | Example |
|---|---|---|
| Drop some items | After the `for` | `[n for n in nums if n > 0]` |
| Choose between two values | Before the `for` | `["even" if n % 2 == 0 else "odd" for n in nums]` |
| Both | Both places | `["even" if n % 2 == 0 else "odd" for n in nums if n > 0]` |

**Rule:** `else` is only allowed in the conditional expression. A trailing filter can never have an `else`.

## Multiple Conditions

```python
[n for n in range(30) if n % 2 == 0 and n % 3 == 0]
[n for n in range(30) if n % 2 == 0 if n % 3 == 0]   # chained if = and
```

Prefer `and` / `or`. It is easier to read.

## Multiple `for` Clauses

Several `for` clauses work like **nested loops**, from left to right.

```python
[(x, y) for x in [1, 2, 3] for y in "ab"]
```

is the same as:

```python
result = []
for x in [1, 2, 3]:
    for y in "ab":
        result.append((x, y))
```

The first `for` is the **outer** loop. The last `for` is the **inner** loop.

## Flattening a Nested Iterable

```python
matrix = [[1, 2, 3], [4, 5, 6]]
[x for row in matrix for x in row]   # [1, 2, 3, 4, 5, 6]
```

Read it as: `for row in matrix: for x in row: x`.

## Nested Comprehensions

A comprehension can be the **expression** of another comprehension. This builds a list of lists.

```python
[[0 for _ in range(3)] for _ in range(2)]        # 2 rows, 3 columns
[[i * j for j in range(1, 4)] for i in range(1, 4)]   # multiplication table
```

## Multiple `for` vs Nested Comprehension

| Form | Result | Example |
|---|---|---|
| Multiple `for` in one comprehension | One flat list | `[x for row in m for x in row]` |
| Comprehension inside a comprehension | List of lists | `[[x * 2 for x in row] for row in m]` |

## Transposing a Matrix

```python
[[row[i] for row in matrix] for i in range(len(matrix[0]))]
[list(col) for col in zip(*matrix)]   # same result
```

## The Aliasing Trap

`[[0] * 3] * 3` does **not** create three rows. It repeats the **same** inner list three times.

```python
bad = [[0] * 3] * 3
bad[0][0] = 99        # every row changes

good = [[0] * 3 for _ in range(3)]
good[0][0] = 99       # only the first row changes
```

## Keep It Readable

- Use at most two `for` clauses in one comprehension.
- Use at most two nesting levels.
- If it needs more, use a normal loop or a helper function.

````python
nums = [-3, -2, -1, 0, 1, 2, 3]

# Filter: if at the END drops items
print([n for n in nums if n > 0])

# Conditional expression: if / else at the START chooses a value
print(["even" if n % 2 == 0 else "odd" for n in nums])

# Both together
print(["even" if n % 2 == 0 else "odd" for n in nums if n > 0])

# Multiple conditions
print([n for n in range(30) if n % 2 == 0 and n % 3 == 0])
print([n for n in range(30) if n % 2 == 0 if n % 3 == 0])   # chained if = and

# Multiple for clauses = nested loops, left to right
pairs = [(x, y) for x in [1, 2, 3] for y in "ab"]
print(pairs)

loop_pairs = []
for x in [1, 2, 3]:
    for y in "ab":
        loop_pairs.append((x, y))
print(pairs == loop_pairs)

# Unique pairs with a condition
print([(i, j) for i in range(4) for j in range(4) if i < j])

# Flatten a matrix
matrix = [[1, 2, 3], [4, 5, 6], [7, 8, 9]]
print([x for row in matrix for x in row])

# Build a matrix with a nested comprehension
print([[0 for _ in range(3)] for _ in range(2)])
print([[i * j for j in range(1, 4)] for i in range(1, 4)])    # multiplication table

# Transpose
print([[row[i] for row in matrix] for i in range(len(matrix[0]))])
print([list(col) for col in zip(*matrix)])                     # same result

# Multiple for vs nested comprehension
print([x * 2 for row in matrix for x in row])                  # flat
print([[x * 2 for x in row] for row in matrix])                # list of lists

# The aliasing trap: [[0] * 3] * 3 repeats the SAME inner list
bad = [[0] * 3] * 3
bad[0][0] = 99
print(bad)                       # every row changed

good = [[0] * 3 for _ in range(3)]
good[0][0] = 99
print(good)                      # only the first row changed
````

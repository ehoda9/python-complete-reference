# Generators

A **generator** is a special type of iterator that produces values **one at a time** instead of creating and storing all values at once.

Generators are created using a function containing the `yield` keyword.

---

## Generator Function

A generator function looks like a normal function, but uses:

`yield`

instead of, or in addition to, `return`.

### Syntax

`def generator_function():`

```
`yield value`
```

Calling the generator function does **not** immediately execute its body.

Instead, it returns a **generator object**.

---

## `yield` vs `return`

| `return`                                 | `yield`                                |
| ---------------------------------------- | -------------------------------------- |
| Ends the function                        | Pauses the function                    |
| Returns one final result                 | Produces a value                       |
| Function execution is finished           | Execution can resume later             |
| Calling again starts a new function call | `next()` resumes from where it stopped |

---

## Using `next()`

`next(generator)`

gets the next value from the generator.

After yielding a value, the generator pauses.

The next `next()` call resumes execution **from the point where the previous `yield` occurred**.

It does not start from the beginning.

---

## Generator Execution

Consider:

`def numbers():`

```
`yield 1`

`yield 2`

`yield 3`
```

When we call:

`generator = numbers()`

the function body does not start yet.

Then:

`next(generator)` => `1`

The generator pauses after `yield 1`.

`next(generator)` => `2`

It resumes from where it paused.

`next(generator)` => `3`

After the final `yield`, the generator is exhausted.

Another `next()` raises:

`StopIteration`

---

## Multiple `yield`s

A generator function can contain one or many `yield` statements.

Each `yield` can produce another value.

Example:

`def numbers():`

```
`yield 1`

`yield 2`

`yield 3`
```

Each call to `next()` continues execution until the next `yield`.

---

## Generators and `for`

Generators are iterators, so they can be used directly with a `for` loop.

`for number in numbers():`

```
`pass`
```

The `for` loop automatically calls `next()` and handles `StopIteration`.

---

## Lazy Evaluation

Generators use **lazy evaluation**.

They produce values only when they are requested.

This means they can save memory when working with large amounts of data.

### Example

A list:

`numbers = [x for x in range(1000000)]`

creates and stores all values.

A generator:

`numbers = (x for x in range(1000000))`

produces values when needed.

---

## Generator Expression

A generator expression looks similar to a list comprehension, but uses parentheses.

List comprehension:

`[x * 2 for x in range(5)]`

Generator expression:

`(x * 2 for x in range(5))`

The list comprehension creates a list immediately.

The generator expression creates values lazily.

---

## Important Notes

* A generator function contains `yield`.
* Calling a generator function does not execute its body immediately.
* It returns a generator object.
* `next()` resumes execution from the previous `yield`.
* A generator can have multiple `yield` statements.
* Generators are iterators.
* Generators are consumed as they are iterated.
* Once exhausted, they cannot be restarted; create a new generator if needed.
* `for` automatically handles `StopIteration`.
* Generators are useful for large data and memory-efficient processing.
* A generator usually uses less memory than creating the complete collection at once.


````python
# ----------------
# -- Generators --
# ----------------

# ---------------------------------
# -- Basic Generator --
# ---------------------------------

# yield => pause function and produce a value
def numbers():
    yield 1
    yield 2
    yield 3


# Calling the function does NOT execute its body
generator = numbers()

# next(generator) => resume until the next yield
next(generator)
next(generator)
next(generator)

# No more values => StopIteration
# next(generator)


# ---------------------------------
# -- Multiple yield Statements --
# ---------------------------------

def generate_numbers():
    yield 10
    yield 20
    yield 30
    yield 40

generator = generate_numbers()

next(generator)
next(generator)
next(generator)
next(generator)


# ---------------------------------
# -- Generator + for Loop --
# ---------------------------------

def count_to_five():
    yield 1
    yield 2
    yield 3
    yield 4
    yield 5

for number in count_to_five():
    pass


# ---------------------------------
# -- yield Resumes From Its Position --
# ---------------------------------

def demo():
    yield "First"

    # Execution resumes here after the first next()
    yield "Second"

    # Execution resumes here after the second next()
    yield "Third"

generator = demo()

next(generator)  # First
next(generator)  # Second
next(generator)  # Third


# ---------------------------------
# -- Generator With a Loop --
# ---------------------------------

def count_to(number):
    current = 1

    while current <= number:
        yield current
        current += 1

generator = count_to(5)

next(generator)
next(generator)

for number in generator:
    pass


# ---------------------------------
# -- Generator Expression --
# ---------------------------------

# List comprehension => creates the complete list
numbers_list = [x * 2 for x in range(5)]

# Generator expression => produces values lazily
numbers_generator = (x * 2 for x in range(5))

next(numbers_generator)
next(numbers_generator)

# Consume remaining values
for number in numbers_generator:
    pass


# ---------------------------------
# -- Generator Is an Iterator --
# ---------------------------------

generator = count_to(3)

iter(generator) is generator

next(generator)
next(generator)
next(generator)


# ---------------------------------
# -- Generator With return --
# ---------------------------------

def example():
    yield 1
    yield 2
    return "Finished"

generator = example()

next(generator)
next(generator)

# The return value becomes the StopIteration value
# next(generator)  # StopIteration: Finished
````

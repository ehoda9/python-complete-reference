# While Loop

A `while` loop is used to repeat a block of code **as long as a condition is `True`**.

### Syntax

`while condition:`

```
# code
```

The condition is checked **before every iteration**.

* If the condition is `True` => the code runs.
* If the condition is `False` => the loop stops.
* If the condition is initially `False` => the loop does not run at all.

---

## How a `while` Loop Works

1. Check the condition.
2. If it is `True`, run the loop body.
3. Update the value that affects the condition.
4. Check the condition again.
5. Repeat until the condition becomes `False`.

---

## Important

The loop should normally contain something that eventually makes the condition `False`.

Otherwise, the loop may become an **infinite loop**.

Example:

`x = 1`

`while x <= 5:`

```
`x += 1`
```

Here, `x` changes on every iteration until `x <= 5` becomes `False`.

---

## Increment and Decrement

Python does **not** use `++` or `--` operators.

Use:

`x += 1`

instead of:

`x++`

And:

`x -= 1`

instead of:

`x--`

---

## `while` With `else`

Python allows an `else` block with a `while` loop.

The `else` block runs when the loop finishes normally because its condition becomes `False`.

It does **not** run if the loop is stopped using `break`.

### Syntax

`while condition:`

```
`# code`
```

`else:`

```
`# code`
```

---

## `break`

`break` immediately stops the loop.

The loop ends even if its condition is still `True`.

---

## `continue`

`continue` skips the rest of the current iteration and starts the next iteration.

---

## Common Uses

`while` loops are useful when:

* The number of iterations is not known beforehand.
* The loop depends on a condition.
* Waiting for valid user input.
* Repeating an action until a specific value is entered.
* Running until a condition changes.

---

## Important Notes

* `while` depends on a condition.
* The condition is checked before each iteration.
* A `while` loop can run zero or more times.
* Make sure the condition can eventually become `False`.
* `break` stops the loop completely.
* `continue` skips the current iteration.
* Python does not have `++` or `--`.
* `while` can have an `else` block.


````python
# -------------------
# -- While Loop --
# -------------------

# Basic while loop
counter = 1

while counter <= 5:
    counter += 1


# -------------------
# -- Increment --
# -------------------

x = 1

while x <= 5:
    x += 1


# -------------------
# -- Decrement --
# -------------------

x = 5

while x >= 1:
    x -= 1


# -------------------
# -- User Input --
# -------------------

password = ""

while password != "1234":
    password = input("Enter password: ")


# -------------------
# -- break --
# -------------------

x = 1

while x <= 10:
    if x == 5:
        break

    x += 1


# -------------------
# -- continue --
# -------------------

x = 0

while x < 5:
    x += 1

    if x == 3:
        continue


# -------------------
# -- while / else --
# -------------------

x = 1

while x <= 5:
    x += 1
else:
    pass


# -------------------
# -- break + else --
# -------------------

x = 1

while x <= 5:
    if x == 3:
        break

    x += 1
else:
    # This does not run because break stopped the loop
    pass


# -------------------
# -- Practical Example --
# -- Sum Until 0 --
# -------------------

total = 0
number = int(input("Enter a number (0 to stop): "))

while number != 0:
    total += number
    number = int(input("Enter a number (0 to stop): "))

# total contains the final sum
````

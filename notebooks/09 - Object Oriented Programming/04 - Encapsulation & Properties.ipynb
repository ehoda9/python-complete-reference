# 09.4 - Encapsulation & Properties

## Encapsulation

Encapsulation means keeping data and the operations that work with that data together, while controlling how the object is accessed.

Python does not provide Java-style enforced private fields. It relies heavily on conventions and language mechanisms.

## Access Conventions

| Form | Meaning |
|---|---|
| `name` | Public API by convention |
| `_name` | Protected/internal convention; not enforced |
| `__name` | Private-style name mangling |

### Single Underscore

```python
self._balance
```

This means:

> "This is intended for internal use."

Python does not prevent outside access.

### Double Underscore

```python
self.__balance
```

Python applies **name mangling**, usually turning it into:

```text
_ClassName__balance
```

It is not absolute security or true private memory.

## Properties

`@property` allows a method to be accessed like an attribute:

```python
user.age_in_days
```

instead of:

```python
user.age_in_days()
```

A property is useful when you want attribute-style syntax with computed or controlled access.

## Property Getter / Setter / Deleter

```python
@property
def value(self):
    ...

@value.setter
def value(self, new_value):
    ...

@value.deleter
def value(self):
    ...
```

## Getter and Setter Methods

Explicit methods such as:

```python
get_name()
set_name(...)
```

are valid, but properties often produce a cleaner Python API when the concept behaves like an attribute.

## `@functools.cached_property`

- Computes a property the first time it is accessed.
- Stores the result as an instance attribute.
- Later accesses use the stored value until it is deleted.
- Useful for expensive, mostly stable per-instance computations.

## Important

Encapsulation is about **API design and invariants**, not simply hiding variables.

Example invariant:

```text
age >= 0
balance >= 0
```

A setter or method can enforce such rules.

## Source

- Descriptor Guide: https://docs.python.org/3/howto/descriptor.html
- `functools.cached_property`: https://docs.python.org/3/library/functools.html

````python
from functools import cached_property


class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Balance cannot be negative.")
        self._balance = value

    @cached_property
    def account_summary(self):
        print("Calculating summary...")
        return f"{self.owner}: ${self.balance}"


account = BankAccount("Ahmed", 100)

print(account.balance)

account.balance = 250
print(account.balance)

print(account.account_summary)
print(account.account_summary)
````

## Docstrings => Commenting vs Documenting

* A **docstring** is a string used to document a **module, class, function, or method**.
* A docstring is written as the **first statement** inside the object being documented.
* Python commonly uses triple quotes `"""..."""` for docstrings.
* Docstrings can be accessed through the `__doc__` attribute.
* They are also used by tools such as `help()`.
* Docstrings explain **what something does, how it should be used, and sometimes its parameters and return value**.
* Docstrings can be **one line** or **multiple lines**.
* A **comment** explains the code to developers and is ignored by Python.
* A **docstring** is actually a string object stored as documentation metadata.

### Comment vs Docstring

| Comment                                         | Docstring                                            |
| ----------------------------------------------- | ---------------------------------------------------- |
| Starts with `#`                                 | Usually uses `"""..."""`                             |
| Explains code or implementation                 | Documents a module, class, function, or method       |
| Ignored by Python                               | Stored in `__doc__`                                  |
| Not available through `help()` as documentation | Used by `help()`                                     |
| Can appear almost anywhere                      | Must be the first statement of the documented object |

### One-Line Docstring

```text
def greet(name):
    """Return a greeting for the given name."""
```

### Multi-Line Docstring

```text
def calculate_average(numbers):
    """
    Calculate the average of a list of numbers.

    Parameters:
        numbers: A collection of numeric values.

    Returns:
        The average of the numbers.
    """
```

### Important Difference

```text
# This is a comment.
```

The comment is not stored as the function's documentation.

```text
def greet():
    """Return a greeting message."""
```

The string becomes the function's docstring and can be accessed with:

```text
greet.__doc__
```

or:

```text
help(greet)
```


````python
# Docstring: documents the function and can be accessed through __doc__ and help()
def greet(name):
    """Return a greeting message for the given name."""
    
    # Comment: explains an implementation detail
    message = f"Hello, {name}!"
    
    return message


print(greet("Mahmoud"))

# Access the function's docstring
print(greet.__doc__)

# Display documentation
help(greet)
````

# Functions Basics

A function is a **reusable block of code** designed to perform a specific task.

Functions help organize code, avoid repetition, and make programs easier to understand and maintain.

---

## Why Use Functions?

* Reuse the same code multiple times.
* Avoid repeating code (**DRY - Don't Repeat Yourself**).
* Break a large program into smaller tasks.
* Make code easier to read and maintain.
* Make code easier to test and debug.

---

## Creating a Function

Use the `def` keyword.

### Syntax

`def function_name():`

```
# code
```

The function body must be indented.

---

## Calling a Function

Defining a function does not execute it.

The function runs when you **call** it.

### Example

`def greet():`

```
`print("Hello")`
```

`greet()`

---

## Parameters

Parameters are variables defined inside the function definition.

They allow the function to receive data.

`def greet(name):`

```
`print(name)`
```

Here, `name` is a **parameter**.

---

## Arguments

Arguments are the actual values passed when calling the function.

`greet("Mahmoud")`

Here, `"Mahmoud"` is an **argument**.

---

## Return Value

A function can return data using `return`.

`def add(a, b):`

```
`return a + b`
```

The returned value can be stored or used in another expression.

`result = add(10, 5)`

---

## Functions Without `return`

A function can perform a task without explicitly returning a value.

If a function reaches the end without a `return`, Python returns `None`.

---

## Built-in vs User-Defined Functions

### Built-in Functions

Functions already provided by Python.

Examples:

* `print()`
* `len()`
* `type()`
* `input()`
* `sum()`
* `max()`
* `min()`

### User-Defined Functions

Functions created by the programmer using `def`.

---

## Function Structure

A function can contain:

* Parameters
* Local variables
* Conditions
* Loops
* Other function calls
* `return`

---

## Important Notes

* `def` creates a function.
* Calling the function executes it.
* Parameters are defined in the function.
* Arguments are passed when calling the function.
* A function can return a value.
* A function can return nothing explicitly, in which case the result is `None`.
* Functions are reusable and help prevent duplicated code.


````python
# -------------------------
# -- Function Basics --
# -------------------------

# Simple function
def greet():
    print("Hello")


# Calling the function
greet()


# Function with a parameter
def greet_user(name):
    print("Hello", name)


greet_user("Mahmoud")


# Function with multiple parameters
def add(a, b):
    return a + b


result = add(10, 5)


# Function without an explicit return
def say_hello():
    print("Hello")


result = say_hello()


# Built-in function
length = len("Python")


# User-defined function
def multiply(a, b):
    return a * b


result = multiply(4, 5)
````

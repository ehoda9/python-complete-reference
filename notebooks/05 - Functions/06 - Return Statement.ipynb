# Function `return`

The `return` statement sends a value back from a function.

It also immediately ends the function execution.

---

## Basic Syntax

`def add(a, b):`

```
`return a + b`
```

The returned value can be stored:

`result = add(10, 5)`

---

## `return` vs `print()`

`print()` displays a value.

`return` sends a value back to the code that called the function.

### `print()`

`def add(a, b):`

```
`print(a + b)`
```

The function displays the result but does not return it.

### `return`

`def add(a, b):`

```
`return a + b`
```

The result can be stored, reused, or passed to another function.

---

## `return` Ends the Function

Any code after an executed `return` in the same function will not run.

---

## Multiple Return Values

Python allows a function to return multiple values.

`return name, age`

Python packs them into a tuple.

---

## No `return`

If a function does not explicitly return a value, it returns `None`.

---

## Conditional Return

A function can return different values depending on a condition.

---

## Important Notes

* `return` sends data back to the caller.
* `return` immediately stops the current function execution.
* A function can have multiple `return` statements.
* Only the first executed `return` is reached.
* Multiple returned values are packed into a tuple.
* A function without an explicit return gives `None`.


````python
# --------------------
# -- Function Return --
# --------------------

# Returning a value
def add(a, b):
    return a + b


result = add(10, 5)


# print vs return
def add_with_print(a, b):
    print(a + b)


def add_with_return(a, b):
    return a + b


add_with_print(10, 5)

result = add_with_return(10, 5)


# return stops the function
def check_number(number):
    if number > 0:
        return "Positive"

    return "Zero or Negative"


check_number(10)


# Multiple return values
def get_user():
    name = "Mahmoud"
    age = 20

    return name, age


name, age = get_user()


# No explicit return
def say_hello():
    print("Hello")


result = say_hello()


# Conditional return
def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"

    return "F"
````

# Decorators

A decorator adds behavior to a function without changing its code.

| Topic | Point |
|---|---|
| Decorator | A function that takes a function, wraps it and returns the wrapper |
| `@decorator` | Shorthand for `greet = decorator(greet)` |
| Wrapper | The inner function that adds the extra behavior |
| Functions with parameters | The wrapper forwards them with `*args` and `**kwargs` |
| Typical uses | Logging, authentication, validation, timing, caching, access control |

---

## Decorators => Introduction

* A **decorator** is a function that takes another function, adds or changes its behavior, and returns a function.
* Decorators are based on Python's **first-class functions** concept: functions are objects and can be stored in variables, passed as arguments, and returned from other functions.
* A decorator is a **higher-order function** because it works with functions as arguments or return values.
* A decorator **wraps** another function to enhance its behavior without changing the original function's code.
* Decorators are commonly used for logging, authentication, validation, timing, caching, and access control.
* The `@decorator` syntax is shorthand for replacing a function with the decorated version.

### Basic Structure

```text
decorator(function)
       =>
    wrapper
       =>
enhanced function
```

### `@decorator` Syntax

These two forms are equivalent:

```text
@decorator
def greet():
    ...
```

```text
def greet():
    ...

greet = decorator(greet)
```

### Important Idea

The decorator does **not** usually modify the original function directly.

Instead:

1. The original function is passed to the decorator.
2. The decorator creates a wrapper function.
3. The wrapper adds extra behavior.
4. The decorator returns the wrapper.
5. The original function name now refers to the decorated function.


````python
# Decorator: takes a function and returns an enhanced function
def my_decorator(func):
    def wrapper():
        print("Before the function")
        func()
        print("After the function")
    
    return wrapper


# @my_decorator: automatically applies my_decorator to greet
@my_decorator
def greet():
    print("Hello!")


greet()
````

## Decorators => Function With Parameters

When the decorated function has parameters, the wrapper must also be able to receive those arguments.

A common solution is:

* `*args` => receives positional arguments.
* `**kwargs` => receives keyword arguments.
* `func(*args, **kwargs)` => passes all received arguments to the original function.

This makes the decorator reusable with functions that have different parameter lists.

### General Pattern

```text
def decorator(func):
    def wrapper(*args, **kwargs):
        # Extra behavior
        result = func(*args, **kwargs)
        # More behavior
        return result

    return wrapper
```

### Why `*args` and `**kwargs`?

Suppose one function receives:

```text
greet("Mahmoud")
```

while another receives:

```text
add(10, 20)
```

and another receives:

```text
login(username="Mahmoud", role="admin")
```

Using `*args` and `**kwargs` allows the same decorator to work with all of them.


````python
# Decorator: accepts a function and supports any positional/keyword arguments
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Before the function")

        result = func(*args, **kwargs)

        print("After the function")

        return result

    return wrapper


# Decorator applied to a function with parameters
@my_decorator
def greet(name):
    print(f"Hello, {name}!")


greet("Mahmoud")


# Same decorator works with multiple parameters
@my_decorator
def add(a, b):
    return a + b


result = add(10, 20)
print(result)
````

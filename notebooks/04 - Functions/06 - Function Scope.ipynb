# Function Scope

Scope determines where a variable can be accessed.

| Topic | Meaning |
|---|---|
| Local scope | Variables created inside a function |
| Global scope | Variables created outside functions |
| Local vs global | A local variable hides a global one with the same name |
| `global` | Lets a function change a global variable |
| `nonlocal` | Lets an inner function change a variable of the enclosing function |
| Closure | An inner function that remembers the variables of the function that created it |
| LEGB rule | Python looks in Local, Enclosing, Global, then Built-in |

---

A variable created inside a function normally belongs to that function's **local scope**.

---

## Local Scope

A variable created inside a function is local to that function.

`def greet():`

```
`name = "Mahmoud"`
```

`name` cannot normally be accessed outside the function.

---

## Global Scope

A variable created outside functions belongs to the global scope.

`name = "Mahmoud"`

Functions can read global variables.

---

## Local vs Global

If a local variable has the same name as a global variable, the local variable takes priority inside the function.

---

## `global`

The `global` keyword allows a function to modify a global variable.

Example:

`count = 0`

`def increase():`

```
`global count`

`count += 1`
```

Using global variables extensively can make code harder to understand, so local variables and return values are usually preferred.

---

## LEGB Rule

Python searches for names using the LEGB order:

| Scope | Meaning   |
| ----- | --------- |
| `L`   | Local     |
| `E`   | Enclosing |
| `G`   | Global    |
| `B`   | Built-in  |

Python searches from the nearest scope outward.

---

## Important Notes

* Function parameters are local variables.
* Variables created inside a function are normally local.
* Global variables can be read inside functions.
* `global` is required when assigning to a global variable from inside a function.
* A function creates its own local scope.
* `if`, `for`, and `while` do not create a separate local scope in Python.

## `nonlocal`

`nonlocal` lets an inner function change a variable of the **enclosing** function.

```python
def counter():
    count = 0

    def increase():
        nonlocal count
        count += 1
        return count

    return increase
```

* Without `nonlocal`, `count += 1` would create a new local variable and fail.
* `global` refers to the module level. `nonlocal` refers to the nearest enclosing function.

---

## Closures

A **closure** is an inner function that remembers the variables of the function that created it, even after that function has finished.

* The inner function uses a variable from the enclosing scope.
* The enclosing function returns the inner function.
* Typical uses: counters, function factories and decorators.

````python
# --------------------
# -- Function Scope --
# --------------------

# Global variable
name = "Mahmoud"


def greet():
    # Local variable
    message = "Hello"

    # Can read global variable
    return message, name


greet()


# Local variable
def calculate():
    number = 10
    return number


result = calculate()


# Same name: local takes priority
name = "Mahmoud"


def show_name():
    name = "Ahmed"
    return name


show_name()


# global
count = 0


def increase():
    global count
    count += 1


increase()
increase()


# Enclosing scope
def outer():
    message = "Hello"

    def inner():
        return message

    return inner()


outer()

# nonlocal
def counter():
    count = 0

    def increase():
        nonlocal count
        count += 1
        return count

    return increase


tick = counter()
tick()
tick()


# Closure: a function factory
def multiplier(factor):
    def multiply(number):
        return number * factor

    return multiply


double = multiplier(2)
triple = multiplier(3)
double(5)
triple(5)
````

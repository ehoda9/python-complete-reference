# Functions => Coding Style (PEP 8)

**PEP 8** is Python's style guide. Code that follows it is easier to read, review and share.

| Topic | Rule |
|---|---|
| Indentation | 4 spaces per level. Never mix tabs and spaces |
| Line length | 79 characters (72 for comments and docstrings) |
| Blank lines | 2 around top-level functions and classes, 1 between methods |
| Imports | One per line, at the top: standard library, then third-party, then your own |
| Functions and variables | `snake_case` |
| Classes | `CapWords` |
| Constants | `UPPER_CASE` |
| Comparisons | `is None`, `if items:`, `x not in items` |
| Comments | Explain **why**, in complete sentences |
| Docstrings | Triple double quotes, a one-line summary first |
| Tools | `black`, `ruff`, `flake8`, `pycodestyle`, `isort` |

---

## Naming

| Kind | Convention | Example |
|---|---|---|
| Variable, function, method | `snake_case` | `total_price`, `read_file()` |
| Class | `CapWords` | `ShoppingCart` |
| Constant | `UPPER_CASE` | `MAX_RETRIES` |
| Module, package | short `lowercase` | `utils`, `mypackage` |
| "Private" name | leading underscore | `_cache` |
| Unused variable | `_` | `for _ in range(3)` |

### Important

* Names should describe **meaning**: `days_left`, not `d`.
* Avoid the single letters `l`, `O` and `I`. They look like `1` and `0`.
* Never name a variable after a built-in (`list`, `sum`, `type`, `id`) or a keyword.

## Layout

```python
import os
import sys

from pathlib import Path


def add(a, b):
    return a + b


class Cart:
    def total(self):
        ...
```

* Two blank lines around top-level definitions.
* Long calls can break after an opening bracket. Indent the next line by 4 spaces.

---

## Whitespace

| Do | Do not |
|---|---|
| `x = y + 1` | `x=y+1` |
| `f(a, b)` | `f (a, b)` or `f( a, b )` |
| `items[0]` | `items [0]` |
| `{"a": 1}` | `{"a" : 1}` |
| `def f(x=1):` | `def f(x = 1):` |
| `x: int = 5` | `x:int=5` |

---

## Idioms

| Avoid | Prefer |
|---|---|
| `if x == None:` | `if x is None:` |
| `if flag == True:` | `if flag:` |
| `if len(items) == 0:` | `if not items:` |
| `if not x in items:` | `if x not in items:` |
| `for i in range(len(items)):` | `for item in items:` or `enumerate(items)` |
| `except:` | `except ValueError:` |
| `lambda` assigned to a name | A normal `def` |

---

## Comments and Docstrings

* A comment that repeats the code is noise. Explain **why**.
* Keep comments up to date. A wrong comment is worse than none.
* Public functions, classes and modules get docstrings.

---

## Tools

| Tool | What it does |
|---|---|
| `black` | Formats the code automatically |
| `ruff` | Very fast linter and formatter |
| `flake8` / `pycodestyle` | Report style problems |
| `isort` | Sorts imports |

````python
import keyword

# Names: valid identifiers, and names that are not allowed
for name in ("total_price", "ShoppingCart", "MAX_RETRIES", "2fast", "my-name", "class"):
    print(f"{name:<14} valid identifier: {name.isidentifier() and not keyword.iskeyword(name)}")

# Avoid names that hide built-ins
list = [1, 2, 3]                      # bad: hides the built-in list()
print(type(list).__name__)            # a list object now, not the built-in function
del list                              # the built-in is visible again
print(list("abc"))

# Idioms: comparisons
value = None
print(value is None, value == None)   # `is None` is the correct test

flag = True
items = []
print(bool(flag), not items)          # `if flag:` and `if not items:` instead of == True / len(items) == 0
print("x" not in items)               # `not in`, not `not x in`

# Iterate directly instead of using range(len(...))
names = ["ann", "bob"]
for index, name in enumerate(names):
    print(index, name)

# A well-styled function
MAX_RETRIES = 3


def retry_count(attempts, limit=MAX_RETRIES):
    """Return how many retries are left."""
    return max(limit - attempts, 0)


print(retry_count(1), retry_count(5))
````

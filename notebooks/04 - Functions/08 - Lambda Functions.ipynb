# Lambda Functions

A `lambda` is a small **anonymous function**.

| Point | Detail |
|---|---|
| Syntax | `lambda parameters: expression` |
| No name | A lambda is anonymous unless you assign it to a name |
| Automatic return | The expression after `:` is returned |
| One expression only | No statements inside |
| As an argument | Passed to functions that accept another function |
| Lambda vs `def` | Use `def` when the logic is longer or reused |

---

It is useful when you need a short function for a simple operation.

---

## Syntax

`lambda parameters: expression`

Example:

`lambda x: x * 2`

This creates a function that receives `x` and returns `x * 2`.

---

## Lambda Has No Name

Unlike a normal function:

`def double(x):`

```
`return x * 2`
```

a lambda does not need a function name.

However, a lambda can be assigned to a variable:

`double = lambda x: x * 2`

In this case, `double` refers to the function.

---

## Lambda Returns Automatically

The expression after `:` is automatically returned.

`lambda x: x * 2`

is similar to:

`def double(x):`

```
`return x * 2`
```

---

## Lambda Is One Expression

A lambda can contain only **one expression**.

It cannot contain a normal block of statements such as multiple lines, loops, or regular `if` statements.

A conditional expression can still be used:

`lambda x: "Even" if x % 2 == 0 else "Odd"`

---

## Calling a Lambda

`double = lambda x: x * 2`

`double(5)`

=> `10`

A lambda can also be called immediately:

`(lambda x: x * 2)(5)`

=> `10`

---

## Lambda as an Argument

Lambdas are commonly passed to functions that accept another function.

Common examples:

* `sorted()`
* `map()`
* `filter()`

Example:

`sorted(names, key=lambda name: len(name))`

---

## Lambda vs `def`

| Lambda                           | `def`                           |
| -------------------------------- | ------------------------------- |
| Anonymous function               | Named function                  |
| One expression                   | Can contain multiple statements |
| Automatically returns expression | Uses `return` explicitly        |
| Best for short/simple operations | Better for larger logic         |
| Commonly used inline             | Usually defined separately      |

---

## Important Notes

* A lambda is an object of type `function`.
* Lambda functions are useful for short operations.
* Lambda can accept multiple parameters.
* Lambda can have default parameters.
* Lambda can use conditional expressions.
* Lambda is not a replacement for normal functions.
* Use `def` when the logic becomes large or needs to be reused clearly.


````python
# ------------------------
# -- Function => lambda --
# -- Anonymous Function --
# ------------------------

# Simple lambda
double = lambda x: x * 2

double(5)


# Multiple parameters
add = lambda a, b: a + b

add(10, 5)


# Lambda with a condition
check_number = lambda number: "Even" if number % 2 == 0 else "Odd"

check_number(10)


# Lambda can be called directly
(lambda x: x * 2)(5)


# Lambda with sorted()
names = ["Mahmoud", "Ali", "Ibrahim", "Mona"]

sorted(names, key=lambda name: len(name))


# Lambda with map()
numbers = [1, 2, 3, 4, 5]

list(map(lambda number: number * 2, numbers))


# Lambda with filter()
list(filter(lambda number: number % 2 == 0, numbers))


# Lambda type
type(lambda x: x * 2)
````

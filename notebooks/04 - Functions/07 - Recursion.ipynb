# Function Recursion

Recursion happens when a function **calls itself**.

| Part | Meaning |
|---|---|
| Base case | The condition that stops the recursion |
| Recursive case | The function calls itself with a smaller or simpler problem |
| Call stack | Python keeps track of every pending call |
| Recursion vs loop | The same problem can often be solved either way |
| When to use it | Problems made of smaller versions of themselves |

---

A recursive function solves a problem by breaking it into smaller versions of the same problem.

---

## The Two Important Parts

A recursive function needs:

### 1. Base Case

The condition that stops the recursion.

Without a base case, the function keeps calling itself until Python raises a `RecursionError`.

### 2. Recursive Case

The part where the function calls itself with a smaller or simpler problem.

---

## Basic Structure

`def function(value):`

```
`if stopping_condition:`

    `return result`

`return function(smaller_value)`
```

---

## Example

Factorial:

`5! = 5 × 4 × 3 × 2 × 1`

The recursive idea is:

`factorial(n) = n × factorial(n - 1)`

The base case:

`factorial(0) = 1`

---

## How Recursion Works

When a function calls itself, Python keeps track of each function call using the **call stack**.

Each call waits for the next call to finish.

Once the base case is reached, the calls return one by one.

---

## Recursion vs Loop

| Recursion                                     | Loop                                         |
| --------------------------------------------- | -------------------------------------------- |
| Function calls itself                         | Repeats a block                              |
| Uses call stack                               | Usually simpler memory usage                 |
| Needs a base case                             | Needs a condition                            |
| Can make recursive problems easier to express | Often simpler for straightforward repetition |

---

## When to Use Recursion

Recursion can be useful for problems that naturally contain smaller versions of themselves, such as:

* Tree structures
* Nested data
* Divide-and-conquer algorithms
* Graph traversal
* Mathematical recursive problems

For simple repetition, a loop is usually easier.

---

## Important Notes

* A recursive function must have a base case.
* Each recursive call should move toward the base case.
* Too much recursion can cause `RecursionError`.
* Python has a recursion depth limit.
* Recursion is not automatically better than loops.


````python
# ------------------------
# -- Function Recursion --
# ------------------------

# Basic recursion
def countdown(number):
    if number == 0:
        return

    countdown(number - 1)


countdown(5)


# ------------------------
# -- Factorial --
# ------------------------

def factorial(number):
    if number == 0:
        return 1

    return number * factorial(number - 1)


factorial(5)


# ------------------------
# -- Recursive Sum --
# ------------------------

def recursive_sum(number):
    if number == 0:
        return 0

    return number + recursive_sum(number - 1)


recursive_sum(5)


# ------------------------
# -- Recursive Countdown --
# ------------------------

def countdown(number):
    if number == 0:
        return

    print(number)
    countdown(number - 1)


countdown(5)
````

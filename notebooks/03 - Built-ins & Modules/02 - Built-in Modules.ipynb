# Modules => Built-in Modules

A **module** is a Python file that contains reusable code such as functions, classes, and variables.

Modules help us:

* Reuse code instead of rewriting it.
* Organize large programs.
* Keep related code together.
* Save development time.
* Use functionality provided by Python or other developers.
* Create our own reusable code.

## Built-in Modules

Python provides many modules as part of the standard library.

Examples:

* `math` => mathematical operations
* `random` => random values
* `datetime` => dates and times
* `os` => operating-system functionality
* `sys` => Python runtime/system functionality
* `json` => JSON data
* `re` => regular expressions

You do not need to memorize every module. The important concept is how to import and use them.

---

## `import`

Imports the entire module.

`import module_name`

Then access its members using:

`module_name.member`

---

## Import With `as`

Creates an alias for the module.

`import module_name as alias`

Then use:

`alias.member`

Useful when the module name is long or when you want a shorter name.

---

## `from ... import`

Imports specific members from a module.

`from module_name import member`

Then you can use the member directly:

`member()`

---

## Import Multiple Members

`from module_name import member1, member2`

You can import multiple functions, classes, or variables from the same module.

---

## Import Everything

`from module_name import *`

This imports names from the module into the current namespace.

### Important

This style is generally **not recommended** because:

* It can make it unclear where names came from.
* It can cause name conflicts.
* It makes large programs harder to understand.

---

## Import Multiple Modules

You can import multiple modules.

`import math, random`

However, separate import statements are usually easier to read.

---

## Custom Modules

You can create your own module simply by creating a `.py` file.

Example structure:

`math_tools.py`

```
`def add(a, b):`

    `return a + b`
```

Then another Python file can import it:

`import math_tools`

and use:

`math_tools.add(10, 20)`

---

## `from` With Custom Modules

You can import specific members from your own module:

`from math_tools import add`

Then:

`add(10, 20)`

---

## Module Search

When importing a module, Python searches locations listed in its module search path.

`sys.path`

This is why a custom module in the same project/directory can normally be imported directly.

---

## Important Notes

* A module is normally a `.py` file.
* A module can contain functions, classes, variables, and executable code.
* `import module` keeps the module namespace separate.
* `from module import name` imports a specific name directly.
* `as` creates an alias.
* `from module import *` is generally discouraged.
* You can create your own modules.
* Built-in/standard-library modules do not normally require installation.
* External packages usually need to be installed separately.

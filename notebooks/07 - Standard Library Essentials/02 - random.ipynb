# Standard Library => random

`random` generates **pseudo-random** numbers and makes random choices.

| Function | Purpose | Example |
|---|---|---|
| `random.random()` | Float in `[0.0, 1.0)` | `0.37...` |
| `random.uniform(a, b)` | Float between `a` and `b` | `random.uniform(1, 5)` |
| `random.randint(a, b)` | Integer, **both ends included** | `random.randint(1, 6)` |
| `random.randrange(a, b, step)` | Integer, `b` **excluded** | `random.randrange(0, 10, 2)` |
| `random.choice(seq)` | One random item | `random.choice("abc")` |
| `random.choices(seq, k=n)` | `n` items, **with** repeats (optional `weights=`) | |
| `random.sample(seq, k=n)` | `n` unique items, **no** repeats | |
| `random.shuffle(list)` | Shuffle a list **in place** | |
| `random.seed(n)` | Make the sequence reproducible | |
| `random.gauss(mu, sigma)` | Normal distribution | |
| `random.Random(seed)` | Independent generator object | |
| `secrets` module | Cryptographically secure values | `secrets.token_hex(16)` |

```python
import random
```

---

## `randint` vs `randrange`

* `randint(1, 6)` can return `1`, `2`, `3`, `4`, `5` or `6` (a die).
* `randrange(1, 6)` can return `1` to `5`, like `range()`.

---

## Choosing and Sampling

| Need | Use |
|---|---|
| One item | `choice` |
| Several items, repeats allowed | `choices` (supports `weights=`) |
| Several unique items | `sample` |
| Reorder a list | `shuffle` (changes the list itself) |

### Important

* `sample(seq, k)` raises `ValueError` if `k` is larger than the sequence.
* `choice([])` raises `IndexError`.
* `shuffle` returns `None`. To keep the original, shuffle a copy or use `sample(seq, k=len(seq))`.

---

## Reproducible Results

The same **seed** gives the same sequence. This is useful for tests and experiments.

```python
random.seed(42)
```

For code that must not affect the global generator, use its own generator: `rng = random.Random(42)`.

---

## Not for Security

`random` is **predictable**. Never use it for passwords, tokens or keys. Use the `secrets` module instead.

````python
import random
import secrets

# The same seed gives the same sequence (the exact numbers can differ between Python versions)
random.seed(42)
first = [random.random() for _ in range(3)]
random.seed(42)
second = [random.random() for _ in range(3)]
print(first == second)

# Ranges
print(all(0 <= random.random() < 1 for _ in range(100)))
print(all(1 <= random.uniform(1, 5) <= 5 for _ in range(100)))
print(sorted({random.randint(1, 6) for _ in range(500)}))         # a die: 1 to 6
print(sorted({random.randrange(1, 6) for _ in range(500)}))       # 6 is excluded
print(sorted({random.randrange(0, 10, 2) for _ in range(500)}))   # even numbers only

# Choosing and sampling
letters = "abcde"
print(random.choice(letters) in letters)
print(len(random.choices(letters, k=8)))                          # repeats allowed
picked = random.sample(letters, k=3)
print(len(picked), len(set(picked)))                              # unique items
print(random.choices(["a", "b"], weights=[0, 1], k=5))            # weight 0 is never chosen

# shuffle works in place and returns None
numbers = [1, 2, 3, 4, 5]
print(random.shuffle(numbers), sorted(numbers))

# Errors
try:
    random.sample(letters, k=10)
except ValueError as error:
    print("ValueError:", error)

try:
    random.choice([])
except IndexError as error:
    print("IndexError:", error)

# An independent generator does not touch the global one
rng = random.Random(7)
print(rng.randint(1, 100) == random.Random(7).randint(1, 100))

# Normal distribution
print(type(random.gauss(0, 1)).__name__)

# secrets: for anything security related
token = secrets.token_hex(8)
print(len(token), secrets.choice("abc") in "abc")
````

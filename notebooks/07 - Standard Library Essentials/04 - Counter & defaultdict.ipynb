# `collections` => Counter & defaultdict

The `collections` module provides specialized container types. These two solve the most common dictionary chores: **counting** and **grouping**.

| Tool | Purpose | Example |
|---|---|---|
| `Counter(iterable)` | Count hashable items | `Counter("hello")` |
| `Counter.most_common(n)` | The `n` most frequent items | `c.most_common(2)` |
| `Counter.update()` / `.subtract()` | Add or remove counts | `c.update("aa")` |
| `Counter.elements()` | Repeat each item by its count | `list(c.elements())` |
| `Counter.total()` | Sum of all counts (Python 3.10+) | `c.total()` |
| `+` `-` `&` `\|` | Combine two counters | `c1 + c2` |
| `defaultdict(factory)` | A dict that creates missing values | `defaultdict(list)` |

```python
from collections import Counter, defaultdict
```

---

## `Counter`

A `Counter` is a `dict` subclass that maps each item to its **count**.

### Creating a Counter

```python
Counter("mississippi")            # from an iterable
Counter({"a": 3, "b": 1})         # from a dict
Counter(a=3, b=1)                 # from keyword arguments
```

### Missing Items Count as 0

A missing key returns `0` and is **not** added to the counter.

```python
c = Counter("aab")
c["z"]        # 0
```

### Most Common

```python
Counter("mississippi").most_common(2)    # [('i', 4), ('s', 4)]
```

Items with equal counts keep the order in which they were first seen.

### Update and Subtract

* `update(iterable)` **adds** counts.
* `subtract(iterable)` **removes** counts. Results can become zero or negative.

### `elements()` and `total()`

* `elements()` returns an iterator that repeats each item by its count. Items with a count below 1 are skipped.
* `total()` returns the sum of all counts.

---

## Counter Arithmetic

| Operation | Meaning | Keeps |
|---|---|---|
| `c1 + c2` | Add counts | Positive results only |
| `c1 - c2` | Subtract counts | Positive results only |
| `c1 & c2` | Minimum of each count | Intersection |
| `c1 \| c2` | Maximum of each count | Union |
| `+c` | Remove zero and negative counts | Positive counts only |

---

## `defaultdict`

A `defaultdict` calls a **factory function** to create the value of a missing key.

### Syntax

`defaultdict(default_factory)`

| Factory | Default value | Typical use |
|---|---|---|
| `int` | `0` | Counting |
| `list` | `[]` | Grouping items |
| `set` | `set()` | Grouping unique items |
| `lambda: "N/A"` | Any custom value | Custom default |

```python
groups = defaultdict(list)
for word in ["apple", "avocado", "banana"]:
    groups[word[0]].append(word)
```

### Important

* Reading a missing key **creates** it. `key in d` and `d.get(key)` do **not**.
* The factory takes no arguments.
* `dict(d)` converts it back to a normal dict.
* Nested defaults: `defaultdict(lambda: defaultdict(int))`.

---

## Which One?

| Need | Use |
|---|---|
| Count occurrences | `Counter` |
| Group items by key | `defaultdict(list)` |
| Group unique items | `defaultdict(set)` |
| Simple default for one lookup | `d.get(key, default)` |
| Group **sorted** data lazily | `itertools.groupby` (see the Itertools part) |

````python
from collections import Counter, defaultdict

# Creating counters
print(Counter("mississippi"))
print(Counter({"a": 3, "b": 1}), Counter(a=3, b=1))

# Missing items count as 0 and are not added
c = Counter("aab")
print(c["z"], "z" in c)

# most_common
words = "the cat and the hat and the bat".split()
freq = Counter(words)
print(freq.most_common(2))

# update / subtract
c = Counter(a=3, b=1)
c.update("aab")
print(c)                                   # a: 5, b: 2
c.subtract({"a": 4, "b": 5})
print(c)                                   # a: 1, b: -3

# elements() skips counts below 1; total() sums all counts
print(sorted(Counter(a=2, b=1, c=0, d=-1).elements()))
print(Counter(a=2, b=3).total())

# Arithmetic
one, two = Counter(a=3, b=1), Counter(a=1, b=2)
print(one + two)                           # a: 4, b: 3
print(one - two)                           # a: 2  (b dropped: not positive)
print(one & two)                           # minimum: a: 1, b: 1
print(one | two)                           # maximum: a: 3, b: 2
print(+Counter(a=2, b=0, c=-1))            # only positive counts

# defaultdict: grouping
groups = defaultdict(list)
for word in ["apple", "avocado", "banana", "blueberry", "cherry"]:
    groups[word[0]].append(word)
print(dict(groups))

# defaultdict(int): counting, defaultdict(set): unique grouping
letters = defaultdict(int)
for ch in "hello":
    letters[ch] += 1
print(dict(letters))

unique = defaultdict(set)
for key, value in [("a", 1), ("a", 1), ("a", 2), ("b", 3)]:
    unique[key].add(value)
print(dict(unique))

# Reading a missing key creates it; `in` and .get() do not
d = defaultdict(list)
print("x" in d, d.get("x"), len(d))
d["x"]
print("x" in d, len(d))

# Custom default and nested defaults
labels = defaultdict(lambda: "N/A")
print(labels["missing"])

nested = defaultdict(lambda: defaultdict(int))
nested["fruit"]["apple"] += 2
print({k: dict(v) for k, v in nested.items()})
````

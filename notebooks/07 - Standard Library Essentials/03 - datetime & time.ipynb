# Standard Library => datetime & time

`datetime` works with dates, times and durations. `time` works with the clock and with pauses.

| Tool | Purpose | Example |
|---|---|---|
| `date(y, m, d)` | A calendar date | `date(2026, 9, 20)` |
| `time(h, m, s)` | A time of day | `time(14, 30)` |
| `datetime(y, m, d, h, mi, s)` | Date and time together | `datetime(2026, 9, 20, 14, 30)` |
| `datetime.now()` / `date.today()` | The current moment / day | |
| `timedelta(days=, hours=, ...)` | A duration | `timedelta(days=7)` |
| `dt.strftime(fmt)` | Date -> text | `dt.strftime("%Y-%m-%d")` |
| `datetime.strptime(text, fmt)` | Text -> date | `strptime("2026-09-20", "%Y-%m-%d")` |
| `dt.isoformat()` / `fromisoformat()` | ISO 8601 text | `2026-09-20T14:30:00` |
| `dt.replace(...)` | Copy with changed fields | `dt.replace(hour=0)` |
| `dt.weekday()` | Monday is `0` | |
| `timezone(timedelta(...))` | A fixed UTC offset | `timezone.utc` |
| `time.time()` | Seconds since 1970 (epoch) | |
| `time.perf_counter()` | Precise timer for measuring | |
| `time.sleep(seconds)` | Pause the program | `time.sleep(0.5)` |

```python
from datetime import date, time, datetime, timedelta, timezone
import time
```

---

## Creating Values

```python
moment = datetime(2026, 9, 20, 14, 30)
moment.year, moment.month, moment.day, moment.hour, moment.minute
moment.date()      # date(2026, 9, 20)
moment.time()      # time(14, 30)
```

---

## Arithmetic With `timedelta`

| Expression | Result |
|---|---|
| `datetime + timedelta` | A later `datetime` |
| `datetime - timedelta` | An earlier `datetime` |
| `datetime - datetime` | A `timedelta` |
| `timedelta.days` | Whole days |
| `timedelta.total_seconds()` | The whole duration in seconds |

Dates and datetimes can be compared with `<`, `>`, `==`.

---

## Format Codes (`strftime` / `strptime`)

| Code | Meaning | Example |
|---|---|---|
| `%Y` / `%y` | Year (4 / 2 digits) | `2026` / `26` |
| `%m` | Month number | `09` |
| `%B` / `%b` | Month name (full / short) | `September` / `Sep` |
| `%d` | Day of the month | `20` |
| `%A` / `%a` | Weekday name (full / short) | `Sunday` / `Sun` |
| `%H` / `%I` | Hour (24h / 12h) | `14` / `02` |
| `%M` / `%S` | Minute / second | `30` / `00` |
| `%p` | AM or PM | `PM` |
| `%j` | Day of the year | `263` |
| `%z` | UTC offset | `+0200` |

Formatting also works inside f-strings: `f"{moment:%d/%m/%Y}"`.

---

## Time Zones

* A **naive** datetime has no time zone. An **aware** datetime does.
* `datetime.now(timezone.utc)` gives the current time in UTC.
* Naive and aware datetimes **cannot be compared or subtracted**.
* Named zones (`Africa/Cairo`) come from the `zoneinfo` module. On Windows it needs `pip install tzdata`.
* `datetime.utcnow()` is deprecated since Python 3.12. Use `datetime.now(timezone.utc)`.

---

## `time`

| Function | Use |
|---|---|
| `time.time()` | Wall-clock time, for timestamps |
| `time.perf_counter()` | Measuring how long code takes |
| `time.sleep(s)` | Wait, for example between retries |

### Important

* For timestamps and dates, use `datetime`.
* Store times in **UTC** and convert only when showing them to users.

````python
import time
from datetime import date, datetime, time as clock_time, timedelta, timezone

moment = datetime(2026, 9, 20, 14, 30)

# Fields and conversions
print(moment.year, moment.month, moment.day, moment.hour, moment.minute)
print(moment.date(), moment.time(), moment.weekday(), moment.isoweekday())   # Monday is 0
print(date(2026, 9, 20), clock_time(14, 30), moment.replace(hour=0, minute=0))

# timedelta arithmetic
week = timedelta(days=7)
print(moment + week, moment - timedelta(hours=36))
difference = datetime(2026, 12, 25) - moment
print(difference.days, difference.total_seconds())
print(moment < moment + week, timedelta(hours=1) == timedelta(minutes=60))

# Formatting: strftime and f-strings
print(moment.strftime("%Y-%m-%d %H:%M"))
print(moment.strftime("%A, %d %B %Y, %I:%M %p"))
print(f"{moment:%d/%m/%Y}", moment.strftime("%j"))

# Parsing: strptime and ISO 8601
parsed = datetime.strptime("20/09/2026 14:30", "%d/%m/%Y %H:%M")
print(parsed == moment, moment.isoformat(), datetime.fromisoformat("2026-09-20T14:30:00") == moment)

try:
    datetime.strptime("2026-13-45", "%Y-%m-%d")
except ValueError as error:
    print(type(error).__name__, "for an invalid date")

# Time zones: naive vs aware
cairo = timezone(timedelta(hours=2))
aware = moment.replace(tzinfo=cairo)
print(aware, aware.utcoffset(), aware.astimezone(timezone.utc))
print(aware.strftime("%Y-%m-%d %H:%M %z"))

try:
    aware - moment
except TypeError as error:
    print(type(error).__name__, "when mixing naive and aware datetimes")

now = datetime.now(timezone.utc)
print(now.tzinfo, type(date.today()).__name__)

# time module
start = time.perf_counter()
time.sleep(0.01)
elapsed = time.perf_counter() - start
print(elapsed >= 0.01, type(time.time()).__name__)
````

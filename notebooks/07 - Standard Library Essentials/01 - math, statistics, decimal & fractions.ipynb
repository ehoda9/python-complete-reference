# Standard Library => math, statistics, decimal & fractions

Python's built-in numbers cover everyday work. These modules add **mathematical functions**, **statistics** and **exact arithmetic**.

| Tool | Purpose | Example |
|---|---|---|
| `math.sqrt(x)` | Square root | `math.sqrt(16)` gives `4.0` |
| `math.floor(x)` / `math.ceil(x)` | Round down / up | `math.floor(2.7)` gives `2` |
| `math.trunc(x)` | Cut off the decimals | `math.trunc(-2.7)` gives `-2` |
| `math.gcd(a, b)` / `math.lcm(a, b)` | Greatest common divisor / least common multiple | `math.gcd(12, 18)` gives `6` |
| `math.factorial(n)` | `n!` | `math.factorial(5)` gives `120` |
| `math.comb(n, k)` / `math.perm(n, k)` | Combinations / permutations | `math.comb(5, 2)` gives `10` |
| `math.log(x, base)` / `math.exp(x)` | Logarithm / `e**x` | `math.log(8, 2)` gives `3.0` |
| `math.sin` / `cos` / `tan` | Trigonometry (**radians**) | `math.sin(math.pi / 2)` gives `1.0` |
| `math.radians(d)` / `math.degrees(r)` | Convert angles | `math.radians(180)` gives `pi` |
| `math.hypot(*xs)` / `math.dist(p, q)` | Length / distance | `math.hypot(3, 4)` gives `5.0` |
| `math.isclose(a, b)` | Compare floats safely | `math.isclose(0.1 + 0.2, 0.3)` |
| `math.pi`, `math.e`, `math.tau`, `math.inf`, `math.nan` | Constants | |
| `math.prod(it)` / `math.fsum(it)` | Product / accurate sum | `math.prod([2, 3, 4])` gives `24` |
| `statistics.mean` / `median` / `mode` | Average, middle value, most common | |
| `statistics.stdev` / `variance` | Spread of the data | |
| `decimal.Decimal` | Exact decimal numbers (money) | `Decimal("0.1") + Decimal("0.2")` |
| `fractions.Fraction` | Exact fractions | `Fraction(1, 3) + Fraction(1, 6)` |

---

## `math`

```python
import math
```

### Rounding

| Function | `2.5` | `-2.5` | Meaning |
|---|---|---|---|
| `math.floor` | `2` | `-3` | Toward negative infinity |
| `math.ceil` | `3` | `-2` | Toward positive infinity |
| `math.trunc` | `2` | `-2` | Toward zero |
| `round` (built-in) | `2` | `-2` | To the nearest **even** number on ties |

### Important

* `round(2.5)` is `2` and `round(3.5)` is `4`. This is **banker's rounding**, not "always round half up".
* Trigonometric functions use **radians**. Convert with `math.radians()`.
* `math.sqrt(-1)` raises `ValueError`. Use `cmath` for complex results.

---

## Floats Are Approximate

```python
0.1 + 0.2 == 0.3                  # False
math.isclose(0.1 + 0.2, 0.3)      # True
```

Never compare floats with `==`. Use `math.isclose()`.

### `inf` and `nan`

* `math.inf` is larger than every number.
* `math.nan` is "not a number". It is **not equal to itself**, so test it with `math.isnan(x)`.

---

## `statistics`

```python
import statistics
data = [2, 4, 4, 4, 5, 5, 7, 9]
```

| Function | Result for `data` | Meaning |
|---|---|---|
| `mean(data)` | `5` | Average |
| `median(data)` | `4.5` | Middle value |
| `mode(data)` | `4` | Most common value |
| `stdev(data)` | `2.138...` | **Sample** standard deviation |
| `pstdev(data)` | `2.0` | **Population** standard deviation |
| `variance(data)` | `4.571...` | Sample variance |
| `fmean(data)` | `5.0` | Faster mean, always a float |

Empty data raises `statistics.StatisticsError`.

---

## `decimal`: Exact Decimal Arithmetic

```python
from decimal import Decimal
Decimal("0.1") + Decimal("0.2")     # Decimal('0.3')
```

* Create a `Decimal` from a **string**, not from a float.
* Use it for **money** and any case where exact decimals matter.

---

## `fractions`: Exact Fractions

```python
from fractions import Fraction
Fraction(1, 3) + Fraction(1, 6)     # Fraction(1, 2)
```

Fractions are always kept in lowest terms.

````python
import math
import statistics
from decimal import Decimal
from fractions import Fraction

# Constants
print(round(math.pi, 5), round(math.e, 5), round(math.tau, 5), math.inf > 10**100)

# Powers, roots, logarithms
print(math.sqrt(16), math.pow(2, 10), 2 ** 10, math.isqrt(17))
print(math.log(8, 2), round(math.log10(1000), 6), round(math.exp(1), 5))

# Rounding: floor, ceil, trunc, round
for value in (2.5, -2.5):
    print(value, math.floor(value), math.ceil(value), math.trunc(value), round(value))
print(round(3.5), round(4.5), round(2.675, 2))          # banker's rounding and float error

# Integer helpers
print(math.gcd(12, 18), math.lcm(4, 6), math.factorial(5), math.comb(5, 2), math.perm(5, 2))
print(math.prod([2, 3, 4]), math.fsum([0.1] * 10))

# Trigonometry uses radians
print(round(math.sin(math.pi / 2), 6), round(math.cos(0), 6), math.degrees(math.pi), round(math.radians(180), 5))
print(math.hypot(3, 4), math.dist((0, 0), (3, 4)))

# Comparing floats
print(0.1 + 0.2 == 0.3, math.isclose(0.1 + 0.2, 0.3))

# inf and nan
print(math.isnan(math.nan), math.nan == math.nan, math.isinf(-math.inf))

# Domain errors
try:
    math.sqrt(-1)
except ValueError as error:
    print("ValueError:", error)

# statistics
data = [2, 4, 4, 4, 5, 5, 7, 9]
print(statistics.mean(data), statistics.median(data), statistics.mode(data))
print(round(statistics.stdev(data), 3), statistics.pstdev(data), round(statistics.variance(data), 3))
print(statistics.fmean(data), statistics.multimode([1, 1, 2, 2, 3]))

try:
    statistics.mean([])
except statistics.StatisticsError as error:
    print("StatisticsError:", error)

# decimal: exact decimal arithmetic
print(0.1 + 0.2, Decimal("0.1") + Decimal("0.2"))
print(Decimal("19.99") * 3, Decimal(0.1) == Decimal("0.1"))      # a float carries its own error

# fractions: exact fractions in lowest terms
print(Fraction(1, 3) + Fraction(1, 6), Fraction(6, 8), Fraction("0.25"), float(Fraction(1, 4)))
````

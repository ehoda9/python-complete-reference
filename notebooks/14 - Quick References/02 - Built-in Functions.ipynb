# Built-in Functions => General & Inspection

Python provides many built-in functions that are available directly without importing a module.

## Common General Functions

| Function     | Purpose                         | Example           | Important Note                                     |
| ------------ | ------------------------------- | ----------------- | -------------------------------------------------- |
| `print()`    | Display output                  | `print("Hello")`  | Returns `None`                                     |
| `input()`    | Get user input                  | `input("Name: ")` | Always returns `str`                               |
| `len()`      | Get number of items             | `len([1, 2, 3])`  | Object must have a length                          |
| `type()`     | Get an object's type            | `type(10)`        | `isinstance()` is usually better for type checking |
| `id()`       | Get object identity             | `id(x)`           | Do not use it to compare values                    |
| `dir()`      | List available names/attributes | `dir("Python")`   | Very useful for exploration                        |
| `help()`     | Show documentation              | `help(len)`       | Mainly useful interactively                        |
| `callable()` | Check if object can be called   | `callable(print)` | Returns `True` / `False`                           |

## `all()`

Returns `True` if **all elements are Truthy**.

`all([True, 1, "Python"])` => `True`

`all([True, 0, "Python"])` => `False`

An empty iterable returns `True`.

## `any()`

Returns `True` if **at least one element is Truthy**.

`any([False, 0, "Python"])` => `True`

`any([False, 0, ""])` => `False`

An empty iterable returns `False`.

## `id()`

Returns an integer representing the identity of an object during its lifetime.

`id(object)`

Use `==` to compare values, not `id()`.

## `type()` vs `isinstance()`

`type(10) == int`

checks the exact type.

`isinstance(10, int)`

checks whether the object is an instance of a type, including subclasses.

For normal type checking, `isinstance()` is usually preferred.

## Common Mistakes

* `len(100)` => `TypeError`
* `input()` returns a string even when the user enters a number.
* `all()` checks Truthy/Falsy values; it does not require literal `True`.
* `any()` returns a Boolean, not the matching element.
* `id()` is not a value comparison tool.
* `dir()` shows available names; it does not execute them.


```python
# -----------------------------------
# -- General & Inspection Functions --
# -----------------------------------

# print(*objects, sep=" ", end="\n") => display output
print("Hello", "Python")

# input(prompt="") => get user input as str
# name = input("Enter your name: ")

# len(object) => return number of items
len([10, 20, 30])
len("Python")

# type(object) => return object's type
type(10)
type("Python")

# id(object) => return object's identity
x = 10
id(x)

# dir(object=None) => list available names/attributes
dir("Python")

# help(request=None) => display documentation
# help(len)

# callable(object) => check whether object can be called
callable(print)
callable(10)

# all(iterable) => True if every item is Truthy
all([True, 1, "Python"])
all([True, 0, "Python"])
all([])

# any(iterable) => True if at least one item is Truthy
any([False, 0, "Python"])
any([False, 0, ""])
any([])

# isinstance(object, classinfo) => check object type
isinstance(10, int)
isinstance("Python", str)

# type(object) == type => exact type check
type(10) == int
```

# Built-in Functions => Conversion & Creation

Many built-ins are also built-in types. They can create objects or convert values.

| Function / Type | Purpose                          | Example                |
| --------------- | -------------------------------- | ---------------------- |
| `bool()`        | Convert to Boolean               | `bool(1)`              |
| `int()`         | Convert to integer               | `int("10")`            |
| `float()`       | Convert to float                 | `float("10.5")`        |
| `complex()`     | Create complex number            | `complex(2, 3)`        |
| `str()`         | Convert to string                | `str(100)`             |
| `list()`        | Create list                      | `list("ABC")`          |
| `tuple()`       | Create tuple                     | `tuple([1, 2])`        |
| `set()`         | Create set                       | `set([1, 2, 2])`       |
| `frozenset()`   | Create immutable set             | `frozenset([1, 2])`    |
| `dict()`        | Create dictionary                | `dict(name="Mahmoud")` |
| `bytes()`       | Create immutable bytes           | `bytes([65, 66])`      |
| `bytearray()`   | Create mutable bytes             | `bytearray([65, 66])`  |
| `memoryview()`  | View binary data without copying | `memoryview(b"ABC")`   |
| `slice()`       | Create a slice object            | `slice(1, 5, 2)`       |

These were already introduced in the Type Conversion section, so this section focuses on their additional/common uses.

## Important Notes

* `int("10.5")` => `ValueError`
* `int(10.9)` => `10`; it does not round.
* `list()` on a string creates one item per character.
* `set()` removes duplicate elements.
* `dict()` requires valid key-value pairs when created from an iterable.
* `bytes` is immutable.
* `bytearray` is mutable.
* `memoryview` provides a view of binary data without copying it.
* `slice(start, stop, step)` creates a slice object that can be used as an index: `[10, 20, 30, 40, 50][slice(1, 4)]` => `[20, 30, 40]`.


```python
# --------------------------------
# -- Conversion & Creation --
# --------------------------------

# bool(value) => convert value to bool
bool(1)
bool(0)
bool("Python")

# int(value, base=10) => convert to integer
int("10")
int("101", 2)

# float(value) => convert to floating-point number
float("10.5")

# complex(real=0, imag=0) => create complex number
complex(2, 3)
complex("2+3j")

# str(object="") => convert object to string
str(100)
str(True)

# list(iterable) => create a list
list("ABC")
list((1, 2, 3))

# tuple(iterable) => create a tuple
tuple([1, 2, 3])

# set(iterable) => create a set
set([1, 2, 2, 3])

# frozenset(iterable) => create immutable set
frozenset([1, 2, 2, 3])

# dict(...) => create dictionary
dict(name="Mahmoud", age=20)
dict([("name", "Mahmoud"), ("age", 20)])

# bytes(source) => create immutable bytes
bytes([65, 66, 67])

# bytearray(source) => create mutable bytearray
bytearray([65, 66, 67])

# memoryview(object) => create binary memory view
memoryview(b"Python")

# slice(start, stop, step) => create a slice object
slice(1, 5, 2)
[10, 20, 30, 40, 50, 60][slice(1, 5, 2)]
```

# Built-in Functions => Numbers & Mathematics

Python provides several built-ins for common numeric operations.

| Function   | Purpose              | Example          | Result   |
| ---------- | -------------------- | ---------------- | -------- |
| `abs()`    | Absolute value       | `abs(-10)`       | `10`     |
| `round()`  | Round a number       | `round(10.6)`    | `11`     |
| `pow()`    | Power calculation    | `pow(2, 3)`      | `8`      |
| `divmod()` | Quotient + remainder | `divmod(10, 3)`  | `(3, 1)` |
| `sum()`    | Add iterable values  | `sum([1, 2, 3])` | `6`      |
| `min()`    | Smallest value       | `min(3, 1, 5)`   | `1`      |
| `max()`    | Largest value        | `max(3, 1, 5)`   | `5`      |
| `bin()`    | Binary string        | `bin(10)`        | `'0b1010'` |
| `oct()`    | Octal string         | `oct(8)`         | `'0o10'`  |
| `hex()`    | Hexadecimal string   | `hex(255)`       | `'0xff'`  |

## `abs()`

Returns the absolute value.

`abs(-10)` => `10`

For complex numbers, it returns the magnitude.

## `round()`

Rounds a number.

`round(10.567, 2)` => `10.57`

Be careful with `.5`: Python uses **round half to even** in applicable cases.

## `pow()`

`pow(base, exponent)`

is similar to:

`base ** exponent`

It also supports:

`pow(base, exponent, modulus)`

which calculates the modular power efficiently.

## `divmod()`

Returns a tuple containing:

`(quotient, remainder)`

Example:

`divmod(10, 3)` => `(3, 1)`

## `sum()`

Adds values from an iterable.

`sum([1, 2, 3])` => `6`

The optional `start` value is added to the result.

## `min()` and `max()`

Find the smallest/largest item.

They support `key=` for custom comparison and `default=` when the iterable may be empty.

## `bin()`, `oct()` and `hex()`

Return an integer as a string in base 2, 8 or 16 (prefix `0b`, `0o` or `0x`).

`bin(10)` => `'0b1010'`

`oct(8)` => `'0o10'`

`hex(255)` => `'0xff'`

Use `int(text, base)` to convert back: `int("ff", 16)` => `255`.

### Common Mistakes

* `min([])` / `max([])` => `ValueError`
* `sum(["a", "b"])` => `TypeError`
* `round()` does not always behave like manual `.5 => up` rounding.
* `abs()` does not modify the original value.


```python
# -----------------------------
# -- Numbers & Math Functions --
# -----------------------------

# abs(number) => absolute value
abs(-10)
abs(-3.5)

# round(number, ndigits=None) => round number
round(10.6)
round(10.5678, 2)

# pow(base, exp, mod=None) => calculate power
pow(2, 3)
pow(2, 3, 5)

# divmod(a, b) => return (quotient, remainder)
divmod(10, 3)

# sum(iterable, start=0) => sum values
sum([1, 2, 3])
sum([1, 2, 3], 10)

# min(iterable, *, key=None, default=...) => smallest item
min([10, 5, 20])

# max(iterable, *, key=None, default=...) => largest item
max([10, 5, 20])

# min/max with key => compare using a custom rule
names = ["Ali", "Mahmoud", "Mona"]
min(names, key=len)
max(names, key=len)

# max with default => avoid ValueError on empty iterable
max([], default=0)

# bin(x) / oct(x) / hex(x) => integer as a binary / octal / hexadecimal string
bin(10)
oct(8)
hex(255)
```

# Built-in Functions => Iteration & Functional

These functions are especially useful when working with loops, iterables, and functions.

| Function      | Purpose                                   |
| ------------- | ----------------------------------------- |
| `enumerate()` | Add an index while iterating              |
| `zip()`       | Combine multiple iterables                |
| `map()`       | Apply a function to items                 |
| `filter()`    | Keep items that satisfy a condition       |
| `iter()`      | Create an iterator                        |
| `next()`      | Get the next item from an iterator        |
| `reversed()`  | Iterate in reverse                        |
| `sorted()`    | Return a sorted list                      |
| `range()`     | Generate an immutable sequence of numbers |

## `enumerate()`

Useful when you need both the index and the value.

`enumerate(iterable, start=0)`

## `zip()`

Combines items from multiple iterables.

`zip(iterable1, iterable2, ...)`

By default, it stops when the shortest iterable ends.

## `map()`

Applies a function to every item.

`map(function, iterable, ...)`

It returns an iterator, so `list()` is commonly used to see all results.

## `filter()`

Keeps elements for which the function returns Truthy.

`filter(function, iterable)`

## `iter()` and `next()`

`iter()` creates an iterator.

`next()` gets the next item.

When an iterator is exhausted:

`next(iterator)` => `StopIteration`

You can provide a default:

`next(iterator, default)`

## `sorted()`

Returns a **new sorted list**.

It does not modify the original iterable.

Supports:

* `key=`
* `reverse=`

## `reversed()`

Returns a reverse iterator.

It does not modify the original object.

## `range()`

Creates an immutable sequence of numbers.

`range(stop)`

`range(start, stop, step)`

The `stop` value is excluded.

### Important

`range()` is a built-in sequence type, not a normal function.

### Common Mistakes

* `map()` and `filter()` return iterators.
* `zip()` also returns an iterator.
* `sorted()` creates a new list.
* `list.sort()` modifies the list in place.
* `range(5)` produces `0` through `4`, not `5`.
* `next()` can raise `StopIteration`.


```python
# ----------------------------------
# -- Iteration & Functional Tools --
# ----------------------------------

# enumerate(iterable, start=0) => index + value
names = ["Ahmed", "Mona", "Ali"]

for index, name in enumerate(names, start=1):
    pass

# zip(*iterables, strict=False) => combine iterables
names = ["Ahmed", "Mona", "Ali"]
ages = [20, 21, 19]

for name, age in zip(names, ages):
    pass

# map(function, iterable, *iterables) => transform items
numbers = [1, 2, 3, 4]

doubled = map(lambda x: x * 2, numbers)
list(doubled)

# filter(function, iterable) => keep items where function is Truthy
numbers = [1, 2, 3, 4, 5]

even_numbers = filter(lambda x: x % 2 == 0, numbers)
list(even_numbers)

# iter(iterable) => create iterator
numbers = [10, 20, 30]
iterator = iter(numbers)

# next(iterator, default=None) => get next item
next(iterator)
next(iterator)
next(iterator)
next(iterator, "No more items")

# reversed(sequence) => reverse iterator
numbers = [1, 2, 3, 4]
list(reversed(numbers))

# sorted(iterable, *, key=None, reverse=False) => new sorted list
numbers = [5, 2, 8, 1]
sorted(numbers)
sorted(numbers, reverse=True)

# sorted() with key
names = ["Ali", "Mahmoud", "Mona"]
sorted(names, key=len)

# range(stop) => numbers from 0 to stop - 1
range(5)

# range(start, stop, step=1) => numeric sequence
list(range(2, 10, 2))
list(range(5, 0, -1))
```

# Built-in Functions => Objects & Attributes

These functions allow you to inspect and manipulate object attributes dynamically.

| Function    | Purpose                           |
| ----------- | --------------------------------- |
| `getattr()` | Get an attribute                  |
| `setattr()` | Set/create an attribute           |
| `delattr()` | Delete an attribute               |
| `hasattr()` | Check whether an attribute exists |
| `vars()`    | Get an object's `__dict__`        |
| `globals()` | Get the global namespace          |
| `locals()`  | Get the current local namespace   |
| `hash()`    | Get an object's hash value        |

## `getattr()`

Gets an attribute dynamically.

`getattr(object, name)`

It can also receive a default:

`getattr(object, name, default)`

## `setattr()`

Creates or changes an attribute.

`setattr(object, name, value)`

## `delattr()`

Deletes an attribute.

`delattr(object, name)`

## `hasattr()`

Checks whether an object has an attribute.

`hasattr(object, name)`

## `vars()`

Without an argument, returns the current local namespace.

With an object, it normally returns the object's `__dict__`.

## `globals()` / `locals()`

Return mappings representing the global/local namespaces.

These are mainly useful for introspection and advanced code.

## `hash()`

Returns an object's hash value.

Hashable objects can be used as:

* dictionary keys
* set elements

Examples:

`hash("Python")`

`hash((1, 2, 3))`

Mutable objects such as lists are not hashable.

`hash([1, 2, 3])` => `TypeError`

### Important

Do not assume hash values are permanent across Python runs.


```python
# ---------------------------------
# -- Objects & Attribute Functions --
# ---------------------------------

class User:
    pass

user = User()

# setattr(object, name, value) => create/change attribute
setattr(user, "name", "Mahmoud")
setattr(user, "age", 20)

# getattr(object, name, default=None) => get attribute
getattr(user, "name")
getattr(user, "email", "Not Found")

# hasattr(object, name) => check attribute
hasattr(user, "name")
hasattr(user, "email")

# delattr(object, name) => delete attribute
delattr(user, "age")

# vars(object=None) => return object's __dict__
vars(user)

# globals() => return global namespace
globals()

# locals() => return current local namespace
locals()

# hash(object) => return hash value
hash("Python")
hash((1, 2, 3))

# hash([1, 2, 3])  # TypeError: list is unhashable
```

# Built-in Functions => Strings & Representation

These functions are useful for converting characters, formatting values, and representing objects.

| Function   | Purpose                           | Example               |
| ---------- | --------------------------------- | --------------------- |
| `chr()`    | Unicode code point => character   | `chr(65)` => `'A'`    |
| `ord()`    | Character => Unicode code point   | `ord('A')` => `65`    |
| `ascii()`  | ASCII-safe representation         | `ascii("café")`       |
| `repr()`   | Developer-oriented representation | `repr("Hello")`       |
| `format()` | Format a value                    | `format(10.5, ".2f")` |

## `chr()`

Converts a Unicode code point into a character.

`chr(65)` => `"A"`

Valid code points range from `0` to `0x10FFFF`.

## `ord()`

Does the opposite:

`ord("A")` => `65`

The input must contain exactly one character when using a string.

## `ascii()`

Returns a printable representation using ASCII characters where possible, escaping non-ASCII characters.

## `repr()`

Returns a representation intended mainly for developers/debugging.

It is different from `str()`.

## `format()`

Formats a value according to a format specification.

`format(value, format_spec="")`

Examples:

`format(10.567, ".2f")` => `"10.57"`

`format(255, "x")` => `"ff"`

### Common Mistakes

* `ord("AB")` => `TypeError`
* `chr(-1)` => `ValueError`
* `chr(0x110000)` => `ValueError`
* `repr()` returns a string; it does not return the original object.
* `format()` returns a string.


```python
# --------------------------------
# -- String & Representation --
# --------------------------------

# chr(codepoint) => Unicode number to character
chr(65)
chr(8364)

# ord(character) => character to Unicode number
ord("A")
ord("€")

# ascii(object) => ASCII representation
ascii("café")
ascii("Python")

# repr(object) => developer-oriented representation
repr("Hello\nPython")
repr([1, 2, 3])

# format(value, format_spec="") => format value as string
format(10.5678, ".2f")
format(255, "x")
format(0.75, ".0%")
```

# Built-in Functions => Files & Dynamic Execution

Some built-ins interact with files or dynamically execute Python code.

| Function       | Purpose                      |
| -------------- | ---------------------------- |
| `open()`       | Open a file                  |
| `compile()`    | Compile source code          |
| `eval()`       | Evaluate a Python expression |
| `exec()`       | Execute Python statements    |
| `breakpoint()` | Enter the debugger           |
| `__import__()` | Low-level import mechanism   |

## `open()`

Opens a file and returns a file object.

`open(file, mode="r", encoding=None)`

Common modes:

* `"r"` => read
* `"w"` => write / truncate
* `"a"` => append
* `"x"` => create
* `"b"` => binary
* `"t"` => text
* `"+"` => read and write

Use `with open(...)` whenever possible.

## `compile()`

Compiles source code into a code object.

`compile(source, filename, mode)`

Modes:

* `"eval"` => expression
* `"exec"` => statements
* `"single"` => single interactive statement

## `eval()`

Evaluates a Python **expression**.

`eval(source, globals=None, locals=None)`

### Security Warning

Never use `eval()` on untrusted user input.

## `exec()`

Executes Python statements.

`exec(source, globals=None, locals=None)`

### Security Warning

Never execute untrusted input with `exec()`.

## `breakpoint()`

Starts the Python debugger.

`breakpoint()`

It is mainly useful while debugging.

## `__import__()`

Low-level function used internally by Python's import machinery.

Normal code should use:

`import module`

instead of calling `__import__()` directly.

### Important

`eval()` and `exec()` are powerful but dangerous when handling untrusted input.


```python
# -----------------------------------------
# -- Files & Dynamic Execution --
# -----------------------------------------

# open(file, mode="r", encoding=None) => open file
with open("data.txt", "w", encoding="utf-8") as file:
    file.write("Hello Python")

# compile(source, filename, mode) => compile source
code = compile("10 + 20", "<string>", "eval")

# eval(source, globals=None, locals=None) => evaluate expression
result = eval("10 + 20")

# NEVER use eval() with untrusted user input.
# user_input = input("Expression: ")
# eval(user_input)

# exec(source, globals=None, locals=None) => execute statements
code = """
x = 10
y = 20
result = x + y
"""

namespace = {}
exec(code, namespace)

# NEVER use exec() with untrusted user input.
# exec(user_input)

# breakpoint(*args, **kwargs) => start debugger
# breakpoint()

# __import__(name, globals=None, locals=None, fromlist=(), level=0)
# => low-level import mechanism
math_module = __import__("math")
```

# Built-in Functions => Classes & OOP

Some built-ins are especially important when working with classes and objects.

| Built-in         | Purpose                           |
| ---------------- | --------------------------------- |
| `object()`       | Create a base object              |
| `type()`         | Inspect/create a type             |
| `isinstance()`   | Check object type                 |
| `issubclass()`   | Check class inheritance           |
| `property()`     | Create managed attributes         |
| `classmethod()`  | Create class methods              |
| `staticmethod()` | Create static methods             |
| `super()`        | Access parent/superclass behavior |

## `object()`

Creates a basic object.

`object()`

It is also the base class of Python's class hierarchy.

## `isinstance()`

Checks whether an object is an instance of a class.

`isinstance(object, classinfo)`

It also considers inheritance.

## `issubclass()`

Checks whether one class inherits from another.

`issubclass(class, classinfo)`

## `type()`

One argument:

`type(object)` => inspect type.

Three arguments:

`type(name, bases, namespace)` => dynamically create a class.

## `property()`

Creates a property that behaves like an attribute but can execute getter/setter logic.

`property(fget=None, fset=None, fdel=None, doc=None)`

Usually used through the `@property` decorator.

## `classmethod()`

Makes a method receive the class (`cls`) instead of the instance.

## `staticmethod()`

Creates a method that does not receive an automatic `self` or `cls`.

## `super()`

Returns a proxy used to access methods from a parent/superclass.

### Important

`classmethod()` and `staticmethod()` are usually used as decorators:

`@classmethod`

`@staticmethod`


```python
# -----------------------------
# -- Classes & OOP Functions --
# -----------------------------

# object() => create basic object
obj = object()

# isinstance(object, classinfo) => check instance type
isinstance(10, int)
isinstance("Python", str)

class Animal:
    pass

class Dog(Animal):
    pass

dog = Dog()

isinstance(dog, Dog)
isinstance(dog, Animal)

# issubclass(class, classinfo) => check inheritance
issubclass(Dog, Animal)
issubclass(Dog, object)

# type(object) => get object's type
type(dog)

# type(name, bases, namespace) => dynamically create class
User = type(
    "User",
    (),
    {"name": "Mahmoud"}
)

user = User()
user.name


# property(fget=None, fset=None, fdel=None, doc=None)
# => create managed attribute

class Person:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

person = Person("Mahmoud")
person.name


# classmethod(function) => method receives cls
class User:
    count = 0

    @classmethod
    def show_count(cls):
        return cls.count

User.show_count()


# staticmethod(function) => method without self/cls
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

Calculator.add(10, 20)


# super(type=None, object_or_type=None) => access parent behavior
class Animal:

    def speak(self):
        return "Animal sound"


class Dog(Animal):

    def speak(self):
        return super().speak()

dog = Dog()
dog.speak()
```

# Built-in Functions => Async

Python also provides built-ins for asynchronous iteration.

These are mainly used with:

* `async`
* `await`
* asynchronous iterators

## `aiter()`

Returns an asynchronous iterator from an asynchronous iterable.

`aiter(async_iterable)`

Unlike `iter()`, it does not have a two-argument sentinel form.

## `anext()`

Returns the next item from an asynchronous iterator when awaited.

`anext(async_iterator)`

It can also receive a default:

`anext(async_iterator, default)`

Without a default, an exhausted iterator raises `StopAsyncIteration`.

### Important

These functions are advanced and are normally used together with:

`async def`

`await`

`async for`

They are not needed for normal beginner-level loops.


| Category          | Built-ins                                                                                                                 |
| ----------------- | ------------------------------------------------------------------------------------------------------------------------- |
| General           | `print`, `input`, `len`, `type`, `id`, `dir`, `help`, `callable`, `all`, `any`                                            |
| Conversion        | `bool`, `int`, `float`, `complex`, `str`, `list`, `tuple`, `set`, `frozenset`, `dict`, `bytes`, `bytearray`, `memoryview`, `slice` |
| Math              | `abs`, `round`, `pow`, `divmod`, `sum`, `min`, `max`, `bin`, `oct`, `hex`                                                                      |
| Iteration         | `enumerate`, `zip`, `map`, `filter`, `iter`, `next`, `reversed`, `sorted`, `range`                                        |
| Attributes        | `getattr`, `setattr`, `delattr`, `hasattr`, `vars`, `globals`, `locals`, `hash`                                           |
| Representation    | `chr`, `ord`, `ascii`, `repr`, `format`                                                                                   |
| Files / Execution | `open`, `compile`, `eval`, `exec`, `breakpoint`, `__import__`                                                             |
| OOP               | `object`, `isinstance`, `issubclass`, `property`, `classmethod`, `staticmethod`, `super`                                  |
| Async             | `aiter`, `anext`                                                                                                          |


```python
# -----------------------
# -- Async Built-ins --
# -----------------------

#aiter(async_iterable) => create async iterator
async_iterator = aiter(async_iterable)

#anext(async_iterator) => get next async item
value = await anext(async_iterator)

# anext(async_iterator, default) => default if exhausted
value = await anext(async_iterator, "Done")
```

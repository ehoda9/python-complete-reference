# Quick References => Methods of Built-in Types

Every public method of `str`, `list`, `tuple`, `dict` and `set`, grouped by what it does. The **Result** column is what the example returns. The examples also run in the code cell below.

| Type | Methods | What they do |
|---|---|---|
| `str` | 47 | Text: case, search, checks, cleaning, splitting, alignment |
| `list` | 11 | Ordered and changeable: add, remove, find, order, copy |
| `tuple` | 2 | Ordered and fixed: count and index |
| `dict` | 11 | Key-value pairs: read, change, create |
| `set` | 17 | Unique values: add and remove, combine, test |

## `str` methods

### Case

| Method | What it does | Example | Result |
|---|---|---|---|
| `.capitalize()` | First character upper case, the rest lower case | `"hello WORLD".capitalize()` | `'Hello world'` |
| `.casefold()` | Aggressive lower case, for case-insensitive comparison | `"Hello".casefold()` | `'hello'` |
| `.lower()` | All lower case | `"Hello".lower()` | `'hello'` |
| `.upper()` | All upper case | `"Hello".upper()` | `'HELLO'` |
| `.swapcase()` | Swap upper and lower case | `"Hello".swapcase()` | `'hELLO'` |
| `.title()` | First letter of every word upper case | `"hello world".title()` | `'Hello World'` |

### Search and count

| Method | What it does | Example | Result |
|---|---|---|---|
| `.count()` | Count non-overlapping occurrences of a substring | `"banana".count("an")` | `2` |
| `.find()` | Index of the first occurrence, or -1 | `"banana".find("na")` | `2` |
| `.rfind()` | Index of the last occurrence, or -1 | `"banana".rfind("na")` | `4` |
| `.index()` | Like `find`, but raises `ValueError` if missing | `"banana".index("na")` | `2` |
| `.rindex()` | Like `rfind`, but raises `ValueError` if missing | `"banana".rindex("na")` | `4` |
| `.startswith()` | Does the string start with a prefix (or a tuple of prefixes)? | `"report.csv".startswith("rep")` | `True` |
| `.endswith()` | Does the string end with a suffix (or a tuple of suffixes)? | `"report.csv".endswith((".csv", ".txt"))` | `True` |

### Check the content (return True or False)

| Method | What it does | Example | Result |
|---|---|---|---|
| `.isalnum()` | Only letters and digits | `"abc123".isalnum()` | `True` |
| `.isalpha()` | Only letters | `"abc".isalpha()` | `True` |
| `.isascii()` | Only ASCII characters | `"abc".isascii()` | `True` |
| `.isdecimal()` | Only decimal digits (0-9 and other scripts) | `"123".isdecimal()` | `True` |
| `.isdigit()` | Only digits | `"123".isdigit()` | `True` |
| `.isidentifier()` | A valid Python name | `"my_var".isidentifier()` | `True` |
| `.islower()` | Cased letters are all lower case | `"abc".islower()` | `True` |
| `.isnumeric()` | Only numeric characters | `"123".isnumeric()` | `True` |
| `.isprintable()` | No control characters such as newline | `"abc".isprintable()` | `True` |
| `.isspace()` | Only whitespace | `"  ".isspace()` | `True` |
| `.istitle()` | Title case: every word starts with a capital | `"Hello World".istitle()` | `True` |
| `.isupper()` | Cased letters are all upper case | `"ABC".isupper()` | `True` |

### Clean and replace

| Method | What it does | Example | Result |
|---|---|---|---|
| `.strip()` | Remove characters from both ends (whitespace by default) | `"  hi  ".strip()` | `'hi'` |
| `.lstrip()` | Remove characters from the start | `"xxhixx".lstrip("x")` | `'hixx'` |
| `.rstrip()` | Remove characters from the end | `"xxhixx".rstrip("x")` | `'xxhi'` |
| `.removeprefix()` | Remove a prefix if it is there | `"test_file".removeprefix("test_")` | `'file'` |
| `.removesuffix()` | Remove a suffix if it is there | `"file.txt".removesuffix(".txt")` | `'file'` |
| `.replace()` | Replace every occurrence (optional maximum count) | `"a-b-c".replace("-", "+")` | `'a+b+c'` |
| `.expandtabs()` | Replace tab characters with spaces | `"a\tb".expandtabs(4)` | `'a   b'` |
| `.maketrans()` | Build a translation table for `translate` | `str.maketrans("ab", "xy")` | `{97: 120, 98: 121}` |
| `.translate()` | Replace characters using a translation table | `"aabbcc".translate(str.maketrans("ab", "xy"))` | `'xxyycc'` |

### Split and join

| Method | What it does | Example | Result |
|---|---|---|---|
| `.split()` | Split into a list (whitespace by default) | `"a,b,c".split(",")` | `['a', 'b', 'c']` |
| `.rsplit()` | Like `split`, but starting from the right | `"a,b,c".rsplit(",", 1)` | `['a,b', 'c']` |
| `.splitlines()` | Split at line breaks | `"one\ntwo".splitlines()` | `['one', 'two']` |
| `.partition()` | Split at the first separator into three parts | `"key=value".partition("=")` | `('key', '=', 'value')` |
| `.rpartition()` | Split at the last separator into three parts | `"a.b.c".rpartition(".")` | `('a.b', '.', 'c')` |
| `.join()` | Join the items of an iterable with this string | `", ".join(["a", "b", "c"])` | `'a, b, c'` |

### Align and pad

| Method | What it does | Example | Result |
|---|---|---|---|
| `.center()` | Center in a given width | `"hi".center(6, "*")` | `'**hi**'` |
| `.ljust()` | Left-align in a given width | `"hi".ljust(5, ".")` | `'hi...'` |
| `.rjust()` | Right-align in a given width | `"hi".rjust(5, ".")` | `'...hi'` |
| `.zfill()` | Pad with zeros on the left | `"42".zfill(5)` | `'00042'` |

### Format and encode

| Method | What it does | Example | Result |
|---|---|---|---|
| `.format()` | Fill `{}` placeholders | `"{} is {}".format("Ann", 30)` | `'Ann is 30'` |
| `.format_map()` | Fill placeholders from a mapping | `"{name} is {age}".format_map({"name": "Ann", "age": 30})` | `'Ann is 30'` |
| `.encode()` | Convert to bytes | `"cafe".encode("utf-8")` | `b'cafe'` |

## `list` methods

### Add

| Method | What it does | Example | Result |
|---|---|---|---|
| `.append()` | Add one item at the end | `items = [1, 2]; items.append(3); items` | `[1, 2, 3]` |
| `.extend()` | Add all items of an iterable at the end | `items = [1, 2]; items.extend([3, 4]); items` | `[1, 2, 3, 4]` |
| `.insert()` | Insert an item before an index | `items = [1, 3]; items.insert(1, 2); items` | `[1, 2, 3]` |

### Remove

| Method | What it does | Example | Result |
|---|---|---|---|
| `.pop()` | Remove and return the item at an index (the last by default) | `items = [1, 2, 3]; items.pop()` | `3` |
| `.remove()` | Remove the first item equal to a value | `items = [1, 2, 3]; items.remove(2); items` | `[1, 3]` |
| `.clear()` | Remove all items | `items = [1, 2, 3]; items.clear(); items` | `[]` |

### Find and count

| Method | What it does | Example | Result |
|---|---|---|---|
| `.index()` | Index of the first item equal to a value | `[10, 20, 30].index(20)` | `1` |
| `.count()` | How many items equal a value | `[1, 2, 2, 3].count(2)` | `2` |

### Order

| Method | What it does | Example | Result |
|---|---|---|---|
| `.sort()` | Sort in place (`key=` and `reverse=` are optional) | `items = [3, 1, 2]; items.sort(); items` | `[1, 2, 3]` |
| `.reverse()` | Reverse in place | `items = [1, 2, 3]; items.reverse(); items` | `[3, 2, 1]` |

### Copy

| Method | What it does | Example | Result |
|---|---|---|---|
| `.copy()` | Shallow copy | `items = [1, 2, 3]; items.copy()` | `[1, 2, 3]` |

## `tuple` methods

### Tuples have two methods

| Method | What it does | Example | Result |
|---|---|---|---|
| `.count()` | How many items equal a value | `(1, 2, 2, 3).count(2)` | `2` |
| `.index()` | Index of the first item equal to a value | `(10, 20, 30).index(20)` | `1` |

## `dict` methods

### Read

| Method | What it does | Example | Result |
|---|---|---|---|
| `.get()` | Value for a key, or a default instead of an error | `person = {"name": "Ann"}; person.get("age", 0)` | `0` |
| `.keys()` | View of the keys | `{"a": 1, "b": 2}.keys()` | `dict_keys(['a', 'b'])` |
| `.values()` | View of the values | `{"a": 1, "b": 2}.values()` | `dict_values([1, 2])` |
| `.items()` | View of (key, value) pairs | `{"a": 1, "b": 2}.items()` | `dict_items([('a', 1), ('b', 2)])` |

### Change

| Method | What it does | Example | Result |
|---|---|---|---|
| `.update()` | Add or overwrite items from another mapping | `person = {"a": 1}; person.update({"b": 2}); person` | `{'a': 1, 'b': 2}` |
| `.setdefault()` | Get a value, inserting a default if the key is missing | `person = {}; person.setdefault("tags", []); person` | `{'tags': []}` |
| `.pop()` | Remove a key and return its value | `person = {"a": 1, "b": 2}; person.pop("a")` | `1` |
| `.popitem()` | Remove and return the last inserted pair | `person = {"a": 1, "b": 2}; person.popitem()` | `('b', 2)` |
| `.clear()` | Remove all items | `person = {"a": 1}; person.clear(); person` | `{}` |

### Create and copy

| Method | What it does | Example | Result |
|---|---|---|---|
| `.copy()` | Shallow copy | `person = {"a": 1}; person.copy()` | `{'a': 1}` |
| `.fromkeys()` | New dict from keys, all with the same value | `dict.fromkeys(["a", "b"], 0)` | `{'a': 0, 'b': 0}` |

## `set` methods

### Add and remove

| Method | What it does | Example | Result |
|---|---|---|---|
| `.add()` | Add one element | `letters = {"a"}; letters.add("b"); sorted(letters)` | `['a', 'b']` |
| `.discard()` | Remove an element if it is present (no error) | `letters = {"a", "b"}; letters.discard("z"); sorted(letters)` | `['a', 'b']` |
| `.remove()` | Remove an element, or raise `KeyError` | `letters = {"a", "b"}; letters.remove("a"); letters` | `{'b'}` |
| `.pop()` | Remove and return an arbitrary element | `letters = {7}; letters.pop()` | `7` |
| `.clear()` | Remove all elements | `letters = {1, 2}; letters.clear(); letters` | `set()` |

### Combine (new set)

| Method | What it does | Example | Result |
|---|---|---|---|
| `.union()` | All elements from both sets | `{1, 2}.union({2, 3})` | `{1, 2, 3}` |
| `.intersection()` | Elements found in both sets | `{1, 2, 3}.intersection({2, 3, 4})` | `{2, 3}` |
| `.difference()` | Elements only in the first set | `{1, 2, 3}.difference({3})` | `{1, 2}` |
| `.symmetric_difference()` | Elements in exactly one of the sets | `{1, 2}.symmetric_difference({2, 3})` | `{1, 3}` |

### Combine (change the set in place)

| Method | What it does | Example | Result |
|---|---|---|---|
| `.update()` | Add all elements of another iterable | `numbers = {1}; numbers.update({2, 3}); numbers` | `{1, 2, 3}` |
| `.intersection_update()` | Keep only elements found in the other set | `numbers = {1, 2, 3}; numbers.intersection_update({2, 3, 9}); numbers` | `{2, 3}` |
| `.difference_update()` | Remove elements found in the other set | `numbers = {1, 2, 3}; numbers.difference_update({3}); numbers` | `{1, 2}` |
| `.symmetric_difference_update()` | Keep elements that are in exactly one of the sets | `numbers = {1, 2}; numbers.symmetric_difference_update({2, 3}); numbers` | `{1, 3}` |

### Test

| Method | What it does | Example | Result |
|---|---|---|---|
| `.issubset()` | Is every element also in the other set? | `{1, 2}.issubset({1, 2, 3})` | `True` |
| `.issuperset()` | Does it contain every element of the other set? | `{1, 2, 3}.issuperset({1, 2})` | `True` |
| `.isdisjoint()` | Do the two sets share no elements? | `{1, 2}.isdisjoint({3, 4})` | `True` |

### Copy

| Method | What it does | Example | Result |
|---|---|---|---|
| `.copy()` | Shallow copy | `numbers = {1, 2}; numbers.copy()` | `{1, 2}` |

```python
# Every example from the tables, ready to run

# ----------------------------------------
# str
# ----------------------------------------

# str.capitalize()
print("hello WORLD".capitalize())

# str.casefold()
print("Hello".casefold())

# str.lower()
print("Hello".lower())

# str.upper()
print("Hello".upper())

# str.swapcase()
print("Hello".swapcase())

# str.title()
print("hello world".title())

# str.count()
print("banana".count("an"))

# str.find()
print("banana".find("na"))

# str.rfind()
print("banana".rfind("na"))

# str.index()
print("banana".index("na"))

# str.rindex()
print("banana".rindex("na"))

# str.startswith()
print("report.csv".startswith("rep"))

# str.endswith()
print("report.csv".endswith((".csv", ".txt")))

# str.isalnum()
print("abc123".isalnum())

# str.isalpha()
print("abc".isalpha())

# str.isascii()
print("abc".isascii())

# str.isdecimal()
print("123".isdecimal())

# str.isdigit()
print("123".isdigit())

# str.isidentifier()
print("my_var".isidentifier())

# str.islower()
print("abc".islower())

# str.isnumeric()
print("123".isnumeric())

# str.isprintable()
print("abc".isprintable())

# str.isspace()
print("  ".isspace())

# str.istitle()
print("Hello World".istitle())

# str.isupper()
print("ABC".isupper())

# str.strip()
print("  hi  ".strip())

# str.lstrip()
print("xxhixx".lstrip("x"))

# str.rstrip()
print("xxhixx".rstrip("x"))

# str.removeprefix()
print("test_file".removeprefix("test_"))

# str.removesuffix()
print("file.txt".removesuffix(".txt"))

# str.replace()
print("a-b-c".replace("-", "+"))

# str.expandtabs()
print("a\tb".expandtabs(4))

# str.maketrans()
print(str.maketrans("ab", "xy"))

# str.translate()
print("aabbcc".translate(str.maketrans("ab", "xy")))

# str.split()
print("a,b,c".split(","))

# str.rsplit()
print("a,b,c".rsplit(",", 1))

# str.splitlines()
print("one\ntwo".splitlines())

# str.partition()
print("key=value".partition("="))

# str.rpartition()
print("a.b.c".rpartition("."))

# str.join()
print(", ".join(["a", "b", "c"]))

# str.center()
print("hi".center(6, "*"))

# str.ljust()
print("hi".ljust(5, "."))

# str.rjust()
print("hi".rjust(5, "."))

# str.zfill()
print("42".zfill(5))

# str.format()
print("{} is {}".format("Ann", 30))

# str.format_map()
print("{name} is {age}".format_map({"name": "Ann", "age": 30}))

# str.encode()
print("cafe".encode("utf-8"))

# ----------------------------------------
# list
# ----------------------------------------

# list.append()
items = [1, 2]
items.append(3)
print(items)

# list.extend()
items = [1, 2]
items.extend([3, 4])
print(items)

# list.insert()
items = [1, 3]
items.insert(1, 2)
print(items)

# list.pop()
items = [1, 2, 3]
print(items.pop())

# list.remove()
items = [1, 2, 3]
items.remove(2)
print(items)

# list.clear()
items = [1, 2, 3]
items.clear()
print(items)

# list.index()
print([10, 20, 30].index(20))

# list.count()
print([1, 2, 2, 3].count(2))

# list.sort()
items = [3, 1, 2]
items.sort()
print(items)

# list.reverse()
items = [1, 2, 3]
items.reverse()
print(items)

# list.copy()
items = [1, 2, 3]
print(items.copy())

# ----------------------------------------
# tuple
# ----------------------------------------

# tuple.count()
print((1, 2, 2, 3).count(2))

# tuple.index()
print((10, 20, 30).index(20))

# ----------------------------------------
# dict
# ----------------------------------------

# dict.get()
person = {"name": "Ann"}
print(person.get("age", 0))

# dict.keys()
print({"a": 1, "b": 2}.keys())

# dict.values()
print({"a": 1, "b": 2}.values())

# dict.items()
print({"a": 1, "b": 2}.items())

# dict.update()
person = {"a": 1}
person.update({"b": 2})
print(person)

# dict.setdefault()
person = {}
person.setdefault("tags", [])
print(person)

# dict.pop()
person = {"a": 1, "b": 2}
print(person.pop("a"))

# dict.popitem()
person = {"a": 1, "b": 2}
print(person.popitem())

# dict.clear()
person = {"a": 1}
person.clear()
print(person)

# dict.copy()
person = {"a": 1}
print(person.copy())

# dict.fromkeys()
print(dict.fromkeys(["a", "b"], 0))

# ----------------------------------------
# set
# ----------------------------------------

# set.add()
letters = {"a"}
letters.add("b")
print(sorted(letters))

# set.discard()
letters = {"a", "b"}
letters.discard("z")
print(sorted(letters))

# set.remove()
letters = {"a", "b"}
letters.remove("a")
print(letters)

# set.pop()
letters = {7}
print(letters.pop())

# set.clear()
letters = {1, 2}
letters.clear()
print(letters)

# set.union()
print({1, 2}.union({2, 3}))

# set.intersection()
print({1, 2, 3}.intersection({2, 3, 4}))

# set.difference()
print({1, 2, 3}.difference({3}))

# set.symmetric_difference()
print({1, 2}.symmetric_difference({2, 3}))

# set.update()
numbers = {1}
numbers.update({2, 3})
print(numbers)

# set.intersection_update()
numbers = {1, 2, 3}
numbers.intersection_update({2, 3, 9})
print(numbers)

# set.difference_update()
numbers = {1, 2, 3}
numbers.difference_update({3})
print(numbers)

# set.symmetric_difference_update()
numbers = {1, 2}
numbers.symmetric_difference_update({2, 3})
print(numbers)

# set.issubset()
print({1, 2}.issubset({1, 2, 3}))

# set.issuperset()
print({1, 2, 3}.issuperset({1, 2}))

# set.isdisjoint()
print({1, 2}.isdisjoint({3, 4}))

# set.copy()
numbers = {1, 2}
print(numbers.copy())
```

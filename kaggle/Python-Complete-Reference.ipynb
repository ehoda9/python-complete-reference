#### Numeric Types
- **`int`**  Integer / Whole numbers
- **`float`**  Floating-point numbers

#### Boolean Type
- **`bool`**  Boolean values
  - Values: `True`, `False`

#### Text Type
- **`str`**  String / Text

#### Sequence Types
- **`list`**  Ordered, mutable collection
- **`tuple`**  Ordered, immutable collection
- **`range`**  Sequence of numbers

#### Mapping Type
- **`dict`**  Key-value pairs

#### Set Types
- **`set`**  Unordered collection of unique values
- **`frozenset`**  Immutable set

#### Binary Types
- **`bytes`**  Immutable sequence of bytes
- **`bytearray`**  Mutable sequence of bytes
- **`memoryview`**  Memory view of a binary object

#### None Type
- **`NoneType`**  Type of `None`
  - `None` represents the absence of a value.

---

#### `type()`

> `type(value)`  returns the type of an object.

#### Important notes

- The **object** has a type.
> Everything in Python is an **object**.

````python
# To know the type of an object, we use the type() built-in function.

# Numeric Types
print(type(10))          # int
print(type(10.5))        # float
print(type(3 + 2j))      # complex

# Boolean
print(type(True), type(False))   # bool

# Text
print(type("Hello Instant"))  # str

# Sequence Types
print(type([1, 2, 3]))       # list
print(type((1, 2, 3)))       # tuple
print(type(range(5)))         # range

# Mapping
print(type({"One": 1}))      # dict

# Set Types
print(type({1, 2, 3}))       # set
print(type(frozenset({1, 2, 3})))  # frozenset

# Binary Types
print(type(b"Hello"))        # bytes
print(type(bytearray(b"Hello")))  # bytearray
print(type(memoryview(b"Hello"))) # memoryview

# None
print(type(None))  # NoneType
````

#### 1.2.3 Strings, Indexing, Slicing & Methods

##### What is a String?

- A **string** is a sequence of characters.
- Strings are created using:
  - Single quotes: `'Hello'`
  - Double quotes: `"Hello"`
  - Triple quotes: `'''Hello'''` / `"""Hello"""`

##### Important Notes

- Strings are **immutable**.
- String methods do **not** modify the original string.
- Most string methods return a **new string**.
- Strings are **Unicode**.
- Strings are **iterable**.
- Strings are **sequences**.
- Indexing returns a single-character string.
- Slicing returns a new string.

````python
my_str = 'Single Quote';   my_strr = "Double Quotes" # we can use ; to write in same line but not good for readAbility

print(type(my_str)) # class => str
print(len(my_str)) # built-in function that get the length of object

# '' accept "" inside and verse is right
my_single = 'This is Single Quote "Test"'
my_double = "This is Double Quotes 'Test'"
############################################

multi_line_str_single = '''First
Second "double" 'single'
Third'''

multi_line_str_double = """this is
string "double" 'single'
i love it"""
````

##### Indexing

Used to access a **single character**.

- Positive indexing starts from `0`.
- Negative indexing starts from `-1` from the end.
- Use square brackets `[]`.

##### Slicing

Used to access a **sequence of characters**.

Syntax:

`string[start:end:step]`

- `start` => starting index, included.
- `end` => ending index, **not included**.
- `step` => number of positions to move.
- Default `start` => beginning.
- Default `end` => end of string.
- Default `step` => `1`.
- Negative `step` can be used to reverse a string.


````python
my_string = "I Love Python"

# Indexing
print(my_string[0])      # I

# Negative Indexing
print(my_string[-1])     # n

print(40*"*") # just a separetor

# ---------------------------------
# Slicing

print(my_string[3:5])    # ov

print(my_string[:10])    # I Love Pyt
print(my_string[5:])     # e Python
print(my_string[:])      # I Love Python

print(my_string[::1])    # I Love Python
print(my_string[::3])    # Io tn

# Reverse
print(my_string[::-1]) # nohtyP evoL I
````

#### 1.2.4 String Methods

##### Case Conversion

- `.capitalize()` => first character uppercase, remaining characters lowercase
- `.casefold()` => aggressive lowercase conversion for caseless matching
- `.lower()` => convert to lowercase
- `.upper()` => convert to uppercase
- `.swapcase()` => swap uppercase and lowercase
- `.title()` => capitalize the first character of each word

---

##### Searching & Counting

- `.count(sub)` => count occurrences of a substring
- `.find(sub)` => return first index of substring, or `-1`
- `.rfind(sub)` => return last index of substring, or `-1`
- `.index(sub)` => return first index of substring, raises `ValueError` if not found
- `.rindex(sub)` => return last index of substring, raises `ValueError` if not found

`find()`, `rfind()`, `index()`, `rindex()`, and `count()` can use: `start` and `end`

---

##### Prefix & Suffix

- `.startswith(prefix)` => check whether string starts with a value
- `.endswith(suffix)` => check whether string ends with a value
  - can use: `start` and `end`
---

##### String Testing / Validation

- `.isalnum()` => all characters are alphanumeric
- `.isalpha()` => all characters are alphabetic
- `.isascii()` => all characters are ASCII
- `.isdecimal()` => all characters are decimal characters
- `.isdigit()` => all characters are digits
- `.isidentifier()` => valid Python identifier
- `.islower()` => all cased characters are lowercase
- `.isnumeric()` => all characters are numeric
- `.isprintable()` => all characters are printable
- `.isspace()` => all characters are whitespace
- `.istitle()` => string follows title-case rules
- `.isupper()` => all cased characters are uppercase

---

##### Removing Characters

- `.strip()` => remove characters from both ends
- `.lstrip()` => remove characters from the left
- `.rstrip()` => remove characters from the right
- `.removeprefix(prefix)` => remove prefix if present
- `.removesuffix(suffix)` => remove suffix if present

###### Important
`strip()`, `lstrip()`, and `rstrip()` remove **characters**, not an exact substring.

---

##### Replacing & Translating

- `.replace(old, new)` => replace occurrences
- `.translate(table)` => replace characters using a translation table
- `.maketrans()` => create a translation table for `translate()`

---

##### Splitting

- `.split()` => split from the left
- `.rsplit()` => split from the right
- `.splitlines()` => split at line boundaries
- `.partition(sep)` => split into `(before, separator, after)`
- `.rpartition(sep)` => partition from the right

###### Important

`split()` returns a **list**.
`partition()` returns a **tuple** containing exactly three elements.

---

##### Joining

- `.join(iterable)` => join strings using the current string as separator

---

##### Alignment & Padding

- `.center(width)` => center string
- `.ljust(width)` => left-align string
- `.rjust(width)` => right-align string
- `.zfill(width)` => pad numeric string with zeros

---

##### Formatting

- `.format()` => format values into a string
- `.format_map(mapping)` => format using a mapping object

###### Modern Formatting

- **f-strings** => `f"Hello {name}"`

> f-strings are syntax, not a `str` method.

---

##### Encoding

- `.encode()` => encode the string into a `bytes` object

### 1.3 Variables

#### What is a Variable?

A variable is a name that refers to an object/value.

#### Syntax
`[Variable Name] [Assignment Operator] [Value]`



#### Variable Naming Rules
1. Can start with: Letters: `a-z`, `A-Z` or Underscore: `_`
2. Cannot start with a number `1name`.

#### Important Notes
- A variable is a **name/reference**, not the object itself.
- Assignment does not require declaring the data type.
- The type belongs to the object, not the variable.
- A variable can refer to objects of different types during its lifetime.

### 1.4 Type Conversion

Type conversion means changing a value from one data type to another.

Python provides built-in functions for converting between common data types.

#### Common Type Conversion Functions

| Function | Converts To | Example | Result |
|---|---|---|---|
| `int()` | `int` | `int("10")` | `10` |
| `float()` | `float` | `float("10.5")` | `10.5` |
| `str()` | `str` | `str(10)` | `"10"` |
| `bool()` | `bool` | `bool(10)` | `True` |
| `list()` | `list` | `list("ABC")` | `['A', 'B', 'C']` |
| `tuple()` | `tuple` | `tuple("ABC")` | `('A', 'B', 'C')` |
| `set()` | `set` | `set("ABC")` | `{'A', 'B', 'C'}` |
| `dict()` | `dict` | `dict([("a", 1)])` | `{'a': 1}` |

#### Converting to `int`

Used to convert compatible values into integers.

- `int("10")` => `10`
- `int(10.9)` => `10`
- `int("10.5")` => `ValueError`

> Converting a `float` to `int` removes the decimal part. It does not round the number.

#### Converting to `float`

Used to convert compatible values into floating-point numbers.

- `float("10")` => `10.0`
- `float("10.5")` => `10.5`
- `float(10)` => `10.0`

#### Converting to `str`

Almost any Python object can be converted to a string representation.

- `str(100)` => `"100"`
- `str(10.5)` => `"10.5"`
- `str(True)` => `"True"`

#### Converting to `bool`

`bool()` follows Python's **truthy / falsy** rules.

- `bool(1)` => `True`
- `bool(0)` => `False`
- `bool("Hello")` => `True`
- `bool("")` => `False`
- `bool([1, 2])` => `True`
- `bool([])` => `False`

#### Converting Between Collections

Some iterable objects can be converted into other collection types.

- `list()` => creates a list
- `tuple()` => creates a tuple
- `set()` => creates a set

### 1.6 Operators

#### Arithmetic Operators

Arithmetic operators are used to perform mathematical operations.

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Addition | `10 + 5` | `15` |
| `-` | Subtraction | `10 - 5` | `5` |
| `*` | Multiplication | `10 * 5` | `50` |
| `/` | Division | `10 / 5` | `2.0` |
| `%` | Modulus | `10 % 3` | `1` |
| `**` | Exponentiation | `2 ** 3` | `8` |
| `//` | Floor Division | `10 // 3` | `3` |

---

The `+` operator can also be used with other compatible types, such as strings and lists.

---

##### Operator Precedence

Multiplication is performed before addition.

`5 + 10 * 100` => `1005`

Use parentheses to change the order:

`(5 + 10) * 100` => `1500`

`*` can also repeat sequences:

`"Hi" * 3` => `"HiHiHi"`

---

##### Important notes

The `/` operator always returns a `float`.

`100 / 20` => `5.0`

Even when the result is mathematically an integer:

`10 / 2` => `5.0`

Division by zero raises `ZeroDivisionError`.

##### Important Difference

`/` returns the normal division result.

`//` returns the floor-divided result.

Floor division rounds toward negative infinity, not simply toward zero.

---
##### Arithmetic with Different Numeric Types

Python can perform arithmetic between compatible numeric types.

Examples:

- `int + int` => `int`
- `int + float` => `float`
- `float + float` => `float`
- `int + complex` => `complex`

##### Division by Zero

`10 / 0` => `ZeroDivisionError`

`10 // 0` => `ZeroDivisionError`

`10 % 0` => `ZeroDivisionError`

````python
# ==========================================
# 1. Arithmetic Operators
# ==========================================


# ------------------------------------------
# Addition +
# ------------------------------------------

print(10 + 30)          # + Addition
print(-30 - 20)         # Subtraction -
print(10 * 3)           # Multiplication *
print(100 / 20)         # Division / always returns float

# Multiplication has higher precedence than addition
# Parentheses change the order
print(5 + 10 * 2)        # 25
print((5 + 10) * 2)      # 30

print(9 % 2)            # Modulus %

# Check even / odd
print(10 % 2 == 0)      # True
print(7 % 2 == 0)       # False


# ------------------------------------------
# Exponentiation **
print(2 ** 5)            # 32
print(2 * 2 * 2 * 2 * 2) # 32

# ------------------------------------------
print(100 // 20)         # Floor Division //
# Difference between / and //
print(7 / 2)             # 3.5
print(7 // 2)            # 3

# Floor means toward negative infinity
print(-7 // 2)           # -4


# ------------------------------------------
# Arithmetic with different numeric types
# ------------------------------------------

print(10 + 2.5)          # 12.5
print(type(10 + 2.5))    # float

print(10 + 2j)           # (10+2j)
print(type(10 + 2j))     # complex


# ------------------------------------------
# Operators with sequences
# ------------------------------------------

print("Hi" * 3)          # HiHiHi
print([1, 2] * 3)        # [1, 2, 1, 2, 1, 2]
````

#### 1.6.1 Assignment Operators

Assignment operators are used to assign a value to a variable or update its current value.

| Operator | Name | Example | Equivalent To |
|---|---|---|---|
| `=` | Assignment | `x = 10` | Assign `10` to `x` |
| `+=` | Add and Assign | `x += 5` | `x = x + 5` |
| `-=` | Subtract and Assign | `x -= 5` | `x = x - 5` |
| `*=` | Multiply and Assign | `x *= 5` | `x = x * 5` |
| `/=` | Divide and Assign | `x /= 5` | `x = x / 5` |
| `**=` | Exponentiate and Assign | `x **= 2` | `x = x ** 2` |
| `%=` | Modulus and Assign | `x %= 3` | `x = x % 3` |
| `//=` | Floor Divide and Assign | `x //= 3` | `x = x // 3` |

##### Important Notes

- `=` assigns a value to a variable.
- The other operators perform an operation and then store the result back in the variable.
- `/=` usually changes the result to `float` because `/` always performs true division.
- `//=` uses floor division.
- Assignment operators can be used with compatible data types.

##### Example

If:

`x = 10`

`x += 5`

means:

`x = x + 5`

````python
# --------------------------
# -- Assignment Operators --
# --------------------------

x = 10

x += 5      # x = x + 5
x -= 2      # x = x - 2
x *= 3      # x = x * 3
x /= 2      # x = x / 2
x **= 2     # x = x ** 2
x %= 3      # x = x % 3
x //= 2     # x = x // 2
````

#### 1.6.2 Comparison Operators

Comparison operators are used to compare two values.

The result is always a Boolean value: `True` or `False`.

| Operator | Name | Example | Result |
|---|---|---|---|
| `==` | Equal | `10 == 10` | `True` |
| `!=` | Not Equal | `10 != 5` | `True` |
| `>` | Greater Than | `10 > 5` | `True` |
| `<` | Less Than | `10 < 5` | `False` |
| `>=` | Greater Than or Equal | `10 >= 10` | `True` |
| `<=` | Less Than or Equal | `10 <= 5` | `False` |

##### Important Notes

- `==` checks if two values are equal.
- `=` is **assignment**, not comparison.
- `!=` checks if two values are different.
- `>=` means greater than **or equal to**.
- `<=` means less than **or equal to**.
- Comparison operators return `True` or `False`.
- They are commonly used with `if`, `while`, and logical operators.

#### 1.6.3 Membership Operators

Membership operators are used to check whether a value exists inside a sequence or collection.

They return a Boolean value: `True` or `False`.

| Operator | Name           | Example              | Result |
| -------- | -------------- | -------------------- | ------ |
| `in`     | Membership     | `2 in [1, 2, 3]`     | `True` |
| `not in` | Not Membership | `5 not in [1, 2, 3]` | `True` |

##### `in`

Checks whether a value exists inside a collection.

Example:

`"Python" in ["Python", "Java", "C++"]`

=> `True`

##### `not in`

Checks whether a value does **not** exist inside a collection.

Example:

`"Java" not in ["Python", "C++"]`

=> `True`

##### With Different Data Types

###### Lists

`"Python" in ["Python", "Java"]`

###### Tuples

`10 in (5, 10, 15)`

###### Sets

`10 in {5, 10, 15}`

###### Strings

`"Py" in "Python"`

###### Dictionaries

Membership checks **keys**, not values.

`"name" in {"name": "Mahmoud", "age": 20}`

=> `True`

`"Mahmoud" in {"name": "Mahmoud", "age": 20}`

=> `False`

To check dictionary values:

`"Mahmoud" in {"name": "Mahmoud", "age": 20}.values()`

=> `True`

##### Important Notes

* `in` and `not in` return `True` or `False`.
* They are commonly used with `if` and `while`.
* With strings, they check whether a substring exists.
* With dictionaries, they check **keys** by default.
* `not in` is the opposite of `in`.


````python
# --------------------------
# -- Membership Operators --
# --------------------------

# Lists
"Python" in ["Python", "Java", "C++"]
"PHP" not in ["Python", "Java", "C++"]

# Tuples
10 in (5, 10, 15)
20 not in (5, 10, 15)

# Sets
10 in {5, 10, 15}
20 not in {5, 10, 15}

# Strings
"Py" in "Python"
"Java" not in "Python"

# Dictionaries
user = {
    "name": "Mahmoud",
    "age": 20
}

"name" in user
"email" not in user

# Check dictionary values
"Mahmoud" in user.values()

# Using membership with if
name = "Mahmoud"

if name in ["Ahmed", "Mahmoud", "Mona"]:
    print("Name found")
````

### 2.1 Lists

#### What is a List?

A list is an **ordered, mutable collection** used to store multiple values.

##### List Characteristics

1. **Ordered**
   - Items have a defined order.
   - Each item has an index.

2. **Mutable**
   - Items can be added, removed, or changed after creation.

3. **Allows Duplicates**
   - The same value can appear multiple times.

4. **Can Contain Different Data Types**
   - A single list can contain different types of objects.

5. **Can Be Nested**
   - A list can contain other lists.

---

#### List Operators

| Operator | Name | Example | Result |
|---|---|---|---|
| `+` | Concatenation | `[1, 2] + [3, 4]` | `[1, 2, 3, 4]` |
| `*` | Repetition | `[1, 2] * 2` | `[1, 2, 1, 2]` |
| `in` | Membership | `2 in [1, 2, 3]` | `True` |
| `not in` | Negative Membership | `5 not in [1, 2, 3]` | `True` |

---

#### Indexing

Indexing is used to access a single item.

Syntax:

`list[index]`

##### Positive Indexing

Starts from `0`.

| Index | `0` | `1` | `2` | `3` |
|---|---|---|---|---|
| Item | `"One"` | `"Two"` | `"Three"` | `"Four"` |

##### Negative Indexing

Starts from `-1` from the end.

| Index | `-4` | `-3` | `-2` | `-1` |
|---|---|---|---|---|
| Item | `"One"` | `"Two"` | `"Three"` | `"Four"` |

---

#### Slicing

Used to access multiple items.

Syntax:

`list[start:end:step]`

- `start` => included
- `end` => excluded
- `step` => movement between items
- Missing `start` => beginning
- Missing `end` => end
- Missing `step` => `1`
- Negative `step` => can move backwards

---

#### Modifying Lists

Lists are mutable, so their contents can be changed.

##### Change One Item

`list[index] = value`

##### Change Multiple Items

`list[start:end] = iterable`

The replacement does not have to contain the same number of items.

Example:

`my_list[0:3] = ["A"]`

can replace three items with one item.

---

#### List Methods

##### 1. Adding Items

| Method | Purpose | Example |
|---|---|---|
| `.append(x)` | Add one item to the end | `items.append(10)` |
| `.extend(iterable)` | Add multiple items | `items.extend([10, 20])` |
| `.insert(i, x)` | Insert item at a specific index | `items.insert(1, "A")` |

###### Difference

`append()` adds the entire object as **one item**.

`extend()` adds the elements of an iterable individually.

---

##### 2. Removing Items

| Method | Purpose | Example |
|---|---|---|
| `.remove(x)` | Remove first matching value | `items.remove(10)` |
| `.pop([i])` | Remove and return item | `items.pop()` |
| `.clear()` | Remove all items | `items.clear()` |

###### Important

- `remove()` uses a **value**.
- `pop()` uses an **index**.
- `pop()` returns the removed item.
- `pop()` without an index removes the last item.
- `remove()` raises `ValueError` if the value does not exist.
- `pop()` raises `IndexError` if the index does not exist.

---

##### 3. Searching & Counting

| Method | Purpose | Example |
|---|---|---|
| `.index(x)` | Return index of first matching value | `items.index(10)` |
| `.count(x)` | Count occurrences | `items.count(10)` |

###### `index()`

Returns the index of the first matching item.

Raises `ValueError` if the item is not found.

---

##### 4. Ordering

| Method | Purpose | Example |
|---|---|---|
| `.sort()` | Sort the list in place | `items.sort()` |
| `.reverse()` | Reverse the list in place | `items.reverse()` |

###### `sort()`

Can use:

- `reverse=True`
- `key=...`

Example:

`items.sort(reverse=True)`

###### Important

`.sort()` changes the original list and returns `None`.

`.reverse()` changes the original list and returns `None`.

---

##### 5. Copying

###### `.copy()`

Creates a **shallow copy** of the list.

Example:

`new_list = old_list.copy()`

This creates a different list object.

---

#### Important List Concepts

##### `len()`

Returns the number of items.

`len(my_list)`

##### `in`

Checks whether an item exists.

`"Python" in my_list`

##### `not in`

Checks whether an item does not exist.

`"Java" not in my_list`

##### Iteration

Lists are iterable and can be used with loops.

`for item in my_list:`

---

#### Nested Lists

A list can contain other lists.

Example:

`matrix = [[1, 2], [3, 4], [5, 6]]`

Accessing nested items:

`matrix[0][1]`

Result:

`2`

---

#### Important Notes

- Lists preserve insertion order.
- Lists are mutable.
- Lists allow duplicate values.
- Lists can contain different data types.
- Lists can contain other lists.
- List indexing starts from `0`.
- Negative indexing starts from `-1`.
- Slicing creates a new list.
- Most list methods modify the original list.
- Methods such as `append()`, `extend()`, `insert()`, `remove()`, `sort()`, and `reverse()` modify the list in place.
- `pop()` both modifies the list and returns the removed item.
- `copy()` creates a shallow copy.

### 2.2 Tuples

A `tuple` is an **ordered, immutable collection**.

#### Tuple Characteristics

| Feature | Tuple |
|---|---|
| Syntax | `(1, 2, 3)` |
| Ordered | ✅ |
| Mutable | ❌ |
| Allows Duplicates | ✅ |
| Indexed | ✅ |
| Different Data Types | ✅ |
| Hashable | ✅* |

\* A tuple is hashable only if all of its elements are hashable.

##### Same as Lists

Tuples share these characteristics with lists:

- Ordered
- Indexed
- Allow duplicate values
- Can contain different data types
- Support indexing
- Support slicing
- Support iteration
- Support `len()`
- Support membership operators: `in`, `not in`
- Support concatenation `+`
- Support repetition `*`

##### Main Difference from Lists

- `list` => **mutable**
- `tuple` => **immutable**

---

#### Tuple Syntax

##### Parentheses

`my_tuple = (1, 2, 3)`

##### Parentheses Are Optional

Python allows tuple packing without parentheses:

`my_tuple = 1, 2, 3`

Both create a tuple.

---

#### Single-Item Tuple

A single-item tuple requires a trailing comma.

`(10,)` => tuple

`(10)` => integer

The comma creates the tuple, not the parentheses.

---

#### Tuple Packing & Unpacking

##### Packing

Multiple values can be packed into a tuple:

`my_tuple = 10, 20, 30`

##### Unpacking

Tuple values can be assigned to multiple variables:

`a, b, c = (10, 20, 30)`

Result:

- `a = 10`
- `b = 20`
- `c = 30`

The number of variables must normally match the number of values.

---

#### Tuple Methods

Because tuples are immutable, they have fewer methods than lists.

| Method | Purpose | Example |
|---|---|---|
| `.count(x)` | Count occurrences of a value | `items.count(10)` |
| `.index(x)` | Find the first index of a value | `items.index(10)` |

These methods are also available in lists.

---

#### Immutability

Tuple items cannot be directly changed, added, or deleted.

Example:

`my_tuple[0] = "A"`

raises:

`TypeError`

##### Important

Immutable means the tuple's structure cannot be changed.

However, a tuple can contain a mutable object.

Example:

`my_tuple = ([1, 2], 3)`

The tuple cannot replace its first item:

`my_tuple[0] = [5, 6]` => ❌

But the list inside the tuple can be modified:

`my_tuple[0].append(3)` => ✅

---

#### Tuple as a Dictionary Key

A tuple can be used as a dictionary key **if all of its elements are hashable**.

Example:

`locations = {(10, 20): "Point A"}`

But:

`locations = {([10, 20], 30): "Point A"}`

is invalid because the tuple contains a `list`.

---

#### When to Use a Tuple?

Use a tuple when:

- The data should not be changed.
- You want to represent a fixed collection of values.
- You need a hashable collection of values.
- You want to use the collection as a dictionary key or set element, when its contents are hashable.

#### Set

A `set` is an **unordered, mutable collection of unique elements**.

##### Set Characteristics

| Feature | Set |
|---|---|
| Syntax | `{1, 2, 3}` |
| Ordered | ❌ |
| Indexed | ❌ |
| Slicing | ❌ |
| Mutable | ✅ |
| Allows Duplicates | ❌ |
| Elements | Hashable only |
| Hashable | ❌ |

##### Same as Lists / Tuples

Sets can:

- Contain different data types
- Be nested inside containers that accept them
- Be iterated over
- Use `len()`
- Use membership operators: `in`, `not in`

##### Main Differences

- Sets do not support indexing.
- Sets do not support slicing.
- Sets contain unique elements.
- Set elements must be hashable.

---

#### Creating a Set

Use curly braces:

`my_set = {1, 2, 3}`

##### Empty Set (Important)

`set()` => empty set

`{}` => empty dictionary, **not** an empty set.

---

#### Set Elements & Hashability

Set elements must be **hashable**.

##### Can Be Set Elements

- `int`
- `float`
- `bool`
- `str`
- `tuple` containing only hashable elements
- `frozenset`

##### Cannot Be Set Elements

- `list`
- `dict`
- `set`

##### Important

A set can contain a tuple:

`{(1, 2), (3, 4)}`

But a tuple containing an unhashable object cannot be a set element:

`{([1, 2], 3)}` => invalid

---

#### Set Operations

Sets are especially useful for mathematical set operations.

| Operator | Operation | Example |
|---|---|---|
| `\|` | Union | `A \| B` |
| `&` | Intersection | `A & B` |
| `-` | Difference | `A - B` |
| `^` | Symmetric Difference | `A ^ B` |

##### Meaning

- **Union** => all elements from both sets
- **Intersection** => elements common to both sets
- **Difference** => elements in the first set but not the second
- **Symmetric Difference** => elements that exist in either set, but not both

---

#### Set Methods

##### Adding

|---|---|
| `.add(x)` | Add one element |
| `.update(iterable)` | Add multiple elements |

##### Removing

| Method | Purpose |
|---|---|
| `.remove(x)` | Remove element; raises `KeyError` if not found |
| `.discard(x)` | Remove element; does nothing if not found |
| `.pop()` | Remove and return an arbitrary element |
| `.clear()` | Remove all elements |

##### Set Operations

| Method | Purpose |
|---|---|
| `.union()` | Union |
| `.intersection()` | Intersection |
| `.difference()` | Difference |
| `.symmetric_difference()` | Symmetric difference |

##### Relationship Testing

| Method | Purpose |
|---|---|
| `.issubset()` | Check whether another set contains all elements |
| `.issuperset()` | Check whether the set contains all elements of another set |
| `.isdisjoint()` | Check whether two sets have no common elements |

---

#### Frozen Set

A `frozenset` is an **immutable version of a set**.

##### Characteristics

| Feature | `set` | `frozenset` |
|---|---|---|
| Ordered | ❌ | ❌ |
| Indexed | ❌ | ❌ |
| Mutable | ✅ | ❌ |

| Hashable | ❌ | ✅ |

- `frozenset.add()` => not available

Use `frozenset` when you need:

- A set that cannot be modified.
- A set as an element inside another set.
- A set as a dictionary key.

Example:

`my_set = {frozenset({1, 2}), frozenset({3, 4})}`

Example dictionary key:

`data = {frozenset({1, 2}): "value"}`

---

#### Important Rules

##### Set

`set` => mutable + unique + unhashable

##### Frozen Set

`frozenset` => immutable + unique + hashable

##### Remember

> **Set = mutable unique collection**

> **Frozenset = immutable unique collection**

### 2.5 Boolean

#### What is Boolean?

`bool` is Python's built-in type for representing one of two truth values:

- `True`
- `False`

Boolean values are used to represent whether a condition or expression is true or false.

---

#### Boolean Values

| Value | Meaning |
|---|---|
| `True` | The condition is true |
| `False` | The condition is false |

`True` and `False` are Boolean constants.

---

#### Boolean Expressions

Comparisons produce Boolean values.

| Operator | Meaning | Example | Result |
|---|---|---|---|
| `>` | Greater than | `10 > 5` | `True` |
| `<` | Less than | `10 < 5` | `False` |
| `>=` | Greater than or equal | `10 >= 10` | `True` |
| `<=` | Less than or equal | `10 <= 5` | `False` |
| `==` | Equal to | `10 == 10` | `True` |
| `!=` | Not equal to | `10 != 5` | `True` |

---

#### `bool()`

`bool()` converts an object to a Boolean value.

Syntax:

`bool(value)`

---

#### Truthy Values

An object is **truthy** when `bool(object)` returns `True`.

Examples include:

- Non-zero numbers
- Non-empty strings
- Non-empty lists
- Non-empty tuples
- Non-empty sets
- Non-empty dictionaries
- `True`

Examples:

`bool(10)` => `True`

`bool("Hello")` => `True`

`bool([1, 2])` => `True`

---

#### Falsy Values

An object is **falsy** when `bool(object)` returns `False`.

Important built-in falsy values:

| Value | Type |
|---|---|
| `False` | `bool` |
| `None` | `NoneType` |
| `0` | `int` |
| `0.0` | `float` |
| `0j` | `complex` |
| `""` | `str` |
| `[]` | `list` |
| `()` | `tuple` |
| `{}` | `dict` |
| `set()` | `set` |
| `range(0)` | `range` |

##### Important

Empty containers are falsy.

Non-empty containers are generally truthy.

---

#### Boolean in Conditions

Python automatically evaluates an object's truth value when it is used in a condition.

Example:

`if "Hello":`

The string is non-empty, so it is treated as `True`.

Example:

`if []:`

The list is empty, so it is treated as `False`.

This is called **truth-value testing**.

---

#### Important Notes

- `True` and `False` are Boolean values.
- `bool` is a built-in Python type.
- `True` and `False` are case-sensitive.
- `true` and `false` are not valid Python Boolean constants.
- `bool()` returns either `True` or `False`.
- Boolean values are commonly produced by comparison and logical expressions.
- `1` is truthy and `0` is falsy, but they are not the same type as `True` and `False`.

````python
# ==========================================
# 1. Boolean Values
# ==========================================

is_logged_in = True
is_admin = False


# ==========================================
# 2. Boolean Expressions
# ==========================================

10 > 5
10 < 5

10 >= 10
10 <= 5

10 == 10
10 != 5


# ==========================================
# 3. bool()
# ==========================================

bool(True)
bool(False)

bool(100)
bool(-10)
bool(0)

bool("Hello")
bool("")

bool([1, 2, 3])
bool([])

bool((1, 2, 3))
bool(())

bool({1, 2, 3})
bool(set())

bool({"name": "Mahmoud"})
bool({})

bool(None)


# ==========================================
# 4. Truthy Values
# ==========================================

bool(100)
bool(-100)
bool(0.5)

bool("Python")
bool([1, 2, 3])
bool((1, 2))
bool({1, 2})
bool({"name": "Mahmoud"})


# ==========================================
# 5. Falsy Values
# ==========================================

bool(False)
bool(None)

bool(0)
bool(0.0)
bool(0j)

bool("")

bool([])
bool(())
bool({})
bool(set())

bool(range(0))


# ==========================================
# 6. Truth-Value Testing
# ==========================================

name = "Mahmoud"

if name:
    pass

items = []

if items:
    pass


# ==========================================
# 7. bool vs int
# ==========================================

print(type(True))
print(type(1))

print(True == 1)
print(False == 0)
````

| `frozenset` | ❌ | ❌ | ❌ | ❌ | ✅ | `frozenset({1, 2, 3})` |
| `dict` | ✅** | ✅ | Keys: ❌ | Key-based | ❌ | `{"a": 1}` |
| `range` | ✅ | ❌ | N/A | ✅ | ✅ | `range(5)` |

###### Notes

- **Ordered** => preserves element order.
- **Mutable** => can be changed after creation.
- **Duplicates** => whether the container can contain duplicate elements.
- **Indexed** => supports access using numeric indexes such as `items[0]`.
- **Hashable** => can be used as a dictionary key or as an element of a set.

\* A tuple is hashable only if all of its elements are hashable.

\** Dictionaries preserve insertion order in modern Python.

---

##### Quick Selection Guide

| If you need... | Use |
|---|---|
| Ordered, changeable collection | `list` |
| Ordered, immutable collection | `tuple` |
| Unique, changeable values | `set` |
| Unique, immutable values | `frozenset` |
| Key => Value relationships | `dict` |
| Immutable sequence of numbers | `range` |

###### General Rule

A container can contain another object if the inner object is allowed as an element/value/key of the outer container.

###### Examples

| `tuple` | Almost any Python object |

| `frozenset` | Hashable objects only |
| `dict` keys | Hashable objects only |
| `dict` values | Almost any Python object |

##### Common Mistakes

* `len(100)` => `TypeError`
* `input()` returns a string even when the user enters a number.
* `all()` checks Truthy/Falsy values; it does not require literal `True`.
* `any()` returns a Boolean, not the matching element.
* `id()` is not a value comparison tool.
* `dir()` shows available names; it does not execute them.

#### 3.1.3 Numbers & Mathematics

Python provides several built-ins for common numeric operations.

| Function   | Purpose              | Example          | Result   |
| ---------- | -------------------- | ---------------- | -------- |
| `abs()`    | Absolute value       | `abs(-10)`       | `10`     |
| `round()`  | Round a number       | `round(10.6)`    | `11`     |
| `pow()`    | Power calculation    | `pow(2, 3)`      | `8`      |
| `divmod()` | Quotient + remainder | `divmod(10, 3)`  | `(3, 1)` |
| `sum()`    | Add iterable values  | `sum([1, 2, 3])` | `6`      |
| `min()`    | Smallest value       | `min(3, 1, 5)`   | `1`      |
| `max()`    | Largest value        | `max(3, 1, 5)`   | `5`      |

##### `abs()`

Returns the absolute value.

`abs(-10)` => `10`

For complex numbers, it returns the magnitude.

##### `round()`

Rounds a number.

`round(10.567, 2)` => `10.57`

Be careful with `.5`: Python uses **round half to even** in applicable cases.

##### `pow()`

`pow(base, exponent)`

is similar to:

`base ** exponent`

It also supports:

`pow(base, exponent, modulus)`

which calculates the modular power efficiently.

##### `divmod()`

Returns a tuple containing:

`(quotient, remainder)`

Example:

`divmod(10, 3)` => `(3, 1)`

##### `sum()`

Adds values from an iterable.

`sum([1, 2, 3])` => `6`

The optional `start` value is added to the result.

##### `min()` and `max()`

Find the smallest/largest item.

They support `key=` for custom comparison and `default=` when the iterable may be empty.

###### Common Mistakes

* `min([])` / `max([])` => `ValueError`
* `sum(["a", "b"])` => `TypeError`
* `round()` does not always behave like manual `.5 => up` rounding.
* `abs()` does not modify the original value.


````python
# -----------------------------
# -- Numbers & Math Functions --
# -----------------------------

# abs(number) => absolute value
abs(-10)
abs(-3.5)

# round(number, ndigits=None) => round number
round(10.6)
round(10.5678, 2)

# pow(base, exp, mod=None) => calculate power
pow(2, 3)
pow(2, 3, 5)

# divmod(a, b) => return (quotient, remainder)
divmod(10, 3)

# sum(iterable, start=0) => sum values
sum([1, 2, 3])
sum([1, 2, 3], 10)

# min(iterable, *, key=None, default=...) => smallest item
min([10, 5, 20])

# max(iterable, *, key=None, default=...) => largest item
max([10, 5, 20])

# min/max with key => compare using a custom rule
names = ["Ali", "Mahmoud", "Mona"]
min(names, key=len)
max(names, key=len)

# max with default => avoid ValueError on empty iterable
max([], default=0)
````

#### 3.1.4 Iteration & Functional

These functions are especially useful when working with loops, iterables, and functions.

| Function      | Purpose                                   |
| ------------- | ----------------------------------------- |
| `enumerate()` | Add an index while iterating              |
| `zip()`       | Combine multiple iterables                |
| `map()`       | Apply a function to items                 |
| `filter()`    | Keep items that satisfy a condition       |
| `iter()`      | Create an iterator                        |
| `next()`      | Get the next item from an iterator        |
| `reversed()`  | Iterate in reverse                        |
| `sorted()`    | Return a sorted list                      |
| `range()`     | Generate an immutable sequence of numbers |

##### `enumerate()`

Useful when you need both the index and the value.

`enumerate(iterable, start=0)`

##### `zip()`

Combines items from multiple iterables.

`zip(iterable1, iterable2, ...)`

By default, it stops when the shortest iterable ends.

##### `map()`

Applies a function to every item.

`map(function, iterable, ...)`

It returns an iterator, so `list()` is commonly used to see all results.

##### `filter()`

Keeps elements for which the function returns Truthy.

`filter(function, iterable)`

##### `iter()` and `next()`

`iter()` creates an iterator.

`next()` gets the next item.

When an iterator is exhausted:

`next(iterator)` => `StopIteration`

You can provide a default:

`next(iterator, default)`

##### `sorted()`

Returns a **new sorted list**.

It does not modify the original iterable.

Supports:

* `key=`
* `reverse=`

##### `reversed()`

Returns a reverse iterator.

It does not modify the original object.

##### `range()`

Creates an immutable sequence of numbers.

`range(stop)`

`range(start, stop, step)`

The `stop` value is excluded.

###### Important

`range()` is a built-in sequence type, not a normal function.

###### Common Mistakes

* `map()` and `filter()` return iterators.
* `zip()` also returns an iterator.
* `sorted()` creates a new list.
* `list.sort()` modifies the list in place.
* `range(5)` produces `0` through `4`, not `5`.
* `next()` can raise `StopIteration`.


````python
# ----------------------------------
# -- Iteration & Functional Tools --
# ----------------------------------

# enumerate(iterable, start=0) => index + value
names = ["Ahmed", "Mona", "Ali"]

for index, name in enumerate(names, start=1):
    pass

# zip(*iterables, strict=False) => combine iterables
names = ["Ahmed", "Mona", "Ali"]
ages = [20, 21, 19]

for name, age in zip(names, ages):
    pass

# map(function, iterable, *iterables) => transform items
numbers = [1, 2, 3, 4]

doubled = map(lambda x: x * 2, numbers)
list(doubled)

# filter(function, iterable) => keep items where function is Truthy
numbers = [1, 2, 3, 4, 5]

even_numbers = filter(lambda x: x % 2 == 0, numbers)
list(even_numbers)

# iter(iterable) => create iterator
numbers = [10, 20, 30]
iterator = iter(numbers)

# next(iterator, default=None) => get next item
next(iterator)
next(iterator)
next(iterator)
next(iterator, "No more items")

# reversed(sequence) => reverse iterator
numbers = [1, 2, 3, 4]
list(reversed(numbers))

# sorted(iterable, *, key=None, reverse=False) => new sorted list
numbers = [5, 2, 8, 1]
sorted(numbers)
sorted(numbers, reverse=True)

# sorted() with key
names = ["Ali", "Mahmoud", "Mona"]
sorted(names, key=len)

# range(stop) => numbers from 0 to stop - 1
range(5)

# range(start, stop, step=1) => numeric sequence
list(range(2, 10, 2))
list(range(5, 0, -1))
````

#### 3.1.8 Classes & OOP

Some built-ins are especially important when working with classes and objects.

| Built-in         | Purpose                           |
| ---------------- | --------------------------------- |
| `object()`       | Create a base object              |
| `type()`         | Inspect/create a type             |
| `isinstance()`   | Check object type                 |
| `issubclass()`   | Check class inheritance           |
| `property()`     | Create managed attributes         |
| `classmethod()`  | Create class methods              |
| `staticmethod()` | Create static methods             |
| `super()`        | Access parent/superclass behavior |

##### `object()`

Creates a basic object.

`object()`

It is also the base class of Python's class hierarchy.

##### `isinstance()`

Checks whether an object is an instance of a class.

`isinstance(object, classinfo)`

It also considers inheritance.

##### `issubclass()`

Checks whether one class inherits from another.

`issubclass(class, classinfo)`

##### `type()`

One argument:

`type(object)` => inspect type.

Three arguments:

`type(name, bases, namespace)` => dynamically create a class.

##### `property()`

Creates a property that behaves like an attribute but can execute getter/setter logic.

`property(fget=None, fset=None, fdel=None, doc=None)`

Usually used through the `@property` decorator.

##### `classmethod()`

Makes a method receive the class (`cls`) instead of the instance.

##### `staticmethod()`

Creates a method that does not receive an automatic `self` or `cls`.

##### `super()`

Returns a proxy used to access methods from a parent/superclass.

###### Important

`classmethod()` and `staticmethod()` are usually used as decorators:

`@classmethod`

`@staticmethod`


````python
# -----------------------------
# -- Classes & OOP Functions --
# -----------------------------

# object() => create basic object
obj = object()

# isinstance(object, classinfo) => check instance type
isinstance(10, int)
isinstance("Python", str)

class Animal:
    pass

class Dog(Animal):
    pass

dog = Dog()

isinstance(dog, Dog)
isinstance(dog, Animal)

# issubclass(class, classinfo) => check inheritance
issubclass(Dog, Animal)
issubclass(Dog, object)

# type(object) => get object's type
type(dog)

# type(name, bases, namespace) => dynamically create class
User = type(
    "User",
    (),
    {"name": "Mahmoud"}
)

user = User()
user.name


# property(fget=None, fset=None, fdel=None, doc=None)
# => create managed attribute

class Person:
    def __init__(self, name):
        self._name = name

    @property
    def name(self):
        return self._name

person = Person("Mahmoud")
person.name


# classmethod(function) => method receives cls
class User:
    count = 0

    @classmethod
    def show_count(cls):
        return cls.count

User.show_count()


# staticmethod(function) => method without self/cls
class Calculator:

    @staticmethod
    def add(a, b):
        return a + b

Calculator.add(10, 20)


# super(type=None, object_or_type=None) => access parent behavior
class Animal:

    def speak(self):
        return "Animal sound"


class Dog(Animal):

    def speak(self):
        return super().speak()

dog = Dog()
dog.speak()
````

| Category          | Built-ins                                                                                                                 |

````python
# -----------------------
# -- Async Built-ins --
# -----------------------

import asyncio

async def async_gen():
    yield 1
    yield 2
    yield 3

async def demo_async_builtins():
    async_iterable = async_gen()

    # aiter(async_iterable) => create async iterator
    async_iterator = aiter(async_iterable)

    # anext(async_iterator) => get next async item
    value = await anext(async_iterator)
    print(value)

    # anext(async_iterator, default) => default if exhausted
    value = await anext(async_iterator, "Done")
    print(value)

# Jupyter and Kaggle already run an event loop, so use top-level await here.
# In a plain .py script, use: asyncio.run(demo_async_builtins())
await demo_async_builtins()
````

### code

#### `if - elif - else`

Used when there are multiple conditions.

* `if` checks the first condition.

### 4.2 While Loop

A `while` loop is used to repeat a block of code **as long as a condition is `True`**.

#### Syntax

`while condition:`

```
### code
```

The condition is checked **before every iteration**.

* If the condition is `True` => the code runs.
* If the condition is `False` => the loop stops.
* If the condition is initially `False` => the loop does not run at all.

---

#### How a `while` Loop Works

1. Check the condition.
2. If it is `True`, run the loop body.
3. Update the value that affects the condition.
4. Check the condition again.
5. Repeat until the condition becomes `False`.

---

#### Important

The loop should normally contain something that eventually makes the condition `False`.

Otherwise, the loop may become an **infinite loop**.

Example:

`x = 1`

`while x <= 5:`

```
`x += 1`
```

Here, `x` changes on every iteration until `x <= 5` becomes `False`.

---

#### Increment and Decrement

Python does **not** use `++` or `--` operators.

Use:

`x += 1`

instead of:

`x++`

And:

`x -= 1`

instead of:

`x--`

---

#### `while` With `else`

Python allows an `else` block with a `while` loop.

The `else` block runs when the loop finishes normally because its condition becomes `False`.

It does **not** run if the loop is stopped using `break`.

##### Syntax

`while condition:`

```
`# code`
```

`else:`

```
`# code`
```

---

#### `break`

`break` immediately stops the loop.

The loop ends even if its condition is still `True`.

---

#### `continue`

`continue` skips the rest of the current iteration and starts the next iteration.

---

#### Common Uses

`while` loops are useful when:

* The number of iterations is not known beforehand.
* The loop depends on a condition.
* Waiting for valid user input.
* Repeating an action until a specific value is entered.
* Running until a condition changes.

---

#### Important Notes

* `while` depends on a condition.
* The condition is checked before each iteration.
* A `while` loop can run zero or more times.
* Make sure the condition can eventually become `False`.
* `break` stops the loop completely.
* `continue` skips the current iteration.
* Python does not have `++` or `--`.
* `while` can have an `else` block.


````python
# -------------------
# -- While Loop --
# -------------------

# Basic while loop
counter = 1
while counter <= 5:
    counter += 1
print("Final counter:", counter)

# -------------------
# -- Increment --
# -------------------
x = 1
while x <= 5:
    x += 1
print("Final x (increment):", x)

# -------------------
# -- Decrement --
# -------------------
x = 5
while x >= 1:
    x -= 1
print("Final x (decrement):", x)

# -------------------
# -- User Input --
# -------------------
# password = ""
# while password != "1234":
#     password = input("Enter password: ")

# Simulated version so the notebook runs without blocking on stdin:
attempts = ["0000", "1111", "1234"]
password = ""
i = 0
while password != "1234":
    password = attempts[i]
    i += 1
print("Access granted with password:", password)

# -------------------
# -- break --
# -------------------
x = 1
while x <= 10:
    if x == 5:
        break
    x += 1
print("Stopped at:", x)

# -------------------
# -- continue --
# -------------------
x = 0
while x < 5:
    x += 1
    if x == 3:
        continue
    print("x =", x)

# -------------------
# -- while / else --
# -------------------
x = 1
while x <= 5:
    x += 1
else:
    print("Loop finished normally, x =", x)

# -------------------
# -- break + else --
# -------------------
x = 1
while x <= 5:
    if x == 3:
        break
    x += 1
else:
    # This does not run because break stopped the loop
    print("This will not print")
print("Loop ended with break, x =", x)

# -------------------
# -- Practical Example --
# -- Sum Until 0 --
# -------------------
# total = 0
# number = int(input("Enter a number (0 to stop): "))
# while number != 0:
#     total += number
#     number = int(input("Enter a number (0 to stop): "))

# Simulated version so the notebook runs without blocking on stdin:
total = 0
inputs = [5, 10, 15, 0]
i = 0
number = inputs[i]
while number != 0:
    total += number
    i += 1
    number = inputs[i]
print("Total:", total)
````

### 4.4 Iterable vs Iterator

Python objects that can be traversed one item at a time are called **iterables**.

An **iterator** is an object that keeps track of its current position and provides the next item when requested.

---

#### Iterable

An **iterable** is an object that can return its items one by one.

Common examples:

* `list`
* `tuple`
* `set`
* `dict`
* `str`
* `range`

An iterable can be passed to `iter()` to create an iterator.

`iter(iterable)` => returns an iterator

##### Example

`numbers = [10, 20, 30]`

`iterator = iter(numbers)`

The list is the **iterable**.

The object returned by `iter()` is the **iterator**.

---

#### Iterator

An **iterator** is an object that provides items one at a time using `next()`.

`next(iterator)` => returns the next item

Each call moves the iterator forward.

When there are no more items:

`next(iterator)` => `StopIteration`

##### Example

`numbers = [10, 20, 30]`

`iterator = iter(numbers)`

`next(iterator)` => `10`

`next(iterator)` => `20`

`next(iterator)` => `30`

`next(iterator)` => `StopIteration`

---

#### Iterable vs Iterator

| Feature                       | Iterable    | Iterator               |
| ----------------------------- | ----------- | ---------------------- |
| Contains / provides data      | ✅           | ✅                      |
| Can be used with `for`        | ✅           | ✅                      |
| Can call `iter()`             | ✅           | Usually returns itself |
| Can call `next()` directly    | ❌           | ✅                      |
| Keeps current iteration state | ❌           | ✅                      |
| Exhausted after iteration     | Usually ❌   | ✅                      |
| Example                       | `[1, 2, 3]` | `iter([1, 2, 3])`      |

---

#### How `for` Works

When you write:

`for item in iterable:`

Python internally gets an iterator from the iterable and repeatedly calls `next()`.

Conceptually:

`iterator = iter(iterable)`

Then:

`next(iterator)`

`next(iterator)`

`next(iterator)`

until `StopIteration` occurs.

The `for` loop handles `StopIteration` automatically, so you normally do not see the exception.

---

#### Example Flow

For:

`numbers = [10, 20, 30]`

`for number in numbers:`

The process is approximately:

1. `iter(numbers)` => create iterator
2. `next(iterator)` => `10`
3. `next(iterator)` => `20`
4. `next(iterator)` => `30`
5. `next(iterator)` => `StopIteration`
6. Loop stops

---

#### Important Difference

An iterable can usually be iterated over **again** by creating a new iterator.

An iterator remembers its current position.

Example:

`numbers = [10, 20, 30]`

`iterator = iter(numbers)`

After:

`next(iterator)` => `10`

the iterator is now positioned at `20`.

Calling `next(iterator)` again gives `20`.

---

#### Iterator Protocol

An iterator follows two important methods:

* `__iter__()` => returns the iterator itself
* `__next__()` => returns the next item

Conceptually:

`iter(iterator) is iterator` => `True`

When no items remain, `__next__()` raises `StopIteration`.

---

#### Common Mistakes

* `next([1, 2, 3])` => `TypeError` because a list is iterable, but is not itself an iterator.
* Use `next(iter([1, 2, 3]))` instead.
* An iterator can become **exhausted**.
* Calling `iter()` on an iterable creates an iterator; it does not modify the original iterable.
* `for` handles `StopIteration` automatically.
* Do not confuse an iterable with an iterator.

````python
# ---------------------------------
# -- Iterable vs Iterator --
# ---------------------------------

# Iterable
numbers = [10, 20, 30]

# iter(iterable) => create an iterator
iterator = iter(numbers)

# next(iterator) => get the next item
next(iterator)
next(iterator)
next(iterator)

# next(iterator) => StopIteration
# next(iterator)


# ---------------------------------
# -- Iterable Examples --
# ---------------------------------

# list => iterable
iter([1, 2, 3])

# tuple => iterable
iter((1, 2, 3))

# set => iterable
iter({1, 2, 3})

# string => iterable
iter("Python")

# dictionary => iterable over keys
iter({"name": "Mahmoud", "age": 20})

# range => iterable
iter(range(5))


# ---------------------------------
# -- Iterator Check --
# ---------------------------------

numbers = [10, 20, 30]
iterator = iter(numbers)

# list is iterable, not an iterator
# next(numbers)  # TypeError

# iterator supports next()
next(iterator)

# An iterator returns itself from iter()
iter(iterator) is iterator


# ---------------------------------
# -- for Loop --
# ---------------------------------

numbers = [10, 20, 30]

for number in numbers:
    pass

# Conceptually similar to:
iterator = iter(numbers)

while True:
    try:
        number = next(iterator)
    except StopIteration:
        break


# ---------------------------------
# -- Iterator Exhaustion --
# ---------------------------------

numbers = [10, 20]
iterator = iter(numbers)

next(iterator)
next(iterator)

# No items remain
# next(iterator)  # StopIteration


# ---------------------------------
# -- Iterator Protocol --
# ---------------------------------

iterator = iter([10, 20, 30])

# __iter__() => return iterator
iterator.__iter__() is iterator

# __next__() => return next item
iterator.__next__()
iterator.__next__()
````

<a id="sec-4-5"></a>

### 4.5 Generators

A **generator** is a special type of iterator that produces values **one at a time** instead of creating and storing all values at once.

Generators are created using a function containing the `yield` keyword.

---

#### Generator Function

A generator function looks like a normal function, but uses:

`yield`

instead of, or in addition to, `return`.

##### Syntax

`def generator_function():`

```
`yield value`
```

Calling the generator function does **not** immediately execute its body.

Instead, it returns a **generator object**.

---

#### `yield` vs `return`

| `return`                                 | `yield`                                |
| ---------------------------------------- | -------------------------------------- |
| Ends the function                        | Pauses the function                    |
| Returns one final result                 | Produces a value                       |
| Function execution is finished           | Execution can resume later             |
| Calling again starts a new function call | `next()` resumes from where it stopped |

---

#### Using `next()`

`next(generator)`

gets the next value from the generator.

After yielding a value, the generator pauses.

The next `next()` call resumes execution **from the point where the previous `yield` occurred**.

It does not start from the beginning.

---

#### Generator Execution

Consider:

`def numbers():`

```
`yield 1`

`yield 2`

`yield 3`
```

When we call:

`generator = numbers()`

the function body does not start yet.

Then:

`next(generator)` => `1`

The generator pauses after `yield 1`.

`next(generator)` => `2`

It resumes from where it paused.

`next(generator)` => `3`

After the final `yield`, the generator is exhausted.

Another `next()` raises:

`StopIteration`

---

#### Multiple `yield`s

A generator function can contain one or many `yield` statements.

Each `yield` can produce another value.

Example:

`def numbers():`

```
`yield 1`

`yield 2`

`yield 3`
```

Each call to `next()` continues execution until the next `yield`.

---

#### Generators and `for`

Generators are iterators, so they can be used directly with a `for` loop.

`for number in numbers():`

```
`pass`
```

The `for` loop automatically calls `next()` and handles `StopIteration`.

---

#### Lazy Evaluation

Generators use **lazy evaluation**.

They produce values only when they are requested.

This means they can save memory when working with large amounts of data.

##### Example

A list:

`numbers = [x for x in range(1000000)]`

creates and stores all values.

A generator:

`numbers = (x for x in range(1000000))`

produces values when needed.

---

#### Generator Expression

A generator expression looks similar to a list comprehension, but uses parentheses.

List comprehension:

`[x * 2 for x in range(5)]`

Generator expression:

`(x * 2 for x in range(5))`

The list comprehension creates a list immediately.

The generator expression creates values lazily.

---

#### Important Notes

* A generator function contains `yield`.
* Calling a generator function does not execute its body immediately.
* It returns a generator object.
* `next()` resumes execution from the previous `yield`.
* A generator can have multiple `yield` statements.
* Generators are iterators.
* Generators are consumed as they are iterated.
* Once exhausted, they cannot be restarted; create a new generator if needed.
* `for` automatically handles `StopIteration`.
* Generators are useful for large data and memory-efficient processing.
* A generator usually uses less memory than creating the complete collection at once.


````python
# ----------------
# -- Generators --
# ----------------

# ---------------------------------
# -- Basic Generator --
# ---------------------------------

# yield => pause function and produce a value
def numbers():
    yield 1
    yield 2
    yield 3


# Calling the function does NOT execute its body
generator = numbers()

# next(generator) => resume until the next yield
next(generator)
next(generator)
next(generator)

# No more values => StopIteration
# next(generator)


# ---------------------------------
# -- Multiple yield Statements --
# ---------------------------------

def generate_numbers():
    yield 10
    yield 20
    yield 30
    yield 40

generator = generate_numbers()

next(generator)
next(generator)
next(generator)
next(generator)


# ---------------------------------
# -- Generator + for Loop --
# ---------------------------------

def count_to_five():
    yield 1
    yield 2
    yield 3
    yield 4
    yield 5

for number in count_to_five():
    pass


# ---------------------------------
# -- yield Resumes From Its Position --
# ---------------------------------

def demo():
    yield "First"

    # Execution resumes here after the first next()
    yield "Second"

    # Execution resumes here after the second next()
    yield "Third"

generator = demo()

next(generator)  # First
next(generator)  # Second
next(generator)  # Third


# ---------------------------------
# -- Generator With a Loop --
# ---------------------------------

def count_to(number):
    current = 1

    while current <= number:
        yield current
        current += 1

generator = count_to(5)

next(generator)
next(generator)

for number in generator:
    pass


# ---------------------------------
# -- Generator Expression --
# ---------------------------------

# List comprehension => creates the complete list
numbers_list = [x * 2 for x in range(5)]

# Generator expression => produces values lazily
numbers_generator = (x * 2 for x in range(5))

next(numbers_generator)
next(numbers_generator)

# Consume remaining values
for number in numbers_generator:
    pass


# ---------------------------------
# -- Generator Is an Iterator --
# ---------------------------------

generator = count_to(3)

iter(generator) is generator

next(generator)
next(generator)
next(generator)


# ---------------------------------
# -- Generator With return --
# ---------------------------------

def example():
    yield 1
    yield 2
    return "Finished"

generator = example()

next(generator)
next(generator)

# The return value becomes the StopIteration value
# next(generator)  # StopIteration: Finished
````

### 5.2 Parameters & Arguments

Parameters allow a function to receive data.

Arguments are the actual values passed to those parameters when the function is called.

---

#### Parameter vs Argument

`def add(a, b):`

Here:

* `a` and `b` are **parameters**.

`add(10, 5)`

Here:

* `10` and `5` are **arguments**.

---

#### Positional Arguments

Arguments are assigned to parameters based on their position.

`add(10, 5)`

=> `a = 10`

=> `b = 5`

The order matters.

---

#### Keyword Arguments

Arguments can be passed using parameter names.

`add(a=10, b=5)`

The order does not matter when using parameter names.

`add(b=5, a=10)`

---

#### Mixing Positional and Keyword Arguments

Positional arguments must come before keyword arguments.

Valid:

`add(10, b=5)`

Invalid:

`add(a=10, 5)`

---

#### Multiple Parameters

A function can accept any number of regular parameters.

`def calculate(a, b, c):`

```
`return a + b + c`
```

---

#### Important Notes

* Parameters exist in the function definition.
* Arguments exist in the function call.
* Positional arguments depend on order.
* Keyword arguments use parameter names.
* Positional arguments must come before keyword arguments.
* Passing the wrong number of required arguments raises `TypeError`.


````python
# ---------------------------------------
# -- Function Parameters And Arguments --
# ---------------------------------------

def introduce(name, age):
    return name, age


# Positional arguments
introduce("Mahmoud", 20)


# Keyword arguments
introduce(name="Mahmoud", age=20)


# Keyword arguments can change the order
introduce(age=20, name="Mahmoud")


# Positional + keyword
introduce("Mahmoud", age=20)


# Multiple parameters
def calculate(a, b, c):
    return a + b + c


calculate(10, 20, 30)
calculate(a=10, b=20, c=30)
calculate(10, b=20, c=30)
````

### 5.3 Default Parameters

A default parameter has a value that Python uses when the caller does not provide an argument for that parameter.

#### Syntax

`def function_name(parameter=value):`

---

#### Example

`def greet(name="Guest"):`

```
`return name`
```

`greet()`

=> `"Guest"`

`greet("Mahmoud")`

=> `"Mahmoud"`

---

#### Default + Required Parameters

A function can have both required and default parameters.

`def greet(name, message="Hello"):`

```
`return message, name`
```

The required parameter must be provided.

---

#### Important Rule

Parameters without default values must come **before** parameters with default values.

Valid:

`def greet(name, message="Hello"):`

Invalid:

`def greet(message="Hello", name):`

---

#### Default Values Are Evaluated Once

Default parameter values are created when the function is defined, not every time it is called.

This is especially important when using mutable objects such as lists or dictionaries as defaults.

##### Avoid

`def add_item(item, items=[]):`

Using a mutable default can cause values to remain between calls.

Prefer:

`def add_item(item, items=None):`

```
`if items is None:`

    `items = []`
```

---

#### Important Notes

* Default parameters are optional when calling the function.
* A provided argument replaces the default value.
* Required parameters must come before default parameters.
* Be careful with mutable default values such as `list` and `dict`.


````python
# ---------------------------------
# -- Function Default Parameters --
# ---------------------------------

def greet(name="Guest"):
    return name


greet()
greet("Mahmoud")


# Required + default parameter
def introduce(name, age=18):
    return name, age


introduce("Mahmoud")
introduce("Mahmoud", 20)


# Multiple default parameters
def create_user(name="Guest", age=18, country="Egypt"):
    return name, age, country


create_user()
create_user("Mahmoud")
create_user("Mahmoud", 20)
create_user("Mahmoud", 20, "Egypt")


# Safe pattern for mutable defaults
def add_item(item, items=None):
    if items is None:
        items = []

    items.append(item)
    return items


add_item("Python")
add_item("Java")
````

### 5.4 `*args`, `**kwargs` & Unpacking

Python allows a function to receive a variable number of arguments.

---

#### `*args`

`*args` allows a function to receive any number of **positional arguments**.

Inside the function, `args` is a **tuple**.

##### Syntax

`def function(*args):`

```
`# code`
```

Example:

`def total(*numbers):`

```
`return sum(numbers)`
```

`total(10, 20, 30)`

---

#### `**kwargs`

`**kwargs` allows a function to receive any number of **keyword arguments**.

Inside the function, `kwargs` is a **dictionary**.

##### Syntax

`def function(**kwargs):`

```
`# code`
```

Example:

`def create_user(**user_data):`

```
`return user_data`
```

---

#### `*args` + `**kwargs`

A function can accept both.

`def function(*args, **kwargs):`

* `args` => tuple of positional arguments
* `kwargs` => dictionary of keyword arguments

---

#### Packing

Packing means collecting multiple values into one variable.

##### With `*`

`def total(*numbers):`

The positional arguments are packed into the `numbers` tuple.

##### With `**`

`def show_info(**info):`

The keyword arguments are packed into the `info` dictionary.

---

#### Unpacking

Unpacking means taking the elements of an iterable and passing them as separate arguments.

##### `*` Unpacking

`numbers = [10, 20, 30]`

`add(*numbers)`

The list elements are passed as separate positional arguments.

##### `**` Unpacking

`user = {"name": "Mahmoud", "age": 20}`

`create_user(**user)`

The dictionary keys are matched with function parameter names.

---

#### Parameter Order

When using all common parameter forms, the order is:

1. Regular parameters
2. `*args`
3. Keyword-only parameters
4. `**kwargs`

Example:

`def function(a, b, *args, option=True, **kwargs):`

---

#### Important Notes

* `*args` collects positional arguments into a tuple.
* `**kwargs` collects keyword arguments into a dictionary.
* `*` can unpack an iterable.
* `**` can unpack a mapping such as a dictionary.
* The names `args` and `kwargs` are conventions, not keywords.


````python
# -------------------------------------------------
# -- *args / **kwargs / Packing / Unpacking --
# -------------------------------------------------

# -----------------
# -- *args --
# -----------------

def total(*numbers):
    return sum(numbers)


total(10)
total(10, 20)
total(10, 20, 30, 40)


# *args is a tuple
def show_args(*args):
    return args


show_args(10, 20, 30)


# -----------------
# -- **kwargs --
# -----------------

def show_user(**user_data):
    return user_data


show_user(name="Mahmoud", age=20)


# **kwargs is a dictionary
def show_kwargs(**kwargs):
    return kwargs


show_kwargs(name="Mahmoud", age=20)


# -----------------
# -- *args + **kwargs --
# -----------------

def function(*args, **kwargs):
    return args, kwargs


function(10, 20, name="Mahmoud", age=20)


# -----------------
# -- Unpacking with * --
# -----------------

def add(a, b, c):
    return a + b + c


numbers = [10, 20, 30]

add(*numbers)


# -----------------
# -- Unpacking with ** --
# -----------------

def create_user(name, age):
    return name, age


user = {
    "name": "Mahmoud",
    "age": 20
}

create_user(**user)


# -----------------
# -- Keyword-only --
# -----------------

def account(name, *, active=True):
    return name, active


account("Mahmoud")
account("Mahmoud", active=False)
````

### 5.5 Function Scope

Scope determines where a variable can be accessed.

A variable created inside a function normally belongs to that function's **local scope**.

---

#### Local Scope

A variable created inside a function is local to that function.

`def greet():`

```
`name = "Mahmoud"`
```

`name` cannot normally be accessed outside the function.

---

#### Global Scope

A variable created outside functions belongs to the global scope.

`name = "Mahmoud"`

Functions can read global variables.

---

#### Local vs Global

If a local variable has the same name as a global variable, the local variable takes priority inside the function.

---

#### `global`

The `global` keyword allows a function to modify a global variable.

Example:

`count = 0`

`def increase():`

```
`global count`

`count += 1`
```

Using global variables extensively can make code harder to understand, so local variables and return values are usually preferred.

---

#### LEGB Rule

Python searches for names using the LEGB order:

| Scope | Meaning   |
| ----- | --------- |
| `L`   | Local     |
| `E`   | Enclosing |
| `G`   | Global    |
| `B`   | Built-in  |

Python searches from the nearest scope outward.

---

#### Important Notes

* Function parameters are local variables.
* Variables created inside a function are normally local.
* Global variables can be read inside functions.
* `global` is required when assigning to a global variable from inside a function.
* A function creates its own local scope.
* `if`, `for`, and `while` do not create a separate local scope in Python.


````python
# --------------------
# -- Function Scope --
# --------------------

# Global variable
name = "Mahmoud"


def greet():
    # Local variable
    message = "Hello"

    # Can read global variable
    return message, name


greet()


# Local variable
def calculate():
    number = 10
    return number


result = calculate()


# Same name: local takes priority
name = "Mahmoud"


def show_name():
    name = "Ahmed"
    return name


show_name()


# global
count = 0


def increase():
    global count
    count += 1


increase()
increase()


# Enclosing scope
def outer():
    message = "Hello"

    def inner():
        return message

    return inner()


outer()
````

### 5.6 Return Statement

The `return` statement sends a value back from a function.

It also immediately ends the function execution.

#### Basic Syntax

`def add(a, b):`

```
`return a + b`
```

The returned value can be stored:

`result = add(10, 5)`

#### `return` vs `print()`

`print()` displays a value.

`return` sends a value back to the code that called the function.

##### `print()`

`def add(a, b):`

```
`print(a + b)`
```

The function displays the result but does not return it.

##### `return`

`def add(a, b):`

```
`return a + b`
```

The result can be stored, reused, or passed to another function.

---

#### `return` Ends the Function

Any code after an executed `return` in the same function will not run.

---

#### Multiple Return Values

Python allows a function to return multiple values.

`return name, age`

Python packs them into a tuple.

---

#### No `return`

If a function does not explicitly return a value, it returns `None`.

---

#### Conditional Return

A function can return different values depending on a condition.

---

#### Important Notes

* `return` sends data back to the caller.
* `return` immediately stops the current function execution.
* A function can have multiple `return` statements.
* Only the first executed `return` is reached.
* Multiple returned values are packed into a tuple.
* A function without an explicit return gives `None`.


````python
# --------------------
# -- Function Return --
# --------------------

# Returning a value
def add(a, b):
    return a + b


result = add(10, 5)


# print vs return
def add_with_print(a, b):
    print(a + b)


def add_with_return(a, b):
    return a + b


add_with_print(10, 5)

result = add_with_return(10, 5)


# return stops the function
def check_number(number):
    if number > 0:
        return "Positive"

    return "Zero or Negative"


check_number(10)


# Multiple return values
def get_user():
    name = "Mahmoud"
    age = 20

    return name, age


name, age = get_user()


# No explicit return
def say_hello():
    print("Hello")


result = say_hello()


# Conditional return
def get_grade(score):
    if score >= 90:
        return "A"
    elif score >= 80:
        return "B"
    elif score >= 70:
        return "C"

    return "F"
````

### 5.7 Recursion

Recursion happens when a function **calls itself**.

A recursive function solves a problem by breaking it into smaller versions of the same problem.

---

#### The Two Important Parts

A recursive function needs:

##### 1. Base Case

The condition that stops the recursion.

Without a base case, the function keeps calling itself until Python raises a `RecursionError`.

##### 2. Recursive Case

The part where the function calls itself with a smaller or simpler problem.

---

#### Basic Structure

`def function(value):`

```
`if stopping_condition:`

    `return result`

`return function(smaller_value)`
```

---

#### Example

Factorial:

`5! = 5 × 4 × 3 × 2 × 1`

The recursive idea is:

`factorial(n) = n × factorial(n - 1)`

The base case:

`factorial(0) = 1`

---

#### How Recursion Works

When a function calls itself, Python keeps track of each function call using the **call stack**.

Each call waits for the next call to finish.

Once the base case is reached, the calls return one by one.

---

#### Recursion vs Loop

| Recursion                                     | Loop                                         |
| --------------------------------------------- | -------------------------------------------- |
| Function calls itself                         | Repeats a block                              |
| Uses call stack                               | Usually simpler memory usage                 |
| Needs a base case                             | Needs a condition                            |
| Can make recursive problems easier to express | Often simpler for straightforward repetition |

---

#### When to Use Recursion

Recursion can be useful for problems that naturally contain smaller versions of themselves, such as:

* Tree structures
* Nested data
* Divide-and-conquer algorithms
* Graph traversal
* Mathematical recursive problems

For simple repetition, a loop is usually easier.

---

#### Important Notes

* A recursive function must have a base case.
* Each recursive call should move toward the base case.
* Too much recursion can cause `RecursionError`.
* Python has a recursion depth limit.
* Recursion is not automatically better than loops.


````python
# ------------------------
# -- Function Recursion --
# ------------------------

# Basic recursion
def countdown(number):
    if number == 0:
        return

    countdown(number - 1)


countdown(5)


# ------------------------
# -- Factorial --
# ------------------------

def factorial(number):
    if number == 0:
        return 1

    return number * factorial(number - 1)


factorial(5)


# ------------------------
# -- Recursive Sum --
# ------------------------

def recursive_sum(number):
    if number == 0:
        return 0

    return number + recursive_sum(number - 1)


recursive_sum(5)


# ------------------------
# -- Recursive Countdown --
# ------------------------

def countdown(number):
    if number == 0:
        return

    print(number)
    countdown(number - 1)


countdown(5)
````

### 5.8 Lambda Functions

A `lambda` is a small **anonymous function**.

It is useful when you need a short function for a simple operation.

---

#### Syntax

`lambda parameters: expression`

Example:

`lambda x: x * 2`

This creates a function that receives `x` and returns `x * 2`.

---

#### Lambda Has No Name

Unlike a normal function:

`def double(x):`

```
`return x * 2`
```

a lambda does not need a function name.

However, a lambda can be assigned to a variable:

`double = lambda x: x * 2`

In this case, `double` refers to the function.

---

#### Lambda Returns Automatically

The expression after `:` is automatically returned.

`lambda x: x * 2`

is similar to:

`def double(x):`

```
`return x * 2`
```

---

#### Lambda Is One Expression

A lambda can contain only **one expression**.

It cannot contain a normal block of statements such as multiple lines, loops, or regular `if` statements.

A conditional expression can still be used:

`lambda x: "Even" if x % 2 == 0 else "Odd"`

---

#### Calling a Lambda

`double = lambda x: x * 2`

`double(5)`

=> `10`

A lambda can also be called immediately:

`(lambda x: x * 2)(5)`

=> `10`

---

#### Lambda as an Argument

Lambdas are commonly passed to functions that accept another function.

Common examples:

* `sorted()`
* `map()`
* `filter()`

Example:

`sorted(names, key=lambda name: len(name))`

---

#### Lambda vs `def`

| Lambda                           | `def`                           |
| -------------------------------- | ------------------------------- |
| Anonymous function               | Named function                  |
| One expression                   | Can contain multiple statements |
| Automatically returns expression | Uses `return` explicitly        |
| Best for short/simple operations | Better for larger logic         |
| Commonly used inline             | Usually defined separately      |

---

#### Important Notes

* A lambda is an object of type `function`.
* Lambda functions are useful for short operations.
* Lambda can accept multiple parameters.
* Lambda can have default parameters.
* Lambda can use conditional expressions.
* Lambda is not a replacement for normal functions.
* Use `def` when the logic becomes large or needs to be reused clearly.


````python
# ------------------------
# -- Function => lambda --
# -- Anonymous Function --
# ------------------------

# Simple lambda
double = lambda x: x * 2

double(5)


# Multiple parameters
add = lambda a, b: a + b

add(10, 5)


# Lambda with a condition
check_number = lambda number: "Even" if number % 2 == 0 else "Odd"

check_number(10)


# Lambda can be called directly
(lambda x: x * 2)(5)


# Lambda with sorted()
names = ["Mahmoud", "Ali", "Ibrahim", "Mona"]

sorted(names, key=lambda name: len(name))


# Lambda with map()
numbers = [1, 2, 3, 4, 5]

list(map(lambda number: number * 2, numbers))


# Lambda with filter()
list(filter(lambda number: number % 2 == 0, numbers))


# Lambda type
type(lambda x: x * 2)
````

### 5.9 Decorators

<a id="sec-5-9-1"></a>

#### 5.9.1 Introduction

* A **decorator** is a function that takes another function, adds or changes its behavior, and returns a function.
* Decorators are based on Python's **first-class functions** concept: functions are objects and can be stored in variables, passed as arguments, and returned from other functions.
* A decorator is a **higher-order function** because it works with functions as arguments or return values.
* A decorator **wraps** another function to enhance its behavior without changing the original function's code.
* Decorators are commonly used for logging, authentication, validation, timing, caching, and access control.
* The `@decorator` syntax is shorthand for replacing a function with the decorated version.

##### Basic Structure

```text
decorator(function)
       =>
    wrapper
       =>
enhanced function
```

##### `@decorator` Syntax

These two forms are equivalent:

```text
@decorator
def greet():
    ...
```

```text
def greet():
    ...

greet = decorator(greet)
```

##### Important Idea

The decorator does **not** usually modify the original function directly.

Instead:

1. The original function is passed to the decorator.
2. The decorator creates a wrapper function.
3. The wrapper adds extra behavior.
4. The decorator returns the wrapper.
5. The original function name now refers to the decorated function.


````python
# Decorator: takes a function and returns an enhanced function
def my_decorator(func):
    def wrapper():
        print("Before the function")
        func()
        print("After the function")
    
    return wrapper


# @my_decorator: automatically applies my_decorator to greet
@my_decorator
def greet():
    print("Hello!")


greet()
````

<a id="sec-5-9-2"></a>

#### 5.9.2 Decorators With Parameters

When the decorated function has parameters, the wrapper must also be able to receive those arguments.

A common solution is:

* `*args` => receives positional arguments.
* `**kwargs` => receives keyword arguments.
* `func(*args, **kwargs)` => passes all received arguments to the original function.

This makes the decorator reusable with functions that have different parameter lists.

##### General Pattern

```text
def decorator(func):
    def wrapper(*args, **kwargs):
        # Extra behavior
        result = func(*args, **kwargs)
        # More behavior
        return result

    return wrapper
```

##### Why `*args` and `**kwargs`?

Suppose one function receives:

```text
greet("Mahmoud")
```

while another receives:

```text
add(10, 20)
```

and another receives:

```text
login(username="Mahmoud", role="admin")
```

Using `*args` and `**kwargs` allows the same decorator to work with all of them.


````python
# Decorator: accepts a function and supports any positional/keyword arguments
def my_decorator(func):
    def wrapper(*args, **kwargs):
        print("Before the function")

        result = func(*args, **kwargs)

        print("After the function")

        return result

    return wrapper


# Decorator applied to a function with parameters
@my_decorator
def greet(name):
    print(f"Hello, {name}!")


greet("Mahmoud")


# Same decorator works with multiple parameters
@my_decorator
def add(a, b):
    return a + b


result = add(10, 20)
print(result)
````

### 5.10 Docstrings

* A **docstring** is a string used to document a **module, class, function, or method**.
* A docstring is written as the **first statement** inside the object being documented.
* Python commonly uses triple quotes `"""..."""` for docstrings.
* Docstrings can be accessed through the `__doc__` attribute.
* They are also used by tools such as `help()`.
* Docstrings explain **what something does, how it should be used, and sometimes its parameters and return value**.
* Docstrings can be **one line** or **multiple lines**.
* A **comment** explains the code to developers and is ignored by Python.
* A **docstring** is actually a string object stored as documentation metadata.

#### Comment vs Docstring

| Comment                                         | Docstring                                            |
| ----------------------------------------------- | ---------------------------------------------------- |
| Starts with `#`                                 | Usually uses `"""..."""`                             |
| Explains code or implementation                 | Documents a module, class, function, or method       |
| Ignored by Python                               | Stored in `__doc__`                                  |
| Not available through `help()` as documentation | Used by `help()`                                     |
| Can appear almost anywhere                      | Must be the first statement of the documented object |

#### One-Line Docstring

```text
def greet(name):
    """Return a greeting for the given name."""
```

#### Multi-Line Docstring

```text
def calculate_average(numbers):
    """
    Calculate the average of a list of numbers.

    Parameters:
        numbers: A collection of numeric values.

    Returns:
        The average of the numbers.
    """
```

#### Important Difference

```text
### This is a comment.
```

The comment is not stored as the function's documentation.

```text
def greet():
    """Return a greeting message."""
```

The string becomes the function's docstring and can be accessed with:

```text
greet.__doc__
```

or:

```text
help(greet)
```


````python
# Docstring: documents the function and can be accessed through __doc__ and help()
def greet(name):
    """Return a greeting message for the given name."""
    
    # Comment: explains an implementation detail
    message = f"Hello, {name}!"
    
    return message


print(greet("Mahmoud"))

# Access the function's docstring
print(greet.__doc__)

# Display documentation
help(greet)
````

### 6.1 File Handling

File handling is used to create, read, write, and update files using Python.

Python commonly uses the `open()` function to work with files.

#### `open()`

##### Syntax

`open(file, mode)`

Example:

`file = open("data.txt", "r")`

The `mode` determines what Python is allowed to do with the file.

---

#### File Modes

| Mode  | Name   | Purpose                | File Must Exist? | Creates File? |
| ----- | ------ | ---------------------- | ---------------- | ------------- |
| `"r"` | Read   | Read file contents     | Yes              | No            |
| `"w"` | Write  | Write new content      | No               | Yes           |
| `"a"` | Append | Add content to the end | No               | Yes           |
| `"x"` | Create | Create a new file      | No               | Yes           |

##### `"r"` => Read

* Opens the file for reading.
* This is the default mode.
* Gives `FileNotFoundError` if the file does not exist.
* Does not allow writing.

##### `"w"` => Write

* Opens the file for writing.
* Creates the file if it does not exist.
* If the file already exists, its previous content is **overwritten**.

##### `"a"` => Append

* Opens the file for adding content.
* Creates the file if it does not exist.
* New content is added to the **end** of the file.
* Existing content is preserved.

##### `"x"` => Create

* Creates a new file.
* Gives `FileExistsError` if the file already exists.
* Useful when you want to make sure you do not overwrite an existing file.

---

#### Closing a File

When using `open()` directly, close the file when you finish working with it.

`file = open("data.txt", "r")`

`file.close()`

A better approach is using `with`.

---

#### `with open()`

The `with` statement automatically closes the file after the block finishes.

##### Syntax

`with open("data.txt", "r") as file:`

```
`# work with the file`
```

This is the recommended way to work with files.

---

#### Reading a File

##### `read()`

Reads the entire file.

`file.read()`

##### `readline()`

Reads one line.

`file.readline()`

##### `readlines()`

Reads all lines and returns them as a list.

`file.readlines()`

##### Iterating Through a File

A file can be used as an iterable.

`for line in file:`

```
`# process line`
```

---

#### Writing to a File

##### `write()`

Writes a string to the file.

`file.write("Hello Python")`

##### `writelines()`

Writes multiple strings from an iterable.

`file.writelines(["Hello\n", "Python\n"])`

`writelines()` does **not** automatically add new lines.

---

#### Important Notes

* `open()` returns a file object.
* `"r"` is the default mode.
* `"w"` can overwrite existing content.
* `"a"` preserves existing content and adds to the end.
* `"x"` fails if the file already exists.
* Use `with open(...)` when possible.
* Text files are usually handled as strings.
* `read()` returns a string.
* `readlines()` returns a list of lines.
* `write()` expects a string.
* Always consider the file encoding when working with text files.


````python
# -------------------
# -- File Handling --
# -------------------

# -------------------
# -- Write --
# -------------------

with open("data.txt", "w") as file:
    file.write("Hello Python\n")
    file.write("File Handling\n")


# -------------------
# -- Read --
# -------------------

with open("data.txt", "r") as file:
    content = file.read()


# -------------------
# -- Read Line --
# -------------------

with open("data.txt", "r") as file:
    first_line = file.readline()


# -------------------
# -- Read Lines --
# -------------------

with open("data.txt", "r") as file:
    lines = file.readlines()


# -------------------
# -- Iterate --
# -------------------

with open("data.txt", "r") as file:
    for line in file:
        pass


# -------------------
# -- Append --
# -------------------

with open("data.txt", "a") as file:
    file.write("New line\n")


# -------------------
# -- Create --
# -------------------

with open("new_file.txt", "x") as file:
    file.write("New file")


# -------------------
# -- writelines() --
# -------------------

lines = [
    "Python\n",
    "Java\n",
    "C++\n"
]

with open("languages.txt", "w") as file:
    file.writelines(lines)
````

### 6.2 Additional File Methods

Besides `read()`, `write()`, `readline()`, and `readlines()`, Python file objects provide additional methods for controlling and inspecting the file.

#### File Position

When Python reads or writes a file, it keeps track of a **current position** inside the file.

##### `tell()`

Returns the current position of the file cursor.

`tell()` returns the position as a number of bytes in binary mode and generally as a character position in text mode.

##### `seek()`

Moves the file cursor to a specific position.

`file.seek(position)`

Example:

`file.seek(0)` => move to the beginning of the file.

---

#### `truncate()`

Changes the size of the file.

`file.truncate(size)`

* If `size` is omitted, the file is truncated at the current cursor position.
* If `size` is specified, the file is truncated to that size.
* Content after the specified position is removed.
* Usually used when the file is opened in a mode that allows writing.

Example:

`file.truncate(10)` => keep only the first 10 units of file content.

---

#### File Object Attributes

File objects also provide useful information about the opened file.

| Attribute       | Description                    |
| --------------- | ------------------------------ |
| `file.name`     | File name                      |
| `file.mode`     | Mode used to open the file     |
| `file.closed`   | Whether the file is closed     |
| `file.encoding` | Text encoding used by the file |

---

#### Additional Useful Methods

| Method       | Purpose                               |
| ------------ | ------------------------------------- |
| `tell()`     | Return current cursor position        |
| `seek()`     | Move cursor to a position             |
| `truncate()` | Reduce file size                      |
| `flush()`    | Force buffered data to be written     |
| `fileno()`   | Return the underlying file descriptor |
| `readable()` | Check if the file can be read         |
| `writable()` | Check if the file can be written      |
| `seekable()` | Check if the cursor can be moved      |

---

#### `flush()`

Forces buffered data to be written to the underlying file.

`file.flush()`

Normally, you do not need to call this manually when using `with open(...)`, because Python handles closing the file automatically.

---

#### Checking File Capabilities

These methods return Boolean values.

`file.readable()` => can the file be read?

`file.writable()` => can the file be written?

`file.seekable()` => can the cursor be moved?

---

#### Important Notes

* `tell()` tells you where the cursor currently is.
* `seek()` changes the cursor position.
* `truncate()` changes the file size; it is different from simply moving the cursor.
* `flush()` writes buffered data.
* `readable()`, `writable()`, and `seekable()` return `True` or `False`.
* These methods work on the **file object** returned by `open()`.
* When using `with open(...)`, the file is automatically closed after the block.


````python
# ---------------------------------
# -- Additional File Methods --
# ---------------------------------

# Create a file
with open("data.txt", "w") as file:
    file.write("Hello Python File Handling")

# ---------------------------------
# -- tell() --
# ---------------------------------

with open("data.txt", "r") as file:
    position = file.tell()

    content = file.read()

    new_position = file.tell()


# ---------------------------------
# -- seek() --
# ---------------------------------

with open("data.txt", "r") as file:
    file.seek(0)
    content = file.read()

    file.seek(6)
    content = file.read()


# ---------------------------------
# -- tell() + seek() --
# ---------------------------------

with open("data.txt", "r") as file:
    file.read(5)
    position = file.tell()

    file.seek(0)
    beginning = file.read(5)


# ---------------------------------
# -- truncate() --
# ---------------------------------

with open("data.txt", "w") as file:
    file.write("Hello Python")

with open("data.txt", "r+") as file:
    file.truncate(5)


# ---------------------------------
# -- File Attributes --
# ---------------------------------

file = open("data.txt", "r")

file.name
file.mode
file.closed
file.encoding

file.close()

file.closed


# ---------------------------------
# -- File Capabilities --
# ---------------------------------

file = open("data.txt", "r")

file.readable()
file.writable()
file.seekable()

file.close()


# ---------------------------------
# -- flush() --
# ---------------------------------

file = open("data.txt", "w")

file.write("Hello Python")
file.flush()

file.close()
````

### 7.1 Errors & Exceptions

* An **exception** is an error that occurs while a program is running.
* Exceptions interrupt the normal flow of the program.
* Python provides an error message that helps you understand what went wrong.
* A **traceback** shows where the error occurred and points to the line that caused it.
* Exceptions have different types, such as:

  * `SyntaxError`
  * `NameError`
  * `TypeError`
  * `ValueError`
  * `IndexError`
  * `KeyError`
  * `ZeroDivisionError`
  * `FileNotFoundError`
* Python provides many built-in exception types.
* The `raise` keyword is used to raise an exception manually.

#### Common Exception Types

| Exception           | Usually happens when                         |
| ------------------- | -------------------------------------------- |
| `SyntaxError`       | Python code has invalid syntax               |
| `NameError`         | A name/variable does not exist               |
| `TypeError`         | An operation uses an inappropriate type      |
| `ValueError`        | The type is correct but the value is invalid |
| `IndexError`        | A sequence index does not exist              |
| `KeyError`          | A dictionary key does not exist              |
| `ZeroDivisionError` | Dividing by zero                             |
| `FileNotFoundError` | A requested file does not exist              |

#### `raise`

`raise` allows you to create an exception yourself when a condition is invalid.

```text
raise ValueError("Invalid value")
```

This immediately raises the exception and stops normal execution unless it is handled.


````python
# raise: manually raises an exception with a custom message
age = -5

# Wrapped in try/except only so the whole notebook can run end-to-end without
# stopping here - on its own, raise would simply propagate the exception up.
try:
    if age < 0:
        raise ValueError("Age cannot be negative")
except ValueError as e:
    print("Raised:", e)
````

<a id="sec-7-2"></a>

### 7.2 Exception Handling

Python provides four main keywords for handling exceptions:

| Keyword   | Purpose                                     |
| --------- | ------------------------------------------- |
| `try`     | Contains code that may cause an exception   |
| `except`  | Handles a matching exception                |
| `else`    | Runs only if no exception occurred in `try` |
| `finally` | Runs whether an exception occurred or not   |

#### Basic Structure

```text
try:
    # Code that may cause an exception

except SomeError:
    # Handle the exception

else:
    # Runs when no exception occurs

finally:
    # Runs in both cases
```

#### Important

* `try` must be followed by at least one `except` or `finally`.
* `else` is optional.
* `finally` is optional.
* `else` runs only when the `try` block finishes successfully.
* `finally` normally runs whether the code succeeds or fails.
* If an exception is not handled by any `except`, it continues upward and may terminate the program.


````python
# try: tests code that may raise an exception
try:
    # number = int(input("Enter a number: "))
    number = int("0")  # simulated input (triggers the ZeroDivisionError branch below)
    result = 10 / number

# except: handles a specific exception
except ZeroDivisionError:
    print("Cannot divide by zero.")

# else: runs only when no exception occurs
else:
    print("Result:", result)

# finally: runs whether an exception happened or not
finally:
    print("Finished.")
````

### 8.1 Regular Expressions Foundations

#### What Is a Regular Expression?

A **regular expression (regex)** is a pattern used to search, match, split, replace, and validate text.

Python provides regex support through the built-in `re` module.

```python
import re
```

#### Common Uses

- Find text that follows a pattern
- Validate input formats
- Extract parts of text
- Replace matching text
- Split text using a pattern
- Search large amounts of text efficiently

#### Important Mental Model

```text
Text
  =>
Pattern
  =>
Regex Engine
  =>
Match / Matches / No Match
```

#### Regex Is Pattern Matching, Not Full Parsing

Regex is excellent for structured text patterns, but it is not a universal parser.

For deeply nested or context-sensitive formats, use a dedicated parser when available.

#### Raw Strings

Regex patterns commonly use raw strings:

```python
r"\d+"
```

Raw strings reduce the need to double-escape backslashes.

Compare:

```python
"\\d+"
r"\d+"
```

Both represent a regex pattern containing `\d+`.

#### Two Levels of Escaping

You must think about:

1. Python string escaping
2. Regex escaping

Example:

```python
r"\."
```

means a literal dot in regex.

#### Main Functions

| Function | Main purpose |
|---|---|
| `re.search()` | Find the first match anywhere |
| `re.match()` | Match only at the start |
| `re.fullmatch()` | Match the entire string |
| `re.findall()` | Return all matches |
| `re.finditer()` | Iterate over `Match` objects |
| `re.split()` | Split using a regex |
| `re.sub()` | Replace matches |
| `re.subn()` | Replace and return replacement count |
| `re.compile()` | Create a reusable pattern |
| `re.escape()` | Escape characters for literal matching |

#### Source

Python documentation: https://docs.python.org/3/library/re.html

````python
import re

text = "Python 3.14 is powerful."

# re.search(pattern, string): find the first match anywhere
match = re.search(r"\d+", text)

print(match)
print(match.group() if match else None)

# re.fullmatch(pattern, string): entire string must match
print(re.fullmatch(r"\d+", "123"))
print(re.fullmatch(r"\d+", "123abc"))
````

<a id="sec-8-2"></a>

### 8.2 Regex Syntax, Character Classes & Quantifiers

#### Character Classes

| Pattern | Meaning |
|---|---|
| `.` | Any character except newline by default |
| `\d` | Unicode decimal digit |
| `\D` | Not a Unicode decimal digit |
| `\w` | Unicode word character |
| `\W` | Not a Unicode word character |
| `\s` | Unicode whitespace |
| `\S` | Not whitespace |
| `[abc]` | `a`, `b`, or `c` |
| `[^abc]` | Any character except `a`, `b`, `c` |
| `[a-z]` | Lowercase range |
| `[0-9]` | Digit range |

#### Quantifiers

| Pattern | Meaning |
|---|---|
| `*` | 0 or more |
| `+` | 1 or more |
| `?` | 0 or 1 |
| `{m}` | Exactly `m` |
| `{m,n}` | Between `m` and `n` |
| `{m,}` | At least `m` |

Quantifiers are normally **greedy**.

Add `?` after a quantifier for the lazy form:

```text
*?   +?   ??   {m,n}?
```

#### Anchors and Boundaries

| Pattern | Meaning |
|---|---|
| `^` | Start of string / line depending on flags |
| `$` | End of string / line depending on flags |
| `\A` | Start of string |
| `\Z` | End of string |
| `\b` | Word boundary |
| `\B` | Not a word boundary |

#### Important Unicode Note

In Python's default string regex mode, `\d`, `\w`, and `\s` follow Unicode-aware behavior.

Use `re.ASCII` when ASCII-only semantics are required.

#### Escaping Special Characters

Characters such as these have special regex meanings:

```text
. ^ $ * + ? { } [ ] \ | ( )
```

Use a backslash when you need their literal meaning:

```python
r"\."
r"\?"
r"\["
```

#### `re.escape()`

Useful when user-provided text should be matched literally instead of interpreted as regex syntax.

#### Source

https://docs.python.org/3/library/re.html

````python
import re

text = "Order 123, Order 456, Order 789."

# re.findall(pattern, string): return all matching strings
numbers = re.findall(r"\d+", text)

print(numbers)

# Character class
print(re.findall(r"[A-Z]", "Python Regex"))

# Quantifier
print(re.findall(r"\w{4}", "Python code test"))

# Boundary
print(re.findall(r"\bcat\b", "cat scatter category cat"))

# Literal user text
user_text = "a+b"
safe_pattern = re.escape(user_text)

print(re.fullmatch(safe_pattern, "a+b"))
````

### 8.3 Regex Groups, Alternation & Assertions

#### Capturing Groups

Parentheses create capturing groups:

```text
(...)
```

They let you extract part of a match.

#### Non-Capturing Groups

Use:

```text
(?:...)
```

when you need grouping but do not need to capture the text.

#### Named Groups

Use:

```text
(?P<name>...)
```

This gives the group a readable name.

#### Alternation

The `|` operator means OR:

```text
cat|dog
```

#### Backreferences

A backreference matches text captured earlier:

```text
(\w+)\s+\1
```

This can detect repeated words.

Named backreferences can use:

```text
(?P=name)
```

#### Lookahead

Positive:

```text
(?=...)
```

The following pattern must match, but it is not consumed.

Negative:

```text
(?!...)
```

The following pattern must not match.

#### Lookbehind

Positive:

```text
(?<=...)
```

The preceding pattern must match.

Negative:

```text
(?<!...)
```

The preceding pattern must not match.

Lookbehind patterns have important fixed-length restrictions in the standard `re` engine for many forms.

#### Match Object

Regex matching functions may return a `re.Match` object.

Useful methods:

| Method | Purpose |
|---|---|
| `group()` | Get matched text |
| `group(n)` | Get a numbered group |
| `groups()` | Get all captured groups |
| `groupdict()` | Get named groups as a dictionary |
| `start()` | Start index |
| `end()` | End index |
| `span()` | `(start, end)` |
| `string` | Original string |
| `re` | Compiled pattern |

#### Group Numbering

Groups are numbered from left to right:

```text
( first group )
( second group )
```

Nested groups still receive their own numbers.

#### Source

https://docs.python.org/3/library/re.html

````python
import re

text = "Ahmed: 25"

# re.search(pattern, string): return the first Match object
match = re.search(r"(?P<name>\w+):\s*(?P<age>\d+)", text)

if match:
    print(match.group())
    print(match.group("name"))
    print(match.group("age"))
    print(match.groupdict())
    print(match.span())

# Non-capturing group
print(re.findall(r"(?:Mr|Ms)\.?\s+\w+", "Mr Ahmed, Ms Mona"))

# Alternation
print(re.findall(r"cat|dog", "cat dog bird"))

# Lookahead
print(re.findall(r"\d+(?= USD)", "10 USD 20 EUR 30 USD"))

# Backreference
print(re.findall(r"\b(\w+)\s+\1\b", "hello hello world"))
````

<a id="sec-8-4"></a>

### 8.4 Regex Functions, Match Objects & Replacement

#### Search Functions

##### `re.search()`

Searches anywhere and returns the first `Match`, or `None`.

##### `re.match()`

Checks only at the beginning of the string.

##### `re.fullmatch()`

Requires the entire string to match.

#### Find Functions

##### `re.findall()`

Returns all matches.

Its result depends on capturing groups:

- No capturing groups => list of matched strings
- One capturing group => list of captured strings
- Multiple capturing groups => list of tuples

##### `re.finditer()`

Returns an iterator of `Match` objects.

Use it when you need:

- Match positions
- Groups
- More detailed match information
- Memory-friendly iteration over many results

#### Split

```python
re.split(pattern, string, maxsplit=0)
```

With capturing groups, captured delimiters are included in the result.

#### Replace

```python
re.sub(pattern, repl, string, count=0)
```

`repl` can be:

- A replacement string
- A function that receives a `Match` object

#### `re.subn()`

Returns:

```text
(new_string, replacement_count)
```

#### `count`

For `find`/`sub` related operations where supported, `count`/`maxsplit` can limit work.

#### Compiled Patterns

```python
pattern = re.compile(r"\d+")
```

Then:

```python
pattern.search(text)
pattern.findall(text)
pattern.finditer(text)
pattern.sub(...)
```

Compiled patterns are useful when the same regex is reused.

#### Pattern Object

A compiled pattern is a `re.Pattern`.

Useful attributes:

- `.pattern`
- `.flags`
- `.groups`
- `.groupindex`

#### Match Object

A successful match is a `re.Match`.

#### Source

https://docs.python.org/3/library/re.html

````python
import re

text = "ID: 123, ID: 456, ID: 789"

# findall: return all matched strings
print(re.findall(r"\d+", text))

# finditer: return an iterator of Match objects
for match in re.finditer(r"\d+", text):
    print(match.group(), match.span())

# split: split by a regex pattern
print(re.split(r",\s*", text))

# sub: replace matches
print(re.sub(r"\d+", "XXX", text))

# sub with a function: replacement depends on the Match
def double(match):
    return str(int(match.group()) * 2)

print(re.sub(r"\d+", double, text))

# compile: create a reusable regex Pattern
pattern = re.compile(r"\d+")

print(pattern.findall(text))
print(pattern.pattern)
print(pattern.groups)
````

### 8.5 Regex Flags, Performance & Best Practices

#### Common Flags

| Flag | Meaning |
|---|---|
| `re.I` / `re.IGNORECASE` | Case-insensitive matching |
| `re.M` / `re.MULTILINE` | `^` and `$` work per line |
| `re.S` / `re.DOTALL` | `.` also matches newline |
| `re.X` / `re.VERBOSE` | Allows readable multi-line regexes with comments |
| `re.A` / `re.ASCII` | ASCII-only behavior for certain shorthand classes |
| `re.DEBUG` | Display regex compilation details |

Flags can be combined with `|`:

```python
re.IGNORECASE | re.MULTILINE
```

#### Inline Flags

Flags can sometimes be written inside a pattern:

```text
(?i)
(?m)
(?s)
(?x)
```

#### `re.VERBOSE`

Useful for complex regexes:

- Break patterns into multiple lines
- Add comments
- Improve readability

Remember that spaces/comments are ignored according to verbose-mode rules, so literal spaces must be expressed carefully.

#### Bytes vs Strings

Regex patterns and target text should use compatible types:

```text
str pattern  <=>  str text
bytes pattern <=> bytes text
```

Do not mix `str` and `bytes`.

#### Performance

- Prefer a simple pattern when possible.
- Compile a pattern when it is reused or when the pattern object is useful.
- Avoid unnecessary nested ambiguous quantifiers.
- Be careful with patterns that can cause excessive backtracking.
- Use `re.fullmatch()` for full validation instead of manually combining anchors when that communicates intent more clearly.
- Avoid using regex when simpler string methods are clearer.

#### Regex vs String Methods

For simple fixed text:

```python
text.startswith(...)
text.endswith(...)
text.find(...)
text.replace(...)
text.split(...)
```

may be clearer than regex.

#### Validation

Regex can validate a format, but a regex match does not automatically prove that data is semantically valid.

Example:

```text
A date may match YYYY-MM-DD
without necessarily being a real calendar date.
```

#### Security

Be cautious when processing attacker-controlled patterns or inputs.

Poorly designed patterns can create severe performance problems through excessive backtracking.

#### Best Practices

- Use raw strings for regex patterns.
- Use named groups when they improve readability.
- Use non-capturing groups when capture is unnecessary.
- Keep complex patterns documented.
- Prefer simpler alternatives when regex is not needed.
- Test edge cases.
- Use `re.escape()` for literal user-controlled text.

#### Python 3.14 Note

The standard `re` behavior is continuously maintained with Python releases. Test regexes against the Python version used by your application.

In Python 3.14, one documented behavioral change is that `\B` can match the empty string when used as the entire pattern.

#### Source

https://docs.python.org/3/library/re.html
https://docs.python.org/3.14/whatsnew/3.14.html

````python
import re

text = """Python
PYTHON
python"""

# IGNORECASE: case-insensitive matching
print(re.findall(r"python", text, flags=re.IGNORECASE))

# MULTILINE: ^ and $ work with individual lines
print(re.findall(r"^python$", text, flags=re.IGNORECASE | re.MULTILINE))

# DOTALL: . can match newline
print(bool(re.search(r"Python.*python", text, flags=re.DOTALL | re.IGNORECASE)))

# VERBOSE: make a complex pattern readable
pattern = re.compile(
    r"""
    (?P<name>[A-Za-z]+)
    \s*:\s*
    (?P<age>\d+)
    """,
    re.VERBOSE
)

print(pattern.fullmatch("Ahmed: 25"))
````

### 9.1 OOP Foundations

#### What Is OOP?

- **OOP (Object Oriented Programming)** is a programming paradigm for organizing data and behavior into objects.
- A **class** is a blueprint that defines the structure and behavior of a type.
- An **object / instance** is a concrete object created from a class.
- **Attributes** represent data/state.
- **Methods** represent behavior.
- Python is a **multi-paradigm** language: procedural, object-oriented, functional, and more.
- Python's object model is dynamic: classes are objects too, and classes can be modified at runtime.

#### Core Mental Model

```text
Class
  =>
Blueprint / Type
  =>
Instances (Objects)
  =>
State + Behavior
```

Example:

```text
Car
  Attributes => model, color, price
  Methods    => start(), stop(), accelerate()
```

#### Important Terminology

| Term | Meaning |
|---|---|
| Class | Blueprint / type used to create objects |
| Instance | Object created from a class |
| Attribute | Data associated with an object or class |
| Method | Function defined inside a class |
| State | Current data of an object |
| Behavior | Actions an object can perform |
| Instantiation | Creating an instance from a class |
| Encapsulation | Bundling data and behavior, with controlled access |
| Inheritance | Creating a class from another class |
| Polymorphism | Same interface/operation, different implementations |
| Abstraction | Exposing essential behavior while hiding implementation details |
| Composition | Building objects from other objects |

#### Everything Is An Object

Almost every value you use in Python is an object:

```text
10       => int object
"hello"  => str object
[1, 2]   => list object
function => function object
Class    => class/type object
```

Use `type()` and `__class__` to inspect an object's type.

#### OOP Is Not Only About Inheritance

A strong Python OOP design also uses:

- Encapsulation
- Composition
- Properties
- Class and static methods
- Polymorphism / duck typing
- Abstract base classes
- Special (dunder) methods
- Dataclasses
- Enums
- Descriptors
- Multiple inheritance and MRO when appropriate

#### Python Naming

Class names normally use **PascalCase**:

```text
User
BankAccount
OrderItem
```

Methods and instance attributes normally use **snake_case**:

```text
get_balance()
account_number
```

#### Source

- Python Classes: https://docs.python.org/3/tutorial/classes.html

````python
# A simple class groups data and behavior together.
class Car:
    def __init__(self, model, color):
        self.model = model
        self.color = color

    def describe(self):
        return f"{self.color} {self.model}"


car = Car("Sedan", "Black")

print(car.describe())
print(type(car))
print(car.__class__)
````

### 9.2 Classes, Instances & `__init__`

#### Class Syntax

```python
class ClassName:
    ...
```

- `class` creates a new class.
- The class body executes when the class is created.
- A class creates a new type of object.
- You create an instance by calling the class:

```python
user = User(...)
```

#### `__init__`

`__init__` is the **initializer**, not the low-level object constructor.

- It runs automatically after an instance has been created.
- Its job is to initialize instance state.
- The first parameter is conventionally `self`.
- `__new__` is responsible for creating/returning the instance; `__init__` initializes it.

#### `self`

- `self` refers to the current instance.
- It is passed automatically when an instance method is called through an instance.
- `self` is a convention, not a reserved keyword.
- The first parameter of a normal instance method should represent the instance.

These are conceptually equivalent:

```python
user.say_hello()
User.say_hello(user)
```

#### Each Instance Is Separate

```text
User Class
   |
   +-- user_1
   +-- user_2
   +-- user_3
```

Each instance normally has its own instance attributes.

#### Constructor vs Initializer

| Concept | Role |
|---|---|
| `__new__` | Creates/returns an instance |
| `__init__` | Initializes the created instance |
| `Class(...)` | Triggers the instantiation process |

#### Inspecting Classes and Instances

Useful tools:

- `type(obj)`
- `obj.__class__`
- `vars(obj)`
- `dir(obj)`
- `Class.__dict__`
- `obj.__dict__` when the object has an instance dictionary

#### Calling an Instance Method Through the Class

```python
instance.method()
```

is effectively bound as:

```python
Class.method(instance)
```

This is why `self` exists.

#### Source

- Python Classes: https://docs.python.org/3/tutorial/classes.html
- Python Data Model: https://docs.python.org/3/reference/datamodel.html

````python
class Member:
    def __init__(self, first_name, last_name):
        self.first_name = first_name
        self.last_name = last_name

    def full_name(self):
        return f"{self.first_name} {self.last_name}"


member_one = Member("Ahmed", "Ali")
member_two = Member("Mona", "Mahmoud")

print(member_one.full_name())
print(Member.full_name(member_one))

print(member_one.__class__)
print(vars(member_one))
print(Member.__dict__.keys())
````

### 9.3 Attributes & Methods

#### Instance Attributes

Instance attributes belong to a specific object.

Usually created inside `__init__`:

```python
self.name = name
self.age = age
```

Two instances can have different values.

#### Class Attributes

Class attributes are defined in the class body:

```python
class User:
    users_count = 0
```

They belong to the class namespace and can be accessed through the class and, when not shadowed, through instances.

#### Attribute Lookup

For an expression such as:

```python
obj.name
```

Python performs attribute lookup through the object/class machinery. A simplified mental model:

```text
instance data
   =>
class data
   =>
base classes according to MRO
```

Descriptors can change the exact lookup behavior.

#### Shadowing

An instance attribute can shadow a class attribute with the same name.

```python
class User:
    role = "user"

u = User()
u.role = "admin"
```

Now `u.role` comes from the instance, while `User.role` remains `"user"`.

#### Instance Methods

Normal methods receive the instance as `self`.

They can:

- Read instance attributes
- Modify instance attributes
- Call other instance methods
- Access class attributes
- Call class methods / static methods

#### Method Chaining

A method can return `self` to support chaining:

```python
obj.set_name(...).set_age(...)
```

Use this only when it improves readability.

#### Attribute Tools

| Tool | Purpose |
|---|---|
| `getattr(obj, name)` | Get an attribute dynamically |
| `setattr(obj, name, value)` | Set dynamically |
| `hasattr(obj, name)` | Check whether lookup succeeds |
| `delattr(obj, name)` | Delete dynamically |
| `vars(obj)` | Get `obj.__dict__` when available |
| `dir(obj)` | Inspect accessible names |

#### Mutable Class Attributes

Avoid class-level mutable state when each instance should have its own data.

Bad:

```python
class User:
    tags = []
```

All instances can share the same list.

Prefer:

```python
class User:
    def __init__(self):
        self.tags = []
```

#### Source

- Python Classes: https://docs.python.org/3/tutorial/classes.html

````python
class User:
    users_count = 0

    def __init__(self, name):
        self.name = name
        self.tags = []
        User.users_count += 1

    def add_tag(self, tag):
        self.tags.append(tag)
        return self

    def describe(self):
        return f"{self.name}: {self.tags}"


user = User("Ahmed")
user.add_tag("python").add_tag("oop")

print(user.describe())
print(User.users_count)
print(getattr(user, "name"))
print(hasattr(user, "email"))
print(vars(user))
````

### 9.4 Encapsulation & Properties

#### Encapsulation

Encapsulation means keeping data and the operations that work with that data together, while controlling how the object is accessed.

Python does not provide Java-style enforced private fields. It relies heavily on conventions and language mechanisms.

#### Access Conventions

| Form | Meaning |
|---|---|
| `name` | Public API by convention |
| `_name` | Protected/internal convention; not enforced |
| `__name` | Private-style name mangling |

##### Single Underscore

```python
self._balance
```

This means:

> "This is intended for internal use."

Python does not prevent outside access.

##### Double Underscore

```python
self.__balance
```

Python applies **name mangling**, usually turning it into:

```text
_ClassName__balance
```

It is not absolute security or true private memory.

#### Properties

`@property` allows a method to be accessed like an attribute:

```python
user.age_in_days
```

instead of:

```python
user.age_in_days()
```

A property is useful when you want attribute-style syntax with computed or controlled access.

#### Property Getter / Setter / Deleter

```python
@property
def value(self):
    ...

@value.setter
def value(self, new_value):
    ...

@value.deleter
def value(self):
    ...
```

#### Getter and Setter Methods

Explicit methods such as:

```python
get_name()
set_name(...)
```

are valid, but properties often produce a cleaner Python API when the concept behaves like an attribute.

#### `@functools.cached_property`

- Computes a property the first time it is accessed.
- Stores the result as an instance attribute.
- Later accesses use the stored value until it is deleted.
- Useful for expensive, mostly stable per-instance computations.

#### Important

Encapsulation is about **API design and invariants**, not simply hiding variables.

Example invariant:

```text
age >= 0
balance >= 0
```

A setter or method can enforce such rules.

#### Source

- Descriptor Guide: https://docs.python.org/3/howto/descriptor.html
- `functools.cached_property`: https://docs.python.org/3/library/functools.html

````python
from functools import cached_property


class BankAccount:
    def __init__(self, owner, balance):
        self.owner = owner
        self._balance = balance

    @property
    def balance(self):
        return self._balance

    @balance.setter
    def balance(self, value):
        if value < 0:
            raise ValueError("Balance cannot be negative.")
        self._balance = value

    @cached_property
    def account_summary(self):
        print("Calculating summary...")
        return f"{self.owner}: ${self.balance}"


account = BankAccount("Ahmed", 100)

print(account.balance)

account.balance = 250
print(account.balance)

print(account.account_summary)
print(account.account_summary)
````

### 9.5 Class Methods, Static Methods & Alternative Constructors

#### Three Common Method Types

| Method | First Parameter | Works With |
|---|---|---|
| Instance method | `self` | A particular instance |
| Class method | `cls` | The class |
| Static method | No automatic `self`/`cls` | No instance/class state |

#### `@classmethod`

- Receives the class as `cls`.
- Can be called through the class or an instance.
- Useful for operations involving class-level state.
- Commonly used for **alternative constructors**.

Example idea:

```python
User.from_string("Ahmed,20")
```

#### `@staticmethod`

- Receives no automatic instance/class argument.
- Useful for a function that is logically related to a class but does not need object or class state.

#### Alternative Constructors

A class method can create instances using another input format:

```python
@classmethod
def from_string(cls, text):
    ...
    return cls(...)
```

Using `cls` rather than the class name keeps inheritance-friendly behavior.

#### Decorators Commonly Used With Classes

Important OOP-related decorators:

- `@property`
- `@classmethod`
- `@staticmethod`
- `@abstractmethod`
- `@dataclass`
- `@cached_property`
- `@functools.total_ordering`
- `@functools.singledispatchmethod`

#### Decorator Order Matters

When combining decorators, order can affect the resulting descriptor/behavior.

For abstract methods combined with `classmethod`, `staticmethod`, or `property`, `@abstractmethod` should be the innermost decorator.

#### Source

- Descriptor Guide: https://docs.python.org/3/howto/descriptor.html
- `abc`: https://docs.python.org/3/library/abc.html
- `functools`: https://docs.python.org/3/library/functools.html

````python
class User:
    users_count = 0

    def __init__(self, name, age):
        self.name = name
        self.age = age
        User.users_count += 1

    @classmethod
    def from_string(cls, text):
        # cls: the class receiving the call
        name, age = text.split(",")
        return cls(name, int(age))

    @classmethod
    def show_count(cls):
        return cls.users_count

    @staticmethod
    def is_adult(age):
        # No self or cls is needed
        return age >= 18


user = User.from_string("Ahmed,20")

print(user.name, user.age)
print(User.show_count())
print(User.is_adult(user.age))
print(user.is_adult(15))
````

### 9.6 Inheritance & `super()`

#### Inheritance

Inheritance allows a class to derive from another class.

```python
class Child(Parent):
    ...
```

- Parent is also called **base class** or **superclass**.
- Child is also called **derived class** or **subclass**.

A subclass can:

- Reuse inherited behavior
- Add new behavior
- Override inherited methods
- Extend initialization

#### Method Overriding

A subclass can define a method with the same name as a parent method.

The subclass implementation is then selected for normal method lookup.

#### `super()`

`super()` gives access to the next implementation in the method resolution order.

Common use:

```python
super().__init__(...)
```

Important:

> `super()` does not simply mean "call my parent class".

It follows the class's **MRO** and is designed to cooperate with inheritance, including multiple inheritance.

#### Do Not Hard-Code the Parent When Cooperative Inheritance Is Intended

Prefer:

```python
super().method()
```

over:

```python
Parent.method(self)
```

when you want the inheritance chain to cooperate.

#### `isinstance()` vs `issubclass()`

```python
isinstance(obj, Class)
issubclass(Child, Parent)
```

`isinstance()` checks an object.

`issubclass()` checks class relationships.

#### Inheritance vs Composition

Inheritance expresses an **is-a** relationship.

Composition often expresses a **has-a** relationship.

Use inheritance when the subtype relationship and substitutability make sense. Use composition when an object is built from or delegates to other objects.

#### Source

- Python Classes / Inheritance: https://docs.python.org/3/tutorial/classes.html
- Python Data Model: https://docs.python.org/3/reference/datamodel.html

````python
class Employee:
    def __init__(self, name):
        self.name = name

    def describe(self):
        return f"Employee: {self.name}"


class Developer(Employee):
    def __init__(self, name, language):
        super().__init__(name)
        self.language = language

    def describe(self):
        return f"Developer: {self.name} - {self.language}"


developer = Developer("Ahmed", "Python")

print(developer.describe())
print(isinstance(developer, Developer))
print(isinstance(developer, Employee))
print(issubclass(Developer, Employee))
````

<a id="sec-9-7"></a>

### 9.7 Multiple Inheritance, MRO & Mixins

#### Multiple Inheritance

Python allows a class to inherit from multiple base classes:

```python
class Derived(BaseOne, BaseTwo):
    ...
```

The class can receive attributes and methods through the inheritance graph.

#### Method Resolution Order (MRO)

MRO determines the order Python uses to search for attributes and methods.

Inspect it with:

```python
Class.mro()
Class.__mro__
```

Python uses a consistent linearization called **C3**.

#### Cooperative Multiple Inheritance

When classes use `super()` consistently, each class can participate in a cooperative chain.

A common pattern:

```python
class A:
    def __init__(self, **kwargs):
        super().__init__(**kwargs)
```

Each class handles its own part and passes remaining arguments onward.

#### Mixins

A **mixin** is a small class designed to provide a focused piece of behavior to another class.

Examples:

```text
JsonMixin
LoggingMixin
TimestampMixin
```

A mixin usually:

- Provides reusable behavior
- Is not intended to represent the main domain object by itself
- Has a narrow responsibility

#### Diamond Inheritance

A diamond can occur when multiple paths lead to the same base class.

```text
       Base
      /       Left   Right
      \    /
      Child
```

MRO and cooperative `super()` are what make such structures manageable.

#### Common Mistakes

- Calling a specific parent directly everywhere
- Ignoring MRO
- Creating deep or confusing inheritance trees
- Using multiple inheritance where composition would be clearer

#### Source

- Python Classes: https://docs.python.org/3/tutorial/classes.html
- Python Data Model: https://docs.python.org/3/reference/datamodel.html

````python
class Base:
    def __init__(self):
        print("Base")


class Left(Base):
    def __init__(self):
        print("Left")
        super().__init__()


class Right(Base):
    def __init__(self):
        print("Right")
        super().__init__()


class Child(Left, Right):
    def __init__(self):
        print("Child")
        super().__init__()


child = Child()

print(Child.mro())
print(Child.__mro__)
````

### 9.8 Polymorphism, Duck Typing & Composition

#### Polymorphism

Polymorphism means code can work through a common interface while different objects provide different implementations.

Example:

```python
animal.speak()
```

Different classes can implement `speak()` differently.

#### Duck Typing

Python often cares more about what an object **can do** than its exact class.

> "If it behaves like the required object, use it."

Example:

```python
def start_engine(vehicle):
    vehicle.start()
```

The function can work with any object that provides a compatible `start()` operation.

This is one reason Python OOP can be flexible without a large inheritance hierarchy.

#### Built-in Polymorphism

The language itself uses polymorphism constantly:

```python
len(list_obj)
len(string_obj)
len(dict_obj)
```

Each type implements the required behavior differently.

Operators are polymorphic too:

```python
1 + 2
"Py" + "thon"
```

#### Composition

Composition means one object contains or uses another object.

```text
Car
 |
 +-- Engine
 +-- Transmission
```

This is a **has-a** relationship.

Composition often reduces coupling compared with inheritance.

#### Delegation

An object can forward some work to another object:

```python
self.engine.start()
```

The outer object exposes a higher-level API while another object performs the specialized operation.

#### Association / Aggregation / Composition

These terms describe object relationships with different strengths:

| Relationship | Idea |
|---|---|
| Association | Objects know/use each other |
| Aggregation | Whole uses objects that can exist independently |
| Composition | Whole strongly owns/contains component objects |

In practical Python code, the boundaries are often conceptual rather than enforced by the language.

#### Source

- Python Classes: https://docs.python.org/3/tutorial/classes.html
- Python Data Model: https://docs.python.org/3/reference/datamodel.html

````python
class EmailNotification:
    def send(self, message):
        return f"Email: {message}"


class SMSNotification:
    def send(self, message):
        return f"SMS: {message}"


def notify(service, message):
    # Duck typing: only the required .send() behavior matters.
    return service.send(message)


print(notify(EmailNotification(), "Hello"))
print(notify(SMSNotification(), "Hello"))


class Engine:
    def start(self):
        return "Engine started"


class Car:
    def __init__(self, engine):
        self.engine = engine

    def start(self):
        # Composition + delegation
        return self.engine.start()


print(Car(Engine()).start())
````

### 9.9 Abstract Base Classes (ABCs)

#### Why Abstract Classes?

An abstract base class defines an expected interface for subclasses.

Use ABCs when you want to state:

> "Any concrete subclass must provide these operations."

Python's `abc` module provides this infrastructure.

#### `ABC`

The easiest modern form is:

```python
from abc import ABC, abstractmethod

class Shape(ABC):
    ...
```

#### `@abstractmethod`

```python
@abstractmethod
def area(self):
    ...
```

A concrete subclass must implement abstract methods before it can be instantiated.

#### `ABCMeta`

`ABC` uses `ABCMeta` as its metaclass.

You may define an ABC directly with:

```python
class Shape(metaclass=ABCMeta):
    ...
```

but inheriting from `ABC` is usually simpler.

#### Abstract Properties and Class/Static Methods

Abstract behavior can also be combined with:

- `@property`
- `@classmethod`
- `@staticmethod`

For these combinations, `@abstractmethod` should be the innermost decorator.

#### Abstract Methods Can Have an Implementation

An abstract method can contain code, and subclasses can call it with `super()`.

Being abstract means the subclass must override it before instantiation; it does not necessarily mean the method body must be empty.

#### ABC vs Duck Typing

- **ABC** => explicit interface contract enforced at instantiation time.
- **Duck typing** => behavior-based, usually without requiring inheritance.

Choose based on the design and API you need.

#### Source

- `abc` module: https://docs.python.org/3/library/abc.html

````python
from abc import ABC, abstractmethod


class Shape(ABC):
    @abstractmethod
    def area(self):
        # Interface requirement
        pass


class Rectangle(Shape):
    def __init__(self, width, height):
        self.width = width
        self.height = height

    def area(self):
        return self.width * self.height


shape = Rectangle(4, 5)

print(shape.area())
print(isinstance(shape, Shape))

# Shape() would raise TypeError because area() is abstract.
````

### 9.10 Dunder Methods & Operator Overloading

#### What Are Dunder Methods?

Dunder = **double underscore**.

Examples:

```text
__init__
__str__
__repr__
__len__
__eq__
```

These methods are special protocol hooks used by Python syntax and built-ins.

They are not merely ordinary methods you call manually.

#### Common Representation Methods

| Method | Used By |
|---|---|
| `__str__` | `str(obj)`, `print(obj)` |
| `__repr__` | `repr(obj)`, interactive/debug representation |
| `__format__` | `format(obj, spec)` |
| `__bytes__` | `bytes(obj)` |

A good `__repr__` should generally be useful for debugging.

#### Object Lifecycle

| Method | Role |
|---|---|
| `__new__` | Create/return instance |
| `__init__` | Initialize instance |
| `__del__` | Finalization hook; use with care |

`__del__` is not a substitute for deterministic resource management. Prefer context managers for resources.

#### Size / Truth / Containers

| Method | Protocol |
|---|---|
| `__len__` | `len(obj)` |
| `__bool__` | `bool(obj)` / truth testing |
| `__iter__` | `iter(obj)` / iteration |
| `__next__` | `next(obj)` |
| `__contains__` | `item in obj` |
| `__getitem__` | `obj[key]` |
| `__setitem__` | `obj[key] = value` |
| `__delitem__` | `del obj[key]` |
| `__call__` | `obj(...)` |

#### Comparison Methods

```text
__eq__  ==
__ne__  !=
__lt__  <
__le__  <=
__gt__  >
__ge__  >=
```

Special comparison methods may return `NotImplemented` when they do not support the other operand.

#### Arithmetic Operator Overloading

Common methods:

```text
__add__      +
__sub__      -
__mul__      *
__truediv__  /
__floordiv__ //
__mod__      %
__pow__      **
```

Reflected forms include:

```text
__radd__
__rsub__
__rmul__
...
```

In-place forms include:

```text
__iadd__
__isub__
__imul__
...
```

#### Hashing

`__hash__` is used by `hash()` and hash-based collections.

Important rule:

> Objects that compare equal must have the same hash.

If a class overrides `__eq__` without explicitly restoring compatible hashing, Python normally makes instances unhashable.

Do not make mutable, equality-based objects hashable in a way that allows the hash to change while stored in a set/dict.

#### Special Method Lookup

For implicit operations such as `len(obj)` and arithmetic, Python looks up special methods on the object's type, not by simply checking instance attributes.

#### Source

- Python Data Model: https://docs.python.org/3/reference/datamodel.html

````python
class Money:
    def __init__(self, amount):
        self.amount = amount

    def __str__(self):
        return f"${self.amount:.2f}"

    def __repr__(self):
        return f"Money({self.amount!r})"

    def __add__(self, other):
        if not isinstance(other, Money):
            return NotImplemented
        return Money(self.amount + other.amount)

    def __eq__(self, other):
        if not isinstance(other, Money):
            return NotImplemented
        return self.amount == other.amount


price1 = Money(10)
price2 = Money(15)
total = price1 + price2

print(price1)
print(repr(price1))
print(total)
print(price1 == Money(10))
````

### 9.11 Object Model, Descriptors, `__slots__` & `__new__`

#### `__new__`

`__new__(cls, ...)` is responsible for creating and returning a new instance.

It is useful for advanced object creation patterns, such as:

- Immutable built-in subclasses
- Singleton-like patterns
- Controlling instance creation

Normal application code usually only needs `__init__`.

#### Descriptors

A descriptor is an object implementing one or more of:

```text
__get__
__set__
__delete__
```

A descriptor assigned as a class attribute can control attribute access.

Descriptors are the machinery behind several Python features, including:

- Bound methods
- `property`
- `classmethod`
- `staticmethod`
- `cached_property`
- `super`
- Many framework APIs

#### `__set_name__`

A descriptor can implement:

```python
__set_name__(owner, name)
```

Python calls it when the descriptor is assigned as part of class creation.

This lets a descriptor know the class and attribute name it was assigned to.

#### `__getattribute__`

Called for attribute access:

```python
obj.name
```

It is very powerful and easy to misuse.

Usually prefer `__getattr__` for a fallback when an attribute was not found.

#### `__getattr__`

Called only when normal attribute lookup fails.

This makes it useful for dynamic fallback behavior.

#### `__setattr__` / `__delattr__`

These intercept setting and deleting attributes.

Be careful to avoid infinite recursion. `object.__setattr__` is often used internally.

#### `__slots__`

`__slots__` lets a class declare a fixed set of instance attributes and can remove the normal per-instance `__dict__` unless one is included.

Possible benefits:

- Lower memory overhead for many instances
- Restricting arbitrary new instance attributes
- Supporting some specialized object layouts

Trade-offs:

- More restrictions
- Multiple inheritance can make slot design more complicated
- Some libraries/features expect `__dict__`

Use it for a reason, not automatically.

#### `__class__` and `type(obj)`

Both can show the object's class:

```python
obj.__class__
type(obj)
```

`type(obj)` is generally clearer for ordinary type inspection.

#### Source

- Python Data Model: https://docs.python.org/3/reference/datamodel.html
- Descriptor Guide: https://docs.python.org/3/howto/descriptor.html

````python
class Positive:
    def __set_name__(self, owner, name):
        self.private_name = f"_{name}"

    def __get__(self, obj, owner=None):
        if obj is None:
            return self
        return getattr(obj, self.private_name)

    def __set__(self, obj, value):
        if value <= 0:
            raise ValueError("Value must be positive.")
        setattr(obj, self.private_name, value)


class Product:
    price = Positive()

    def __init__(self, price):
        self.price = price


item = Product(50)
print(item.price)


class Point:
    __slots__ = ("x", "y")

    def __init__(self, x, y):
        self.x = x
        self.y = y


point = Point(10, 20)
print(point.x, point.y)

# point.z = 30  # AttributeError: no slot for "z"
````

### 9.12 Dataclasses & Enums

#### Dataclasses

The `dataclasses` module reduces boilerplate for classes that mainly store data.

Use:

```python
from dataclasses import dataclass
```

and:

```python
@dataclass
class User:
    name: str
    age: int
```

Depending on configuration, `@dataclass` can generate methods such as:

- `__init__`
- `__repr__`
- `__eq__`

Other important options include:

- `frozen=True`
- `order=True`
- `slots=True`
- `kw_only=True`
- `match_args=False`
- `unsafe_hash=True` (use with care)
- `field()`
- `default_factory=...`

#### `field(default_factory=...)`

Use `default_factory` for per-instance mutable defaults:

```python
items: list = field(default_factory=list)
```

Do not use a shared mutable default for instance data.

#### Frozen Dataclasses

```python
@dataclass(frozen=True)
class Point:
    x: int
    y: int
```

Instances become effectively read-only through normal assignment.

This is useful for value-like immutable objects.

#### Slots in Dataclasses

Modern Python dataclasses can generate slots:

```python
@dataclass(slots=True)
class Point:
    x: int
    y: int
```

#### Enums

The `enum` module provides named symbolic values.

Common enum types:

- `Enum`
- `IntEnum`
- `StrEnum`
- `Flag`
- `IntFlag`

#### `auto()`

`auto()` lets the Enum machinery generate values according to the enum type's rules.

#### Why Use Enum?

Prefer an enum when a value must come from a fixed set of named states:

```text
PENDING
PROCESSING
DONE
FAILED
```

This is clearer than passing arbitrary strings everywhere.

#### Source

- Dataclasses: https://docs.python.org/3/library/dataclasses.html
- Enum: https://docs.python.org/3/library/enum.html

````python
from dataclasses import dataclass, field
from enum import StrEnum, auto


@dataclass(slots=True)
class User:
    name: str
    tags: list[str] = field(default_factory=list)

    def add_tag(self, tag: str):
        self.tags.append(tag)


class Status(StrEnum):
    PENDING = auto()
    PROCESSING = auto()
    DONE = auto()


user = User("Ahmed")
user.add_tag("python")

print(user)
print(user.tags)
print(Status.PENDING)
print(Status.PENDING.value)
````

### 9.13 OOP Design, Advanced Patterns & Best Practices

#### 1. Prefer Clear Responsibilities

A class should have a focused purpose.

Good:

```text
User        => user data/behavior
EmailSender => email delivery
Report      => report representation
```

Avoid giant "God classes" that do everything.

#### 2. Composition Over Inheritance

When both are possible, composition can be easier to change:

```text
Order
  =>
PaymentService
ShippingService
NotificationService
```

The order uses these objects instead of inheriting from them.

Inheritance is still useful when there is a genuine subtype relationship and shared contract.

#### 3. Program to an Interface

Code can depend on behavior:

```python
service.send(...)
```

instead of requiring one concrete implementation everywhere.

This works naturally with duck typing and can also be formalized with ABCs.

#### 4. SOLID Overview

| Principle | Idea |
|---|---|
| S - Single Responsibility | One main reason for a class to change |
| O - Open/Closed | Extend behavior without constantly modifying stable code |
| L - Liskov Substitution | Subtypes should remain usable where the base type is expected |
| I - Interface Segregation | Prefer small focused interfaces over large ones |
| D - Dependency Inversion | Depend on abstractions/behavior rather than concrete details |

SOLID is a design guideline, not a Python syntax rule.

#### 5. Avoid Deep Inheritance Trees

Deep inheritance can make behavior hard to trace.

Prefer simple hierarchies and composition when inheritance does not add real value.

#### 6. Keep Public APIs Small

Expose the operations users actually need.

Internal helpers can use `_` naming conventions.

#### 7. Use `@property` for Controlled Attribute Access

Properties are useful when:

- A value is derived
- Validation is needed
- You want to change the implementation without changing the public syntax

#### 8. Use Dataclasses for Data-Centric Objects

If most of a class is:

```text
fields + simple representation + equality
```

a dataclass may be clearer than handwritten boilerplate.

#### 9. Prefer Explicit Protocols

For containers, use Python's protocols through dunder methods:

```text
__len__
__iter__
__getitem__
__contains__
```

For callable objects:

```text
__call__
```

For context managers:

```text
__enter__
__exit__
```

#### 10. Context Managers

Objects can manage resources using:

```python
with resource:
    ...
```

Implement:

```python
__enter__()
__exit__(...)
```

For async resources:

```python
__aenter__()
__aexit__(...)
```

#### 11. `__call__`

An object with `__call__` can be used like a function:

```python
obj(...)
```

This is useful for stateful callable objects.

#### 12. Introspection

Useful OOP inspection tools:

```text
type()
isinstance()
issubclass()
dir()
vars()
getattr()
hasattr()
Class.__dict__
Class.mro()
Class.__mro__
obj.__class__
```

#### 13. Advanced Function Dispatch

`functools.singledispatchmethod` can provide type-based dispatch for methods.

Dispatch is based on the first non-`self` / non-`cls` argument.

Use it when type-specific method behavior is genuinely useful; it is not a replacement for normal polymorphism.

#### 14. OOP Checklist

Before creating a class, ask:

```text
Does this object represent a clear concept?
Does it own state?
Does it have meaningful behavior?
Should this be composition instead?
Is inheritance actually needed?
What should be public?
What should be internal?
Would a dataclass be clearer?
Does the object need special methods?
```

#### Source

- Python Classes: https://docs.python.org/3/tutorial/classes.html
- Data Model: https://docs.python.org/3/reference/datamodel.html
- Dataclasses: https://docs.python.org/3/library/dataclasses.html
- `functools`: https://docs.python.org/3/library/functools.html

````python
from functools import singledispatchmethod


class Formatter:
    @singledispatchmethod
    def format_value(self, value):
        return str(value)

    @format_value.register
    def _(self, value: int):
        return f"INT:{value}"

    @format_value.register
    def _(self, value: list):
        return f"LIST:{len(value)} items"


formatter = Formatter()

print(formatter.format_value("hello"))
print(formatter.format_value(10))
print(formatter.format_value([1, 2, 3]))
````

### 10.1 Type Hints Foundations

#### What Is Type Hinting?

**Type hints** are annotations that describe the expected types of values in Python code.

They improve:

- Readability
- Editor support
- Static analysis
- Refactoring
- API documentation
- Early detection of many type-related mistakes

#### Important

Python type hints generally do **not** enforce types at runtime by themselves.

```python
def add(a: int, b: int) -> int:
    return a + b
```

A type checker can analyze this annotation, but Python normally does not automatically reject every incorrect call at runtime.

#### Variable Annotations

```python
name: str = "Ahmed"
age: int = 25
```

A bare annotation is also possible:

```python
age: int
```

#### Function Annotations

```python
def greet(name: str) -> str:
    return f"Hello, {name}"
```

The syntax is:

```text
parameter: Type
return: Type
```

#### `None`

Use:

```python
def log(message: str) -> None:
    ...
```

to indicate that the function does not return a useful value.

#### Type Hints vs `isinstance()`

| Type hints | `isinstance()` |
|---|---|
| Static/documentation-oriented | Runtime check |
| Used by tools | Executed by Python |
| Does not enforce by itself | Actually checks a value |

#### Built-in Generic Syntax

Modern Python can use built-in collections directly:

```python
list[str]
dict[str, int]
tuple[int, str]
set[int]
```

This syntax is preferred in modern Python when compatible with your supported versions.

#### Type Checker

Type hints are interpreted by external/static analysis tools such as:

- mypy
- pyright
- IDE language servers

#### Source

https://docs.python.org/3/library/typing.html

````python
# Variable annotations
name: str = "Ahmed"
age: int = 25

# Function annotations
def greet(name: str) -> str:
    return f"Hello, {name}"

def log(message: str) -> None:
    print(message)

print(greet(name))
log("Type hints improve readability.")

# Type hints do not automatically enforce runtime types.
print(greet(123))  # Python itself does not reject this call just because of the annotation.
````

### 10.2 Collections, Unions & Type Aliases

#### Collection Types

Modern syntax:

```python
list[str]
dict[str, int]
set[int]
tuple[int, str]
tuple[int, ...]
```

`tuple[int, ...]` means a tuple containing any number of integers.

#### Nested Collections

```python
list[dict[str, int]]
dict[str, list[int]]
```

#### Union Types

Modern Python supports:

```python
int | str
```

This means either `int` or `str`.

It is the concise form of:

```python
Union[int, str]
```

#### Optional Values

A value that can be a type or `None` can be written:

```python
str | None
```

This replaces the older common style:

```python
Optional[str]
```

#### Type Aliases

Python 3.12+ provides the `type` statement:

```python
type UserId = int
```

This creates a type alias with explicit alias semantics.

Older/common syntax:

```python
UserId = int
```

For new code on modern Python, the `type` statement is the preferred explicit alias syntax.

#### `TypeAlias`

`typing.TypeAlias` still exists but is deprecated in favor of the `type` statement as of Python 3.12.

#### Readability

Type aliases are useful when a type expression becomes long or has domain meaning:

```python
type UserRecord = dict[str, int | str]
```

#### Source

https://docs.python.org/3/library/typing.html

````python
# Built-in generic types
scores: list[int] = [90, 85, 100]
users: dict[str, int] = {"Ahmed": 20, "Mona": 22}
coordinates: tuple[int, int] = (10, 20)

# Union with modern | syntax
def normalize(value: int | str) -> str:
    return str(value)

# Optional-style type
def find_user(user_id: int) -> str | None:
    return "Ahmed" if user_id == 1 else None

# Python 3.12+ type alias syntax:
#     type UserId = int
#     type UserRecord = dict[str, int | str]
# exec() keeps this cell runnable on Python 3.11, where the statement is a SyntaxError.
import sys

if sys.version_info >= (3, 12):
    exec("type UserId = int\ntype UserRecord = dict[str, int | str]")
else:
    UserId = int
    UserRecord = dict[str, int | str]

user_id: UserId = 1
user: UserRecord = {"id": 1, "name": "Ahmed"}

print(scores, users, coordinates)
print(normalize(10))
print(find_user(1))
````

### 10.3 TypeVar, Generics & Special Type Forms

#### `TypeVar`

`TypeVar` represents a type that stays consistent across related positions.

```python
T = TypeVar("T")

def first(items: list[T]) -> T:
    return items[0]
```

The type checker can infer that the returned value has the same type as the elements.

#### Constrained TypeVar

```python
T = TypeVar("T", int, float)
```

`T` must be one of the listed types.

#### Bounded TypeVar

```python
T = TypeVar("T", bound=BaseClass)
```

`T` can be a subtype of the bound.

#### Generic Classes

```python
T = TypeVar("T")

class Box(Generic[T]):
    ...
```

Modern Python also supports type parameter syntax:

```python
class Box[T]:
    ...
```

and:

```python
def first[T](items: list[T]) -> T:
    ...
```

#### `Literal`

Use `Literal` for a limited set of exact values:

```python
Literal["read", "write"]
```

Useful when values are fixed and meaningful.

#### `Final`

Marks a name as not intended to be reassigned.

```python
MAX_RETRIES: Final = 3
```

Type checkers can flag reassignment.

#### `ClassVar`

Marks an attribute as class-level rather than instance-level for typing purposes.

```python
class User:
    users_count: ClassVar[int] = 0
```

`ClassVar` is especially relevant when type checkers analyze classes and dataclasses.

#### `Required` / `NotRequired`

These are mainly used with `TypedDict` to describe which dictionary keys must or need not exist.

#### Source

https://docs.python.org/3/library/typing.html

````python
from typing import ClassVar, Final, Generic, Literal, TypeVar

T = TypeVar("T")

# TypeVar: input and return keep the same type
def first(items: list[T]) -> T:
    return items[0]

print(first([10, 20, 30]))
print(first(["a", "b"]))

# Generic class
class Box(Generic[T]):
    def __init__(self, value: T):
        self.value = value

box = Box("Python")
print(box.value)

# Literal: only specific values are valid for static checking
def set_mode(mode: Literal["read", "write"]) -> str:
    return mode.upper()

print(set_mode("read"))

# Final: type checkers can flag reassignment
MAX_RETRIES: Final[int] = 3

class User:
    # ClassVar: class-level attribute
    users_count: ClassVar[int] = 0
````

### 10.4 Protocols, TypedDict, NamedTuple & Callable

#### `Protocol`

A `Protocol` defines behavior structurally.

Instead of requiring inheritance, a type checker can accept an object that has the required members.

This matches Python's **duck typing** style.

```python
class SupportsClose(Protocol):
    def close(self) -> None:
        ...
```

A class does not have to inherit from `SupportsClose` to satisfy it for static type checking.

#### `runtime_checkable`

A protocol decorated with:

```python
@runtime_checkable
```

can be used with certain runtime checks such as:

```python
isinstance(obj, ProtocolType)
```

Only runtime-checkable protocols support these checks, and runtime checking is based on the protocol's runtime-visible members.

#### `TypedDict`

`TypedDict` describes the expected keys and value types of dictionary-like data.

```python
class UserData(TypedDict):
    name: str
    age: int
```

The object is still an ordinary `dict` at runtime.

#### `Required` and `NotRequired`

These can specify required and optional keys in a `TypedDict`.

#### `NamedTuple`

Provides tuple-like objects with named fields.

It is useful when tuple behavior is desired and fields should have names.

#### `Callable`

Use `Callable` to describe functions/callables:

```python
Callable[[int, int], int]
```

meaning:

```text
accepts int, int
returns int
```

For an arbitrary callable:

```python
Callable[..., ReturnType]
```

#### Why These Matter

| Type | Main purpose |
|---|---|
| `Protocol` | Structural interfaces |
| `TypedDict` | Typed dictionary-shaped data |
| `NamedTuple` | Named tuple values |
| `Callable` | Function/callable signatures |

#### Source

https://docs.python.org/3/library/typing.html

````python
from typing import Callable, NamedTuple, NotRequired, Protocol, TypedDict, runtime_checkable


class Closable(Protocol):
    def close(self) -> None:
        ...


class Resource:
    def close(self) -> None:
        print("Closed")


def shutdown(resource: Closable) -> None:
    resource.close()


shutdown(Resource())


class UserData(TypedDict):
    name: str
    age: int
    email: NotRequired[str]


user: UserData = {
    "name": "Ahmed",
    "age": 25,
}

class Point(NamedTuple):
    x: int
    y: int

point = Point(10, 20)


def apply_operation(
    operation: Callable[[int, int], int],
    a: int,
    b: int,
) -> int:
    return operation(a, b)


print(point)
print(apply_operation(lambda x, y: x + y, 10, 20))
````

### 10.5 Advanced Type Hints: Self, TypeGuard, overload, ParamSpec & Annotated

#### `Self`

`Self` represents the current enclosing class.

Useful when a method returns `self` and subclasses should preserve the more specific return type.

```python
class Builder:
    def set_name(self, name: str) -> Self:
        self.name = name
        return self
```

It is especially useful for fluent APIs and alternative constructors.

#### `TypeGuard`

A function returning `TypeGuard[T]` tells a type checker:

> If this function returns `True`, treat the value as type `T`.

Useful for custom type-narrowing helpers.

#### `TypeIs`

Modern typing also provides `TypeIs`, which can express predicates that narrow to an intersection-like compatible type in both branches.

Use it when the predicate truly establishes that relationship.

#### `overload`

`@overload` describes multiple call signatures for a function.

Important:

- Overloaded signatures are for static type checkers.
- Only the final implementation is executed at runtime.

#### `ParamSpec`

`ParamSpec` captures a callable's parameter specification.

Useful when writing decorators that preserve the original function's parameter types.

```python
P = ParamSpec("P")

def decorator(func: Callable[P, T]) -> Callable[P, T]:
    ...
```

#### `Concatenate`

`Concatenate` can describe decorators that add parameters to a callable's signature.

#### `Annotated`

`Annotated[T, metadata]` attaches metadata to a type annotation.

The metadata can be interpreted by frameworks/tools.

It does not automatically enforce validation.

#### `Any`

`Any` disables much of the static type checking for that value.

Use it carefully.

`Any` is not the same as `object`.

#### `object` vs `Any`

- `object` means the value can be any object, but type-safe code cannot assume specific operations without narrowing.
- `Any` allows essentially unchecked operations by type checkers.

#### `Never`

`Never` represents a type with no values.

Useful for functions that never return normally:

```python
def fail() -> Never:
    raise RuntimeError
```

#### `NoReturn`

`NoReturn` has the same type-system meaning as `Never`; `Never` is the modern spelling.

#### Source

https://docs.python.org/3/library/typing.html

````python
from functools import wraps
import sys
from typing import Annotated, Callable, Never, ParamSpec, Self, TypeGuard, TypeVar, overload

if sys.version_info >= (3, 13):
    from typing import TypeIs
else:
    from typing import TypeGuard as TypeIs  # typing.TypeIs needs Python 3.13+; TypeGuard is the older form

T = TypeVar("T")
P = ParamSpec("P")


class Builder:
    def __init__(self):
        self.name = ""

    def set_name(self, name: str) -> Self:
        self.name = name
        return self


def is_int(value: object) -> TypeGuard[int]:
    return isinstance(value, int)


def is_string(value: object) -> TypeIs[str]:
    return isinstance(value, str)


@overload
def stringify(value: int) -> str: ...

@overload
def stringify(value: str) -> str: ...

def stringify(value: int | str) -> str:
    return str(value)


def trace(func: Callable[P, T]) -> Callable[P, T]:
    @wraps(func)
    def wrapper(*args: P.args, **kwargs: P.kwargs) -> T:
        return func(*args, **kwargs)
    return wrapper


def fail(message: str) -> Never:
    raise RuntimeError(message)


Score = Annotated[int, "0..100"]


builder = Builder().set_name("Python")

value: object = 10
if is_int(value):
    print(value + 1)

print(stringify(100))
print(stringify("Python"))
print(builder.name)
````

### 10.6 Typing Tools, Runtime Introspection & Python 3.14 Notes

#### `cast()`

```python
cast(Type, value)
```

Tells a static type checker to treat a value as a specified type.

Important:

> `cast()` does not perform a runtime conversion or validation.

#### `assert_type()`

Used by type-checking tests to state the type you expect a checker to infer.

#### `reveal_type()`

A tool for type checkers to reveal their inferred type during analysis.

Its behavior is primarily for static analysis and is not a normal runtime debugging solution.

#### `TYPE_CHECKING`

```python
if TYPE_CHECKING:
    ...
```

This condition is `False` at runtime and `True` for static type checkers.

Useful for type-only imports and avoiding certain runtime import cycles.

#### `NewType`

`NewType` creates a distinct static type based on an existing type:

```python
UserId = NewType("UserId", int)
```

At runtime, the returned object behaves like the underlying value.

It provides stronger static separation, not runtime validation.

#### `get_type_hints()`

`typing.get_type_hints()` evaluates and returns resolved type hints for supported objects.

Be careful when evaluating annotations from untrusted sources.

#### `__annotations__`

Functions, classes, and modules expose annotations through `__annotations__`.

Do not assume every arbitrary callable has this attribute.

#### Python 3.14: Deferred Annotation Evaluation

Python 3.14 changed annotation evaluation semantics.

Annotations are now evaluated lazily by default under the new annotation model.

Python 3.14 also introduced the standard-library `annotationlib` module for inspecting annotations.

Useful API:

```python
from annotationlib import get_annotations
```

`annotationlib` supports different formats for inspecting annotations, including evaluated values, forward references, and strings.

#### `TypeAlias` Note

`TypeAlias` remains available, but the `type` statement is preferred for explicit type aliases in modern Python.

#### Best Practices

- Prefer precise types over `Any`.
- Use built-in generic syntax such as `list[str]` in modern Python.
- Use `|` for unions on modern Python.
- Keep annotations readable.
- Use `Protocol` when behavior matters more than inheritance.
- Use `TypedDict` for dictionary-shaped external/data structures.
- Use `Self` for fluent methods that return the same subclass type.
- Use `ParamSpec` when writing type-preserving decorators.
- Use `cast()` only when static analysis needs information that runtime logic already guarantees.
- Remember that type hints alone do not validate runtime data.

#### Compatibility

Choose syntax based on the oldest Python version your project supports.

Modern Python supports features that require newer versions, such as:

```text
list[str]        => Python 3.9+
X | Y            => Python 3.10+
type Alias = ... => Python 3.12+
generic syntax   => Python 3.12+
```
